```python
# --- Collect all bucket dictionaries into one list ---
all_dicts = [
    BLEED_PATHOLOGY,
    OTHER_PATHOLOGIES,
    INTRACRANIAL_ANATOMY,
    EXTRACRANIAL_ANATOMY,
    BONE,
    NOTICING,
    LINKERS,
    UNCERTAINTY,
    CEREBRAL_LOCATIONS,
    SUBREGIONAL_ANATOMY,
    MASS_EFFECT,
    HERNIATION_SUBTYPES,
    ACUITY,
    DENSITY,
    PROMINENCE,
    MAGNITUDE,
    MORPHOLOGY,
    NORMAL_STATE,
    MEASUREMENT,
    NEGATION,
    LATERALITY,
    IGNORE,
    REPORT_TITLES,
    PUNCTUATION,
    NUMERICAL
]


# --- Extract multi-word keys ---
multiword_phrases = []

for d in all_dicts:
    for tok in d.keys():
        if " " in tok or "-" in tok:        # contains a space or - (multi-word)
            multiword_phrases.append(tok)

patterns = [nlp.make_doc(text) for text in multiword_phrases]
matcher = PhraseMatcher(nlp.vocab, attr="LOWER")
matcher.add("TOKEN_BUCKET_PHRASES", patterns)

@Language.component("merge_phrases")
def merge_phrases(doc):
    matches = matcher(doc)
    spans = [doc[start:end] for _, start, end in matches]
    filtered = filter_spans(spans)
    with doc.retokenize() as retokenizer:
        for span in filtered:
            retokenizer.merge(span)
    return doc

nlp.add_pipe("merge_phrases", first=True)

# --- Force sentence splitting ONLY by punctuation ---
@Language.component("punct_only_sentences")
def punct_only_sentences(doc):

    for token in doc:
        token.is_sent_start = False

    for i, token in enumerate(doc):

        if i == 0:
            token.is_sent_start = True
            continue

        # normal punctuation
        if doc[i-1].text in {".", ";", ":"}:
            token.is_sent_start = True
            continue

        # implicit sentence break:
        # Capitalized word but NOT all caps
        if (
            token.text[0].isupper()
            and not token.text.isupper()
        ):
            token.is_sent_start = True

    return doc

# GUARANTEED to run after parser
nlp.add_pipe("punct_only_sentences", last=True)
```

### Master Mapping function

```python
def master_variable_mapping_function(
    bleed_trees
):

    final_variables = {}

    for tree_group in bleed_trees:

        for tree in tree_group:

            all_mapped_variables = []
            all_direct_variables = {}

            # -------------------------
            # SDH Tree
            # -------------------------

            if "pathology" in tree:

                sdh_children = []
                association_children = []

                for child in tree["children"]:

                    if (
                        isinstance(child, dict)
                        and "association" in child
                    ):

                        # -------------------------
                        # Pull SDH characteristics
                        # back into sdh_children
                        # -------------------------

                        for association in child["association"]:

                            if "density" in association:

                                sdh_children.extend(
                                    association["density"]
                                )

                            elif "acuity" in association:

                                sdh_children.extend(
                                    association["acuity"]
                                )

                            elif "morphology" in association:

                                sdh_children.extend(
                                    association["morphology"]
                                )

                            elif "thickness" in association:

                                sdh_children.append(
                                    {
                                        "thickness":
                                        association["thickness"]
                                    }
                                )

                            else:

                                association_children.append(
                                    [association]
                                )

                        continue

                    sdh_children.append(
                        child
                    )

                # -------------------------
                # SDH Characteristics
                # -------------------------

                (
                    sdh_mapped_variables,
                    sdh_direct_variables
                ) = SDH_characteristics(
                    sdh_children
                )

                # -------------------------
                # SDH Associations
                # -------------------------

                (
                    assoc_mapped_variables,
                    assoc_direct_variables
                ) = SDH_association_characteristics(
                    association_children
                )

                all_mapped_variables.extend(
                    sdh_mapped_variables
                )

                all_mapped_variables.extend(
                    assoc_mapped_variables
              )

                all_direct_variables.update(
                    sdh_direct_variables
                )

                all_direct_variables.update(
                    assoc_direct_variables
                )

            # -------------------------
            # Negation Tree
            # -------------------------

            elif "negation" in tree:

                (
                    mapped_variables,
                    direct_variables
                ) = negation_sorter(
                    tree["negation"]
                )

                all_mapped_variables.extend(
                    mapped_variables
                )

                all_direct_variables.update(
                    direct_variables
                )

            # -------------------------
            # Outcome Tree
            # -------------------------

            elif "outcome" in tree:

                (
                    mapped_variables,
                    direct_variables
                ) = outcome_function(
                    tree["outcome"]
                )

                all_mapped_variables.extend(
                    mapped_variables
                )

                all_direct_variables.update(
                    direct_variables
                )

            else:

                continue

            # -------------------------
            # Final Mapping
            # -------------------------

            final_array_function(
                all_mapped_variables,
                all_direct_variables,
                final_variables
            )

    return final_variables
```

### Association_mapping

```python
def SDH_association_characteristics(
    association_children,
):

    mapped_variables = []
    direct_variables = {}

    for group in association_children:

        for association in group:

            # -------------------------
            # Effects
            # -------------------------

            if "effect_name" in association:

                (
                    function_mapped_variables,
                    function_direct_variables
                ) = effect_name_sorter(
                    association
                )
                
            # -------------------------
            # Negation
            # -------------------------

            elif "negation" in association:

                for negation in association["negation"]:
            
                    (
                        function_mapped_variables,
                        function_direct_variables
                    ) = negation_sorter(
                        negation
                    )
            
                    mapped_variables.extend(
                        function_mapped_variables
                    )
            
                    direct_variables.update(
                        function_direct_variables
                    )
            
                continue

            elif "prominence" in association:

                (
                    function_mapped_variables,
                    function_direct_variables
                ) = prominence_sorter(
                    association
                )

            else:

                continue

            mapped_variables.extend(
                function_mapped_variables
            )

            direct_variables.update(
                function_direct_variables
            )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Effect name Sorter

```python
def effect_name_sorter(
    association
):

    mapped_variables = []
    direct_variables = {}

    effect_name = association[
        "effect_name"
    ][0].lower()

    laterality_map = {

        "left": "Left",
        "left-sided": "Left",
        "to the left": "Left",

        "right": "Right",
        "right-sided": "Right",
        "to the right": "Right",

        "bilateral": "Bilateral"

    }

    herniation_map = {

        "subfalcine": "herniation_subfalcine",
        "subfalcine herniation": "herniation_subfalcine",
        "uncal herniation": "herniation_uncal",
        "early uncal herniation": "herniation_uncal",
        "transtentorial herniation": "herniation_transtentorial",
        "transtentorial": "herniation_transtentorial",
        "tonsillar": "herniation_tonsillar",
        "tonsillar herniation": "herniation_tonsillar",
        
    }

    # -------------------------
    # Extract magnitude
    # -------------------------

    severity = "Yes NS"

    if "magnitude" in association:

        severity = magnitude_function(
            association["magnitude"]
        )

    # -------------------------
    # Extract location info
    # -------------------------

    location_buckets = set()

    location_laterality = None

    ventricle_type = None

    if "location" in association:

        for text, bucket in association["location"]:

            text = text.lower()

            location_buckets.add(
                bucket
            )

            if bucket == "laterality":

                location_laterality = (
                    laterality_map.get(
                        text
                    )
                )

            elif bucket == "ventricle":

                if text == "lateral ventricle":

                    ventricle_type = (
                        "lateral"
                    )

                elif text == "lateral ventricles":

                    ventricle_type = (
                        "lateral"
                    )

                    # -------------------------
                    # Bilateral override
                    # -------------------------

                    location_laterality = (
                        "Bilateral"
                    )

                else:

                    ventricle_type = (
                        "system"
                    )

    # -------------------------
    # Mass Effect
    # -------------------------

    if effect_name == "mass effect":

        mapped_variables.append(
            f"mass_effect_present_{severity}"
        )

    # -------------------------
    # Midline Structures
    # -------------------------
    
    if "intraparenchymal" in location_buckets:
    
        mapped_variables.append(
            "midline_shift_present_Yes"
        )
    
        if location_laterality:
    
            direct_variables[
                "midline_shift_direction"
            ] = location_laterality

    # Uncal Heriation
    
    if "uncus" in location_buckets:

        mapped_variables.append(
            f"herniation_uncal_{severity}"
        )

    # -------------------------
    # Midline Shift
    # -------------------------

    elif effect_name == "midline shift":

        mapped_variables.append(
            "midline_shift_present_Yes"
        )

        if "numeric_mass_effect" in association:

            shift_data = association[
                "numeric_mass_effect"
            ]

            value_text = shift_data[
                "value"
            ].lower()

            direction_text = shift_data.get(
                "direction"
            )

            if value_text.endswith(
                "mm"
            ):

                mm = float(
                    value_text.replace(
                        "mm",
                        ""
                    ).strip()
                )

                cm = mm / 10

            elif value_text.endswith(
                "cm"
            ):

                cm = float(
                    value_text.replace(
                        "cm",
                        ""
                    ).strip()
                )

                mm = cm * 10

            else:

                mm = None
                cm = None

            if mm is not None:

                direct_variables[
                    "midline_shift_mm"
                ] = mm

            if cm is not None:

                direct_variables[
                    "midline_shift_cm"
                ] = cm

            if direction_text:

                direction = (
                    laterality_map.get(
                        direction_text.lower(),
                        direction_text
                    )
                )

                direct_variables[
                    "midline_shift_direction"
                ] = direction

    # -------------------------
    # Herniation
    # ------------------------

    elif effect_name in herniation_map:

        mapped_variables.append(
            f"{herniation_map[effect_name]}_{severity}"
        )

    # -------------------------
    # Sulcal Effacement
    # -------------------------

    if "sulci" in location_buckets:

        mapped_variables.append(
            f"effacement_sulci_{severity}"
        )

        if location_laterality:

            mapped_variables.append(
                f"effacement_sulci_laterality_{location_laterality}"
            )

    # -------------------------
    # Ventricular Effects
    # -------------------------

    if ventricle_type == "lateral":

        mapped_variables.append(
            f"effacement_lateral_ventricle_{severity}"
        )

        if location_laterality:

            mapped_variables.append(
                f"l_ventricle_laterality_{location_laterality}"
            )

    elif ventricle_type == "system":

        mapped_variables.append(
            f"effacement_ventricular_system_{severity}"
        )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Negation

```python
def negation_sorter(
    negation
):

    # -------------------------
    # Negated pathology
    # -------------------------

    if "pathology" in negation:

        mapped_variables, direct_variables = (
            pathology_function(
                negation["pathology"],
                negation.get(
                    "modifiers",
                    []
                ),
                negation.get(
                    "anatomy",
                    []
                )
            )
        )

    # -------------------------
    # Negated effects
    # -------------------------

    elif "effect_name" in negation:

        mapped_variables, direct_variables = (
            effect_name_sorter(
                negation
            )
        )

    # -------------------------
    # Negated outcomes
    # -------------------------

    elif "outcome" in negation:

        mapped_variables, direct_variables = (
            outcome_function(
                negation
            )
        )

    else:

        return (
            [],
            {}
        )

    # -------------------------
    # Negate mapped variables
    # -------------------------
    mapped_variables = [
        f"N_{variable}"
        for variable in mapped_variables

    ]

    return (
        mapped_variables,
        direct_variables
    )
```

#### Pathology

```python
def pathology_function(
    pathology_list,
    modifier_tokens=None,
    anatomy_tokens=None
):

    pathology_text = []

    for text, bucket in pathology_list:

        pathology_text.append(
            text
        )

    mapped_variables = pathology_filter(
        pathology_text
    )

    # -------------------------
    # Anatomy-based mappings
    # -------------------------

    if anatomy_tokens:

        anatomy_text = []

        for text, bucket in anatomy_tokens:

            anatomy_text.append(
                text
            )

        anatomy_variables = (
            pathology_anatomy_filter(
                anatomy_text
            )
        )

        mapped_variables.extend(
            anatomy_variables
        )

    # -------------------------
    # Generic abnormality
    # -------------------------
    
    if "generic_abnormality" in mapped_variables:
    
        mapped_variables.remove(
            "generic_abnormality"
        )

    # -------------------------
    # Add severity for selected
    # pathology variables
    # -------------------------

    if modifier_tokens:

        severity = magnitude_function(
            modifier_tokens
        )

        severity_variables = {

            "cerebral_atrophy",
            "microangiopathy"

        }

        updated_variables = []

        for variable in mapped_variables:

            if variable in severity_variables:

                updated_variables.append(
                    f"{variable}_{severity}"
                )

            else:

                updated_variables.append(
                    variable
                )

        mapped_variables = (
            updated_variables
        )

    direct_variables = {}

    return (
        list(set(mapped_variables)),
        direct_variables
    )
```

#### Outcome 

```python
def outcome_function(
    outcome
):

    # -------------------------
    # Outcome pathology
    # -------------------------

    if "pathology" in outcome:

        result = pathology_function(
            outcome["pathology"],
            outcome.get(
                "modifiers",
                []
            ),
            outcome.get(
                "anatomy",
                []
            )
        )

        return result

    # -------------------------
    # Outcome normal state
    # -------------------------

    elif "normal_state" in outcome:

        result = normal_state_function(
            outcome
        )

        return result

    return (
        [],
        {}
    )
```

#### Magnitude Mapper

```python
def magnitude_function(
    modifier_tokens
):

    magnitude_map = {

        "mild": "Mild",
        "partical": "Mild",
        "moderate": "Moderate",
        "severe": "Severe",
        "marked": "Severe",
        "significant": "Severe",
        "diffuse": "Severe",

        "generalized": "Yes NS"

    }

    severity = "Yes NS"

    for text, bucket in modifier_tokens:

        if bucket != "magnitude":
            continue

        severity = magnitude_map.get(
            text.lower(),
            severity
        )

    return severity
```

### SDH Characteristics

```python
def SDH_characteristics(children):

    # -------------------------
    # Count laterality mentions
    # -------------------------

    laterality_count = 0

    for child in children:

        if isinstance(child, tuple):

            text, bucket = child

            if bucket == "laterality":
                laterality_count += 1

        elif isinstance(child, dict) and "location" in child:

            for text, bucket in child["location"]:

                if bucket == "laterality":
                    laterality_count += 1

    # -------------------------
    # Bilateral branch
    # -------------------------

    # if laterality_count > 1:

        # return bilateral_SDH_characteristics(
            # children
        # )

    # -------------------------
    # Unilateral logic
    # -------------------------

    cerebral_locations = []
    density_values = []
    acuity_values = []
    magnitude_values = []
    thickness_values = []
    morphology_values = []

    laterality = None

    mapped_variables = []
    direct_variables = {}

    for child in children:

        # -------------------------
        # SDH Negation branch
        # -------------------------

        if (
            isinstance(child, dict)
            and "negation" in child
        ):

            for negation in child["negation"]:

                negation_variables, negation_direct = (
                    negation_sorter(
                        negation
                    )
                )

                mapped_variables.extend(
                    negation_variables
                )

                direct_variables.update(
                    negation_direct
                )

            continue

        # -------------------------
        # Direct tuples
        # -------------------------

        if isinstance(child, tuple):

            text, bucket = child

            if bucket == "laterality":

                if "left" in text.lower():
                    laterality = "L"

                elif "right" in text.lower():
                    laterality = "R"

            elif bucket == "cerebral_location":

                cerebral_locations.append(
                    text
                )

            elif bucket == "density":

                density_values.append(
                    text
                )

            elif bucket == "acuity":

                acuity_values.append(
                    text
                )

            elif bucket == "magnitude":

                magnitude_values.append(
                    text
                )

            elif bucket == "morphology":
            
                morphology_values.append(
                    text
                )

        # -------------------------
        # Association branch
        # -------------------------

        elif (
            isinstance(child, dict)
            and "association" in child
        ):

            for association in child["association"]:

                if "density" in association:

                    for text, bucket in association["density"]:

                        density_values.append(
                            text
                        )

                if "acuity" in association:

                    for text, bucket in association["acuity"]:

                        acuity_values.append(
                            text
                        )

                if "morphology" in association:

                    for text, bucket in association["morphology"]:

                        morphology_values.append(
                            text
                        )

                if "thickness" in association:

                    thickness_values.append(
                        association["thickness"]
                    )

        # -------------------------
        # Thickness branch
        # -------------------------

        elif (
            isinstance(child, dict)
            and "thickness" in child
        ):

            thickness_values.append(
                child["thickness"]
            )

        # -------------------------
        # Location branch
        # -------------------------

        elif (
            isinstance(child, dict)
            and "location" in child
        ):

            for text, bucket in child["location"]:

                if bucket == "laterality":

                    if "left" in text.lower():
                        laterality = "L"

                    elif "right" in text.lower():
                        laterality = "R"

                elif bucket == "cerebral_location":

                    cerebral_locations.append(
                        text
                    )


    # -------------------------
    # Run filters
    # -------------------------

    cerebral_variables = (
        cerebral_location_filter(
            cerebral_locations
        )
    )

    density_variables = (
        density_filter(
            density_values
        )
    )

    acuity_variables = (
        acuity_filter(
            acuity_values
        )
    )

    magnitude_variables = (
        magnitude_filter(
            magnitude_values
        )
    )

    morphology_variables = (
        morphology_filter(
            morphology_values
        )
    )


    # -------------------------
    # Combine outputs
    # -------------------------

    sdh_variables = []

    sdh_variables.extend(
        cerebral_variables
    )

    sdh_variables.extend(
        density_variables
    )

    sdh_variables.extend(
        acuity_variables
    )

    sdh_variables.extend(
        magnitude_variables
    )

    sdh_variables.extend(
        morphology_variables
    )

    # -------------------------
    # Apply side prefix
    # -------------------------

    for variable in sdh_variables:

        if laterality:

            mapped_variables.append(
                f"{laterality}_{variable}"
            )

        else:

            mapped_variables.append(
                variable
            )

    # -------------------------
    # Thickness
    # -------------------------

    for thickness_dict in thickness_values:

        thickness_value = next(
            iter(thickness_dict.keys())
        )

        thickness_type = next(
            iter(thickness_dict.values())
        )

        if thickness_value.endswith("cm"):

            cm = float(
                thickness_value.replace(
                    "cm",
                    ""
                )
            )

            mm = cm * 10

        elif thickness_value.endswith("mm"):

            mm = float(
                thickness_value.replace(
                    "mm",
                    ""
                )
            )

            cm = mm / 10

        else:

            continue

        if laterality:

            direct_variables[
                f"{laterality}_thickness_mm"
            ] = mm

            direct_variables[
                f"{laterality}_thickness_cm"
            ] = cm

            if thickness_type == "thickness_max":

                direct_variables[
                    f"{laterality}_thickness_max"
                ] = "Yes"
    print(mapped_variables)
    return (
        mapped_variables,
        direct_variables
    )
```

## BLEED TREE FUNCTION

```python
def build_bleed_tree(sent_tokens, is_bleed_sentence, existing_tree=None):

    # ============================
    # Initialize / reuse tree
    # ============================
    if existing_tree is not None:
        tree = existing_tree
    else:
        tree = {"pathology": None, "children": []}

        for text, bucket in sent_tokens:
            if bucket == "bleed_pathology":
                tree["pathology"] = text
                break

        if not tree["pathology"]:
            return None

    # ============================
    # Secondary sentence: run association ONCE at start
    # ============================
    association_done = False

    i = 0
    while i < len(sent_tokens):

        text, bucket = sent_tokens[i]

        # -----------------------------------
        # SECONDARY SENTENCE → association first
        # -----------------------------------
        if not is_bleed_sentence and not association_done:
            assoc, next_i = parse_association(sent_tokens, i)
                  
            if assoc:
                tree["children"].append({"association": assoc})
        
            association_done = True
        
            # DO NOT JUMP
            i += 1
            continue

        # -----------------------------------
        # PRIMARY SENTENCE LOGIC
        # -----------------------------------

        if bucket == "negation":

            branch, next_i = parse_negation_branch(
                sent_tokens, 
                i
            )
                      
            tree["children"].append(
                {"negation": [branch]}
            )

            i= next_i            
            continue
            
        # -----------------------------------
        # ASSOCIATION (primary bleed sentence)
        # -----------------------------------
        if is_bleed_sentence and bucket == "association_linker":

            next_bucket = sent_tokens[i+1][1] if i+1 < len(sent_tokens) else None

            # Open association ONLY if next bucket is magnitude or mass effect
            if next_bucket in ("magnitude",) or next_bucket in EFFECT_FAMILY:
                assoc, next_i = parse_association(sent_tokens, i)
                if assoc:
                    tree["children"].append({"association": assoc})
                i = next_i
                continue

            # Otherwise → SKIP association_linker
            i += 1
            continue

        # -----------------------------------
        # DIRECT ATTACH (after association logic)
        # -----------------------------------
        if is_bleed_sentence and bucket in DIRECT_ATTACH:
            tree["children"].append((text, bucket))
            i += 1
            continue

        # -----------------------------------
        # THICKNESS
        # -----------------------------------
        
        if bucket in (
            "numerical",
            "measurement_linker",
            "measurement_linker_max",
            "measurement_max",
            "measurement_type"
        ):
        
            local_tokens = sent_tokens[
                max(0, i - 5):
                min(len(sent_tokens), i + 6)
            ]
        
            has_effect = any(
                bucket in EFFECT_FAMILY
                for _, bucket in local_tokens
            )
        
            if not has_effect:

        
                result, next_i = parse_thickness_branch(
                    sent_tokens,
                    i
                )
        
                if result:
        
                    tree["children"].append(
                        {
                            "thickness": result
                        }
                    )
        
                i = next_i
                continue

        # -----------------------------------
        # LOCATION BRANCH
        # -----------------------------------
        if bucket == "location_linker":
            branch, next_i = parse_location_branch(sent_tokens, i)
            if branch:
                tree["children"].append({"location": branch})
            i = next_i
            continue

        # -----------------------------------
        # Direct Mass Effect Opener
        # -----------------------------------

        if (

            is_bleed_sentence
            and bucket in EFFECT_FAMILY

        ):

            assoc, next_i = parse_association(
                sent_tokens,
                i
            )

            if assoc:

                tree["children"].append(
                    {"association": assoc}
                )

            i = next_i
            continue

        # -----------------------------------
        # CONTINUE ATTACH
        # -----------------------------------
        if bucket in CONTINUE_ATTACH:
            i += 1
            continue

        # -----------------------------------
        # DEFAULT
        # -----------------------------------
        i += 1

    return tree
```

```python
def get_measurement_branch(
    tokens,
    start_i
):

    branch = []
    i = start_i

    while i < len(tokens):

        _, bucket = tokens[i]

        if bucket in (
            "association_linker",
            "stop_and",
            "period",
            "comma",
            "semicolon",
        ):
            break

        branch.append(
            tokens[i]
        )

        i += 1

    return branch
```

### AND function

```python
def classify_and(sent_tokens, idx):

    if idx <= 0 or idx >= len(sent_tokens) - 1:
        return "unknown_and"

    left_text, left_bucket = sent_tokens[idx - 1]
    right_text, right_bucket = sent_tokens[idx + 1]

    # -----------------------------
    # 1. Paired measurements
    # -----------------------------
    if "respectively" in right_text.lower():
        return "paired_and"

    # -----------------------------
    # 1.5 Directional laterality
    # -----------------------------
    if right_text.lower() in (
        "rightward",
        "leftward"
    ):
        return "stop_and"

    # -----------------------------
    # 2. Anatomy families → sandwich_and
    # -----------------------------
    ANATOMY_FAMILY = {
        "intraparenchymal",
        "ventricle",
        "cistern",
        "cerebral_location",
        "extracranial_anatomy",
        "subregional_anatomy",
        "laterality",
        "sulci",
        "sinus",
        "bone",
    }

    if left_bucket in ANATOMY_FAMILY and right_bucket in ANATOMY_FAMILY:
        return "sandwich_and"

    # -----------------------------
    # 3. Pathology families → stop_and
    # -----------------------------
    PATHOLOGY_FAMILY = {
        "bleed_pathology",
        "mass_effect",
        "effacement",
        "compression",
        "midline_shift",
        "pathology",
        "bone_pathology",
    }

    if left_bucket in PATHOLOGY_FAMILY and right_bucket in PATHOLOGY_FAMILY:
        return "stop_and"

    # -----------------------------
    # 4. Mixed categories → stop_and
    # -----------------------------
    if left_bucket != right_bucket:
        return "stop_and"

    # -----------------------------
    # 5. Same bucket → sandwich_and
    # -----------------------------
    return "sandwich_and"
```

### Thickness Parser

```python
def parse_thickness_branch(
    tokens,
    start_i
):

    has_thickness_context = any(
        bucket in (
            "measurement_linker_max",
            "measurement_max",
            "measurement_type",
        )
        for _, bucket in tokens
    )

    if not has_thickness_context:

        return None, start_i+1

    i = start_i
    number = None
    unit = None
    is_max = False

    STOP_BUCKETS = {
        "mass_effect",
        "herniation_subtype",
        "association_linker",
        "comma",
        "period",
        "semicolon",
        "colon"
    }

    while i < len(tokens):
        text, bucket = tokens[i]

        if bucket in STOP_BUCKETS:
            break
        
        if bucket == "noticing":
            i += 1
            continue
        
        if bucket == "location_linker":
            i += 1
            continue

        if bucket == "numerical":
            number = text
            i += 1
            continue

        if bucket == "measurement_unit":
            unit = text
            i += 1
            continue

        if bucket == "measurement_type":
            i += 1
            continue

        if bucket in ("measurement_max", "measurement_linker_max"):
            is_max = True
            i += 1
            continue

        if bucket == "measurement_linker":
            i += 1
            continue

        if bucket == "uncertainty_linker":
            i += 1
            continue

        if bucket in (
            "laterality",
            "cerebral_location",
            "subregional_anatomy"
        ):
            i += 1
            continue

        if bucket is None:
            i += 1
            continue

        break

    if number and unit:
        value = f"{number}{unit}"
        thickness_type = "thickness_max" if is_max else "thickness"
        return {value: thickness_type}, i

    return None, i
```

### Mass Effect Parser

```python
def parse_numeric_mass_effect(tokens, start_i):

    i = start_i

    value = None
    unit = None
    modifier = None
    direction = None

    # ============================
    # Step 1: numeric value
    # ============================
    if tokens[i][1] == "numerical":
        value = tokens[i][0]
        i += 1
    else:
        return None, start_i

    # ============================
    # Step 2: measurement unit
    # ============================
    if i < len(tokens) and tokens[i][1] == "measurement_unit":
        unit = tokens[i][0]
        i += 1

    # ============================
    # Step 3: optional modifier
    # ============================
    if i < len(tokens) and tokens[i][1] == "modifier":
        modifier = tokens[i][0]
        i += 1

    # ============================
    # Step 4: scan forward until STOP
    # ============================
    while i < len(tokens):
        text, bucket = tokens[i]

        # STOP CONDITIONS
        if bucket in MASS_NUM_STOP:
            break

        # Laterality (direction)
        if bucket == "laterality":
            direction = text
            i += 1
            continue

        # Skip harmless joiners
        if bucket in ("sandwich_and",):
            i += 1
            continue

        # Unknown → STOP
        break

    # ============================
    # Return clean measurement block
    # ============================
    measurement = {
        "value": value,
        "unit": unit,
        "modifier": modifier,
        "direction": direction
    }

    return measurement, i
```

### Negation Parser

```python
def parse_negation_branch(tokens, start_i):

    branch = {}
    pre_effect_tokens = []

    i = start_i + 1

    EFFECTS = {
        "mass_effect",
        "effacement",
        "compression",
        "herniation_subtype"
    }

    STOP_BUCKETS = {
        "period",
        "bleed_pathology",
        "negation"
    }

    while i < len(tokens):

        text, bucket = tokens[i]

        if bucket == "location_linker":
            loc_branch, next_i = parse_association_location(tokens, i)

            if loc_branch:
                branch["location"] = loc_branch

            i = next_i
            continue

        if bucket == "association_linker":
            i += 1
            continue

        if bucket in STOP_BUCKETS:
            break

        # save magnitude before effect
        if bucket == "magnitude":
            pre_effect_tokens.append((text, bucket))
            i += 1
            continue

        # effect opener (same pattern as parse_association)
        if bucket in EFFECTS:

            branch["effect_name"] = (text, bucket)

            for t_text, t_bucket in pre_effect_tokens:

                if t_bucket == "magnitude":
                    branch.setdefault("magnitude", []).append(
                        (t_text, t_bucket)
                    )

            i += 1
            continue

        if bucket == "sandwich_and":
            i += 1
            continue

        i += 1
        continue


    return branch, i
```

### Location Parser

```python
def parse_location_branch(sent_tokens, start_idx):

    branch = []
    i = start_idx + 1

    while i < len(sent_tokens):
        text, bucket = sent_tokens[i]

        # --- CONDITIONAL STOP: association_linker ---
        if bucket == "association_linker":
            # Look ahead safely
            if i + 1 < len(sent_tokens):
                next_bucket = sent_tokens[i + 1][1]

                # If next token is a location linker → continue scanning
                if next_bucket == "location_linker":
                    i += 1
                    continue

            # Otherwise → STOP
            break

        # --- NORMAL STOP TOKENS ---
        if bucket in LOCATION_STOP:
            break

        # --- ACCEPTABLE LOCATION TOKENS ---
        if bucket in LOCATION_ACCEPTABLE:
            branch.append((text, bucket))
            i += 1
            continue

        # --- OTHERWISE IGNORE AND CONTINUE ---
        i += 1

    return branch, i
```

### Association Parser

```python
def parse_association(tokens, start_i):

    # 1. Slice tokens into segments using STOP buckets
    segments = []
    current_segment = []

    i = start_i
    while i < len(tokens):
        text, bucket = tokens[i]

        # HARD STOP → end entire association function
        if bucket in ASSOCIATION_HARD_STOP:
            if current_segment:
                segments.append(current_segment)
                current_segment = []
            break

        # BRANCH STOP → end current segment, start new one
        if bucket in ASSOCIATION_BRANCH_STOP:
            if current_segment:
                segments.append(current_segment)
                current_segment = []
            i += 1
            continue

        current_segment.append((text, bucket))
        i += 1

    # Add last segment if any
    if current_segment:
        segments.append(current_segment)

    # 2. Process each segment
    associations = []

    for seg in segments:
               
        # -------------------------
        # Thickness Segment
        # -------------------------
        
        if bucket in (
            "numerical",
            "measurement_linker",
            "measurement_linker_max",
            "measurement_max",
            "measurement_type"
        ):
        
            local_tokens = sent_tokens[
                max(0, i - 5):
                min(len(sent_tokens), i + 6)
            ]
        
            has_effect = any(
                bucket in EFFECT_FAMILY
                for _, bucket in local_tokens
            )
        
            if not has_effect:
        
                thickness_result, _ = (
                    parse_thickness_branch(
                        seg,
                        0
                    )
                )
        
                if thickness_result:
        
                    associations.append(
                        {
                            "thickness": thickness_result
                        }
                    )
        
                    continue

        # -------------------------
        # Numeric Mass Effect
        # -------------------------

        if (
            any(b == "numerical" for _, b in seg)
            and
            any(b in EFFECT_FAMILY for _, b in seg)
        ):

            associations.append(
                num_mass(seg)
            )

        current = None
        pre_effect_tokens = []
        j = 0

        while j < len(seg):

            text, bucket = seg[j]

            # ⭐ NEGATION HANDLING
            if bucket == "negation":

                neg_branch, next_j = (
                    parse_negation_branch(
                        seg,
                        j
                    )
                )

                associations.append(
                    {"negation": [neg_branch]}
                )

                j = next_j
                continue

            # EFFECT OPENER
            if bucket in EFFECT_FAMILY:
               
                if current:
                    associations.append(current)

                current = {
                    "effect_name": (text, bucket)
                }

                for t_text, t_bucket in pre_effect_tokens:

                    if t_bucket in (
                        "laterality",
                        "cerebral_location",
                        "sulci",
                        "ventricle",
                        "uncus",
                    ):

                        current.setdefault(
                            "location",
                            []
                        ).append(
                            (t_text, t_bucket)
                        )

                    if t_bucket == "magnitude":

                        current.setdefault(
                            "magnitude",
                            []
                        ).append(
                            (t_text, t_bucket)
                        )

                pre_effect_tokens = []
                j += 1
                continue

            # DENSITY OPENER
            if bucket == "density":
            
                if current:
                    associations.append(current)
            
                current = {
                    "density": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # ACUITY OPENER
            if bucket == "acuity":
            
                if current:
                    associations.append(current)
            
                current = {
                    "acuity": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # MORPHOLOGY OPENER
            if bucket == "morphology":
            
                if current:
                    associations.append(current)
            
                current = {
                    "morphology": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # PROMINENCE OPENER
            if bucket == "prominence":
            
                if current:
                    associations.append(current)
            
                current = {
                    "prominence": (text, bucket)
                }
            
                for t_text, t_bucket in pre_effect_tokens:
            
                    if t_bucket in (
                        "laterality",
                        "ventricle",
                    ):
            
                        current.setdefault(
                            "location",
                            []
                        ).append(
                            (t_text, t_bucket)
                        )
                        
                pre_effect_tokens = []
                j += 1
                continue

            # LOCATION LINKER
            if bucket == "location_linker" and current:

                loc_list = []
                k = j + 1

                while k < len(seg):

                    t_text2, t_bucket2 = seg[k]

                    if t_bucket2 in ASSO_LOCATION_STOP:
                        break

                    if t_bucket2 == "sandwich_and":
                        k += 1
                        continue

                    if t_bucket2 in (
                        None,
                        "location_linker",
                        "subregional_anatomy"
                    ):
                        k += 1
                        continue

                    loc_list.append(
                        (t_text2, t_bucket2)
                    )

                    k += 1

                if loc_list:

                    current.setdefault(
                        "location",
                        []
                    ).extend(
                        loc_list
                    )

                j = k 
                continue

            # Store pre-effect tokens
            pre_effect_tokens.append(
                (text, bucket)
            )

            j += 1

        if current:
            associations.append(current)

    return associations, i
```

### Mass Effect Num Collector

```python
def num_mass(collected):

    value = None
    unit = None
    direction = None
    effect_name = None

    # Extract components from collected tokens
    for item in collected:

        if (
            isinstance(item, tuple)
            and isinstance(item[1], str)
            and item[1] in (
                "mass_effect",
                "herniation_subtype"
            )
        ):

            text = item[0].lower()

            # Always prefer midline shift
            if text == "midline shift":

                effect_name = item

            # Only use off the midline
            # if nothing better found
            elif (
                text == "off the midline"
                and effect_name is None
            ):

                effect_name = (
                    "midline shift",
                    "mass_effect"
                )

            elif effect_name is None:

                effect_name = item

        # numeric value
        if (
            isinstance(item, tuple)
            and item[1] == "numerical"
        ):
            value = item[0]

        # measurement unit
        if (
            isinstance(item, tuple)
            and item[1] == "measurement_unit"
        ):
            unit = item[0]

        # laterality
        if (
            isinstance(item, tuple)
            and item[1] == "laterality"
        ):
            direction = item[0]

    # Combine number + unit into one string
    combined_value = None

    if value:

        if unit:
            combined_value = f"{value} {unit}"
        else:
            combined_value = value

    branch = {
        "effect_name": effect_name,
        "numeric_mass_effect": {
            "value": combined_value,
            "direction": direction
        }
    }

    return branch
```

```python
def parse_association_location(tokens, start_i):
   
    i = start_i + 1
    locations = []

    while i < len(tokens):
        text, bucket = tokens[i]

        # STOP CONDITIONS
        if bucket in ASSO_LOCATION_STOP:
            break

        # ACCEPTABLE LOCATION TOKENS
        if bucket in LOCATION_ACCEPTABLE:
            locations.append(
                (text, bucket)
            )
            i += 1
            continue


        # JOINERS (sandwich_and only)
        if bucket == "sandwich_and":
            i += 1
            continue

        # UNKNOWN → STOP
        break

    return locations, i
```

## Short Sentences Function

### Other Pathology

```python
def parse_other_pathology(sent_tokens):

    negation_obj = None
    normal_state_token = None

    anatomy_tokens = []
    pathology_tokens = []
    modifier_tokens = []

    i = 0
    while i < len(sent_tokens):

        text, bucket = sent_tokens[i]

        # ============================
        # 1. NEGATION
        # ============================
        if bucket == "negation":
            neg_branch, next_i = parse_pathology_negation_branch(
                sent_tokens,
                i
            )
            negation_obj = neg_branch
            i = next_i
            continue

        # ============================
        # 2. NORMAL STATE / OUTCOME
        # ============================
        if bucket in {"normal_state", "outcome"}:
            normal_state_token = (text, bucket)
            i += 1
            continue

        # ============================
        # 3. ANATOMY
        # ============================
        if bucket in {
            "extracranial_anatomy",
            "sinus",
            "intraparenchymal",
            "sulci",
            "cistern",
            "ventricle",
            "cerebral_location",
            "bone",
        }:
            anatomy_tokens.append((text, bucket))
            i += 1
            continue

        # ============================
        # 4. PATHOLOGY
        # ============================
        if bucket in {
            "pathology",
            "general_pathology"
        }:
            pathology_tokens.append((text, bucket))
            i += 1
            continue

        # ============================
        # 5. MODIFIERS
        # ============================
        if bucket in {
            "magnitude",
            "acuity"
        }:
            modifier_tokens.append((text, bucket))
            i += 1
            continue

        # ============================
        # 6. LOCATION LINKER
        # ============================
        if bucket == "location_linker":
            loc_branch, next_i = parse_association_location(
                sent_tokens,
                i
            )

            anatomy_tokens.extend(loc_branch)
            i = next_i
            continue

        i += 1

    # ============================
    # BUILD FINAL OBJECT
    # ============================
    result = {}

    # NEGATION OVERRIDES EVERYTHING
    if negation_obj:
        result["negation"] = negation_obj
        return result

    # NORMAL OUTCOME
    if normal_state_token:
        result["outcome"] = {
            "normal_state": normal_state_token,
            "anatomy": anatomy_tokens if anatomy_tokens else None
        }
        return result

    # POSITIVE PATHOLOGY
    if pathology_tokens:   
        result["outcome"] = {
            "pathology": pathology_tokens if pathology_tokens else None,
            "anatomy": anatomy_tokens if anatomy_tokens else None,
            "modifiers": modifier_tokens if modifier_tokens else []
        }

    return result

    return result
```

### Other Path Negation

```python
def parse_pathology_negation_branch(tokens, start_i):

    branch = {"pathology": [], "modifiers": [], "anatomy": []}
    i = start_i + 1

    PATHOLOGY_FAMILY = {"pathology", "bone_pathology"}
    MODIFIER_FAMILY = {"acuity", "magnitude", "territorial", "ischemic", "laterality"}
    ANATOMY_FAMILY = {"extracranial_anatomy", "cerebral_location", "sinus", "ventricle", "sulci"}

    STOP_BUCKETS = {"period"}

    while i < len(tokens):

        text, bucket = tokens[i]

        # STOP
        if bucket in STOP_BUCKETS:
            break

        # LOCATION
        if bucket == "location_linker":
            loc_branch, next_i = parse_association_location(tokens, i)
            branch["anatomy"].extend(loc_branch)
            i = next_i
            continue

        # PATHOLOGY
        if bucket in PATHOLOGY_FAMILY:
            branch["pathology"].append((text, bucket))
            i += 1
            continue

        # MODIFIERS
        if bucket in MODIFIER_FAMILY:
            branch["modifiers"].append((text, bucket))
            i += 1            
            continue

        # ANATOMY
        if bucket in ANATOMY_FAMILY:
            branch["anatomy"].append((text, bucket))
            i += 1
            continue

        # pass-through tokens
        if bucket in {"sandwich_and", "logical_connector", "noticing", "comma"}:
            i += 1
            continue

        # ⭐ FIX: allow None bucket to pass through
        if bucket is None:
            i += 1
            continue

        # Unknown → stop
        break

    return branch, i
```

## Detect Pathologies Function

```python
def detect_pathology_root(sent_tokens):
    for text, bucket in sent_tokens:
        if bucket == "bleed_pathology":
            return "bleed_pathology"
        if bucket in ("pathology", "bone_pathology"):
            return bucket
    return None
```

```python
def detect_other_pathology_root(sent_tokens):

    has_target_anatomy = any(bucket in {
        "extracranial_anatomy",
        "sinus",
        "intraparenchymal",
        "sulci",
        "cistern",
        "ventricle",
        "cerebral_location",
        "bone",
    } for _, bucket in sent_tokens)

    has_state = any(bucket in {
        "normal_state",
        "outcome"
    } for _, bucket in sent_tokens)

    if has_target_anatomy and has_state:
        return True

    return False
```

```python
def detect_relevant_sentence(sent_tokens):

    RELEVANT_BUCKETS = {

        "bleed_pathology",
        "pathology",
        "bone_pathology",

        "mass_effect",
        "herniation_subtype",

        "density",
        "acuity",

        "cerebral_location",
        "intraparenchymal",
        "ventricle",
        "cistern",
        "sulci",
        "extracranial_anatomy",
        "bone",

        "normal_state",
        "negation",

    }

    return any(
        bucket in RELEVANT_BUCKETS
        for _, bucket in sent_tokens
    )
```

## Orchestrator Function

```python
def orchestrator(sentence_lists):

    bleed_trees = []
    current_tree = None

    for sent_tokens in sentence_lists:

        # Detect pathology in this sentence
        root_bucket = detect_pathology_root(sent_tokens)

        # ============================
        # CASE 1: BLEED PATHOLOGY
        # ============================
        if root_bucket == "bleed_pathology":

            new_pathology = None
            for text, bucket in sent_tokens:
                if bucket == "bleed_pathology":
                    new_pathology = text
                    break

            # Start new bleed tree if needed
            if current_tree is None or current_tree["pathology"] != new_pathology:
                current_tree = build_bleed_tree(
                    sent_tokens,
                    is_bleed_sentence=True,
                    existing_tree=None
                )
                bleed_trees.append(current_tree)
                continue
                

            # Extend existing bleed tree
            current_tree = build_bleed_tree(
                sent_tokens,
                is_bleed_sentence=True,
                existing_tree=current_tree
            )
            continue

        # ============================
        # CASE 2: OTHER PATHOLOGY (infarct, ischemia, mass, edema, etc.)
        # ============================
        if root_bucket in ("pathology", "bone_pathology"):
            other_result = parse_other_pathology(sent_tokens)
            bleed_trees.append(other_result)
            continue

        # ============================
        # CASE 2B: SINUS / EXTRACRANIAL + NORMAL_STATE/OUTCOME
        # ============================
        if detect_other_pathology_root(sent_tokens):                   
            other_result = parse_other_pathology(sent_tokens)
            bleed_trees.append(other_result)
            continue

        # ============================
        # NOT RELEVANT
        # ============================
        
        if not detect_relevant_sentence(
            sent_tokens
        ):
            continue

        # ============================
        # CASE 3: NO PATHOLOGY → continue bleed tree
        # ============================
        
        if current_tree is not None:
            
            current_tree = build_bleed_tree(
                sent_tokens,
                is_bleed_sentence=False,
                existing_tree=current_tree
            )

    return bleed_trees
```
