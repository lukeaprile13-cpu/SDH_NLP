```python
import spacy
from spacy.matcher import PhraseMatcher
from spacy.language import Language
from spacy.util import filter_spans     # removes overlapping phrases (there is, there is a) to get the longest phrase

nlp = spacy.load("en_core_web_sm")
nlp.disable_pipes(["parser"])
```

```python
# --- Collect all bucket dictionaries into one list ---
all_dicts = [
    BLEED_PATHOLOGY,
    OTHER_PATHOLOGIES,
    INTRACRANIAL_ANATOMY,
    EXTRACRANIAL_ANATOMY,
    BONE,
    NOTICING,
    LINKERS,
    UNCERTAINTY,
    CEREBRAL_LOCATIONS,
    SUBREGIONAL_ANATOMY,
    MASS_EFFECT,
    HERNIATION_SUBTYPES,
    ACUITY,
    DENSITY,
    PROMINENCE,
    MAGNITUDE,
    MORPHOLOGY,
    NORMAL_STATE,
    MEASUREMENT,
    NEGATION,
    LATERALITY,
    IGNORE,
    SINUS,
    REPORT_TITLES,
    PUNCTUATION,
    NUMERICAL
]


# --- Extract multi-word keys ---
multiword_phrases = []

for d in all_dicts:
    for tok in d.keys():
        if " " in tok or "-" in tok or "/" in tok:   # contains a space, - or / (multi-word)
            multiword_phrases.append(tok)

patterns = [nlp.make_doc(text) for text in multiword_phrases]
matcher = PhraseMatcher(nlp.vocab, attr="LOWER")
matcher.add("TOKEN_BUCKET_PHRASES", patterns)

@Language.component("merge_phrases")
def merge_phrases(doc):
    matches = matcher(doc)
    spans = [doc[start:end] for _, start, end in matches]
    filtered = filter_spans(spans)
    with doc.retokenize() as retokenizer:
        for span in filtered:
            retokenizer.merge(span)
    return doc

nlp.add_pipe("merge_phrases", first=True)

# --- Force sentence splitting ONLY by punctuation ---
@Language.component("punct_only_sentences")
def punct_only_sentences(doc):

    for token in doc:
        token.is_sent_start = False

    for i, token in enumerate(doc):

        if i == 0:
            token.is_sent_start = True
            continue

        # normal punctuation
        if doc[i-1].text in {".", ";", ":"}:
            token.is_sent_start = True
            continue

        # implicit sentence break:
        # Capitalized word but NOT all caps
        if (
            token.text[0].isupper()
            and not token.text.isupper()
        ):
            token.is_sent_start = True

    return doc

# GUARANTEED to run after parser
nlp.add_pipe("punct_only_sentences", last=True)
```

```python
import re

# ============================
# REPORT PREPROCESSING
# ============================

# every word that appears in a dictionary key, plus common report words
KNOWN_WORDS = set()

for key in TOKEN_BUCKETS:
    for word in re.findall(r"[a-z]+", key.lower()):
        KNOWN_WORDS.add(word)

KNOWN_WORDS.update({
    "hemorrhage", "dimension", "differentiation", "intraventricular", "excluded",
    "measuring", "measures", "significant", "subdural", "hematoma", "hemorrhage",
    "herniation", "ventricle", "ventricles", "right", "left", "side", "which",
    "blood", "image", "parietal", "lateral", "microangiopathic", "subarachnoid",
    "lateral", "sulci", "interpretation", "consistent", "hemorrhages",
})


def repair_split_words(text):
    """
    Dictated / OCR reports often contain words broken by a space
    ("subd ural", "herni ation", "m easuring", "righ t").
    Join two fragments when neither is a known word but the joined
    word is.
    """

    tokens = text.split(" ")
    repaired = []
    i = 0

    while i < len(tokens):

        if i + 1 < len(tokens):

            left = tokens[i]
            right = tokens[i + 1]

            left_core = left.lower().strip(".,;:()")
            right_core = right.lower().strip(".,;:()")
            joined = (left_core + right_core)

            if (
                left_core.isalpha()
                and right_core.isalpha()
                and left_core not in KNOWN_WORDS
                and (right_core not in KNOWN_WORDS or len(left_core) >= 4 or len(right_core) == 1)
                and joined in KNOWN_WORDS
            ):
                repaired.append(left + right)
                i += 2
                continue

        repaired.append(tokens[i])
        i += 1

    return " ".join(repaired)


def preprocess_report(text):

    text = str(text)

    # "|" is used as a line break inside sentences
    text = text.replace("|", " ")

    # glued words: "injuryTechnique", "2022There", "reviewFINDINGS", "2026.FINDINGS"
    text = re.sub(r"(?<=[a-z0-9])(?=[A-Z][a-z])", " ", text)
    text = re.sub(r"(?<=[a-z])(?=[A-Z]{3,})", " ", text)
    text = re.sub(r"(?<=\d)\.(?=[A-Z][A-Za-z])", ". ", text)

    # names put in by de-identification: "MERWIN-white differentiation",
    # "BASSITI-white matter differentiation" -> gray-white
    text = re.sub(r"\b[A-Za-z]+-white( matter)? diff ?erentiation", r"gray-white\1 differentiation", text)

    # glued side words: "rightinterhemispheric fissure", "lefttentorium"
    text = re.sub(
        r"(?i)\b(right|left)(?=(interhemispheric|tentorium|tentorial|falx|frontal|parietal|temporal|occipital|convexity|hemispheric))",
        r"\1 ", text)

    # "There are subacute and chronic components" = subacute on chronic
    text = re.sub(r"(?i)\b(acute|subacute|recent)\s+and\s+chronic\s+component", r"\1 on chronic component", text)

    # "acute on subacute/chronic" = acute on chronic
    text = re.sub(r"(?i)\bacute on subacute\s*/\s*chronic", "acute on chronic", text)

    # "The midline is intact" says there is no midline shift
    text = re.sub(r"(?i)\bthe midline is intact\b", "No midline shift", text)

    # list items: "COMPONENT.-RIGHT: SMALL" -> "COMPONENT. RIGHT: SMALL"
    text = re.sub(r"\.\s*-\s*(?=[A-Za-z])", ". ", text)

    # "FINDINGS:There" -> "FINDINGS: There"
    text = re.sub(r":(?=[A-Za-z])", ": ", text)

    # "1 cm.There" -> "1 cm. There"   (not decimals)
    text = re.sub(r"(?<=[A-Za-z\)])\.(?=[A-Z])", ". ", text)

    # numbered impression lines "1. Large ..." -> "Large ..."
    text = re.sub(r"(?<![\d.])\b\d\.\s+(?=[A-Z])", " ", text)

    text = re.sub(r"[ \t]+", " ", text)

    return repair_split_words(text)


# ============================
# SECTIONS
# ============================

# sentences under these headings are history / technique, not findings
SKIP_SECTIONS = {
    "clinical history", "history", "clinical indication", "indication",
    "clinical information", "comparison", "comparison study", "technique",
    "procedure", "reference examination", "reference examinations",
    "exam", "contrast",
}


KEEP_SECTIONS = {
    "findings", "imaging findings", "impression", "imaging result", "imaging_result",
    "opinion", "summary", "conclusion", "interpretation",
}


def heading_text(sent_tokens):
    """
    A heading sentence ends in ':' e.g. [("FINDINGS", ...), (":", "colon")].
    It is either a known heading (possibly glued to the end of the
    previous section's text, "... 2009 Imaging Findings :") or a short
    run of words ("CT HEAD:", "Midline:").
    """

    if not sent_tokens or sent_tokens[-1][1] != "colon":
        return None

    words = [text for text, bucket in sent_tokens[:-1] if text not in ("-", "*", "|")]

    if not words:
        return None

    joined = " ".join(words).lower().strip()

    for name in SKIP_SECTIONS | KEEP_SECTIONS:
        if joined == name or joined.endswith(" " + name):
            return name

    if len(words) <= 4:
        return joined

    return None


def reports_finding(sent_tokens):

    buckets = [bucket for _, bucket in sent_tokens]

    if buckets and buckets[0] == "noticing":
        return True

    has_number = "numerical" in buckets and "measurement_unit" in buckets
    has_finding = "bleed_pathology" in buckets or any(b in EFFECT_FAMILY for b in buckets)

    return has_number and has_finding


# these sections are a single line ("COMPARISON: NONE.") - the next
# sentence is findings even when no "FINDINGS:" heading follows
ONE_SENTENCE_SECTIONS = {
    "comparison", "comparison study", "technique", "procedure", "contrast",
    "reference examination", "reference examinations",
}

SIDE_HEADINGS = {"right": "right", "left": "left", "on the right": "right", "on the left": "left"}


def filter_sections(sentence_lists):

    kept = []
    skipping = False
    one_sentence_left = False
    side_label = None           # "-RIGHT: SMALL. 10 MM THICKNESS"

    for sent_tokens in sentence_lists:

        heading = heading_text(sent_tokens)

        if heading is not None and heading.strip("- ") in SIDE_HEADINGS:
            side_label = SIDE_HEADINGS[heading.strip("- ")]
            continue

        if heading is not None:

            skipping = heading in SKIP_SECTIONS
            one_sentence_left = heading in ONE_SENTENCE_SECTIONS
            side_label = None

            continue

        if skipping and one_sentence_left:
            one_sentence_left = False
            skipping = False
            continue

        # every sentence under a side label says which side it is about
        if side_label and not skipping:
            sent_tokens = [(side_label, "laterality")] + list(sent_tokens)

        # a history / technique section ends early when a sentence
        # clearly reports a finding ("There is ...", "Largely isodense
        # left SDH ... measuring 1.7 cm") - headings are often missing
        if skipping and reports_finding(sent_tokens):
            skipping = False

        if not skipping:
            kept.append(sent_tokens)

    return kept


# ============================
# FULL PIPELINE
# ============================

def tokenize_report(report):

    doc = nlp(preprocess_report(report))

    sentence_lists = []

    for sent in doc.sents:

        sent_tokens = []

        for token in sent:
            bucket = get_bucket(token.text)
            sent_tokens.append((token.text, bucket))

        sentence_lists.append(sent_tokens)

    for sent_tokens in sentence_lists:

        for idx, (text, bucket) in enumerate(sent_tokens):

            if text.lower() == "and":

                sent_tokens[idx] = (
                    "and",
                    classify_and(sent_tokens, idx)
                )

    return filter_sections(sentence_lists)


def run_report(report):

    sentence_lists = tokenize_report(report)

    bleed_trees = [orchestrator(sentence_lists)]

    return master_variable_mapping_function(bleed_trees)
```

## VARIABLE MAPPING

### Master Mapping function

```python
def master_variable_mapping_function(
    bleed_trees
):

    final_variables = {}

    # side of the SDH (used for "ipsilateral" and side-less sulcal effacement)
    sdh_side = None

    for tree_group in bleed_trees:

        for tree in tree_group:

            if (
                isinstance(tree, dict)
                and "pathology" in tree
                and tree.get("pathology")
            ):
                sdh_side = summarize_sdh_side(tree["children"])

    for tree_group in bleed_trees:

        for tree in tree_group:

            if not tree:
                continue

            all_mapped_variables = []
            all_direct_variables = {}

            # -------------------------
            # SDH Tree
            # -------------------------

            if "pathology" in tree:

                sdh_children = []
                association_children = []

                for child in tree["children"]:

                    if (
                        isinstance(child, dict)
                        and "association" in child
                    ):

                        # -------------------------
                        # Pull SDH characteristics
                        # back into sdh_children
                        # (in place, so the side they were said on is kept)
                        # -------------------------

                        previous_side = None

                        for association in child["association"]:

                            # "mass effect on the left hemisphere with
                            # effacement of the hemispheric sulci"
                            if "effect_name" in association:
                                sides_here = [
                                    t for t, b in association.get("location", [])
                                    if b == "laterality"
                                ]
                                if sides_here:
                                    previous_side = sides_here[0]
                                elif previous_side and any(
                                    b == "sulci" for _, b in association.get("location", [])
                                ):
                                    association = dict(association)
                                    association["location"] = (
                                        association["location"] + [(previous_side, "laterality")]
                                    )

                            if "density" in association:

                                sdh_children.extend(
                                    association["density"]
                                )

                            elif "acuity" in association:

                                sdh_children.extend(
                                    association["acuity"]
                                )

                            elif "morphology" in association:

                                sdh_children.extend(
                                    association["morphology"]
                                )

                            elif "thickness" in association:

                                sdh_children.append(
                                    {
                                        "thickness":
                                        association["thickness"]
                                    }
                                )

                            elif (
                                "negation" in association
                                and any("density" in n for n in association["negation"])
                            ):

                                sdh_children.append(association)

                            else:

                                association_children.append(
                                    [association]
                                )

                        continue

                    sdh_children.append(
                        child
                    )

                # -------------------------
                # SDH Characteristics
                # -------------------------

                (
                    sdh_mapped_variables,
                    sdh_direct_variables
                ) = SDH_characteristics(
                    sdh_children
                )

                # -------------------------
                # SDH Associations
                # -------------------------

                (
                    assoc_mapped_variables,
                    assoc_direct_variables
                ) = SDH_association_characteristics(
                    association_children,
                    sdh_side
                )

                all_mapped_variables.extend(
                    sdh_mapped_variables
                )

                all_mapped_variables.extend(
                    assoc_mapped_variables
                )

                all_direct_variables.update(
                    sdh_direct_variables
                )

                all_direct_variables.update(
                    assoc_direct_variables
                )

            # -------------------------
            # Negation Tree (a list of negated heads)
            # -------------------------

            elif "negation" in tree:

                negations = tree["negation"]

                if isinstance(negations, dict):
                    negations = [negations]

                for negation in negations:

                    (
                        mapped_variables,
                        direct_variables
                    ) = negation_sorter(
                        negation,
                        sdh_side
                    )

                    all_mapped_variables.extend(
                        mapped_variables
                    )

                    all_direct_variables.update(
                        direct_variables
                    )

            # -------------------------
            # Outcome Tree
            # -------------------------

            elif "outcome" in tree:

                (
                    mapped_variables,
                    direct_variables
                ) = outcome_function(
                    tree["outcome"]
                )

                all_mapped_variables.extend(
                    mapped_variables
                )

                all_direct_variables.update(
                    direct_variables
                )

            # -------------------------
            # Other hemorrhage (SAH / IPH / IVH)
            # -------------------------

            elif "hemorrhage" in tree:

                all_mapped_variables.extend(
                    hemorrhage_variables(tree["hemorrhage"]["type"], tree["hemorrhage"].get("grade"))
                )

            # -------------------------
            # Sinus / mastoid
            # -------------------------

            elif "sinus" in tree:

                all_direct_variables.update(
                    tree["sinus"]
                )

            elif "fracture" in tree:

                all_direct_variables.update(
                    tree["fracture"]
                )

            else:

                continue

            # -------------------------
            # Final Mapping
            # -------------------------

            final_array_function(
                all_mapped_variables,
                all_direct_variables,
                final_variables
            )

    if not final_variables:
        final_array_function([], {}, final_variables)

    return post_process_variables(final_variables)


def summarize_sdh_side(children):
    """
    "L" / "R" / "B" for the SDH, from the lateralities said in the
    bleed sentences.
    """

    named = set()

    for child in children:

        if isinstance(child, tuple) and child[1] == "laterality":
            side = side_of(child[0])
            if side:
                named.add(side)

        elif isinstance(child, dict) and "location" in child:
            for text, bucket in child["location"]:
                if bucket == "laterality" and side_of(text):
                    named.add(side_of(text))

    if "B" in named or {"L", "R"} <= named:
        return "B"

    if named:
        return next(iter(named))

    return None


def hemorrhage_grade(sent_tokens):
    """
    "scattered areas of ... subarachnoid hemorrhage" -> Scattered,
    "minor / small volume / trace subarachnoid blood" -> Mild,
    "moderate" -> Moderate, "diffuse / extensive / large" -> Severe
    """

    # whole words only ("within" is not "thin")
    words = set(" ".join(t.lower() for t, _ in sent_tokens).split())

    if "scattered" in words:
        return "Scattered"

    if any(w in words for w in ("diffuse", "extensive", "large", "severe", "significant")):
        return "Severe"

    if "moderate" in words:
        return "Moderate"

    if any(w in words for w in ("minor", "small", "trace", "mild", "minimal", "tiny", "thin")):
        return "Mild"

    return None


def hemorrhage_variables(btype, grade=None):

    if btype == "subarachnoid" and grade:
        return [f"subarachnoid_hemorrhage_{grade}"]


    names = {
        "subarachnoid": "subarachnoid_hemorrhage",
        "intraparenchymal": "intraparenchymal_hemorrhage",
        "intraventricular": "intraventricular_hemorrhage",
        "epidural": "epidural_hemorrhage",
    }

    if btype in names:
        return [names[btype]]

    return []


def post_process_variables(final_variables):
    """
    Values that depend on the whole report.
    """

    # -------------------------
    # Overall SDH laterality
    # -------------------------
    left = final_variables.pop("_sdh_left", False)
    right = final_variables.pop("_sdh_right", False)

    if left and right:
        final_variables["laterality"] = "Bilateral"
    elif left:
        final_variables["laterality"] = "Left"
    elif right:
        final_variables["laterality"] = "Right"

    return final_variables
```

### Association_mapping

```python
def SDH_association_characteristics(
    association_children,
    sdh_side=None
):

    mapped_variables = []
    direct_variables = {}

    for group in association_children:

        for association in group:

            # -------------------------
            # Effects
            # -------------------------

            if "effect_name" in association:

                (
                    function_mapped_variables,
                    function_direct_variables
                ) = effect_name_sorter(
                    association,
                    sdh_side
                )
                
            # -------------------------
            # Negation
            # -------------------------

            elif "negation" in association:

                for negation in association["negation"]:
            
                    (
                        function_mapped_variables,
                        function_direct_variables
                    ) = negation_sorter(
                        negation,
                        sdh_side
                    )
            
                    mapped_variables.extend(
                        function_mapped_variables
                    )
            
                    direct_variables.update(
                        function_direct_variables
                    )
            
                continue

            elif "prominence" in association:

                (
                    function_mapped_variables,
                    function_direct_variables
                ) = prominence_sorter(
                    association
                )

            else:

                continue

            mapped_variables.extend(
                function_mapped_variables
            )

            direct_variables.update(
                function_direct_variables
            )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Effect name Sorter

```python
LATERALITY_MAP = {

    "left": "Left",
    "left-sided": "Left",
    "left sided": "Left",
    "to the left": "Left",
    "leftward": "Left",
    "towards the left": "Left",
    "on the left": "Left",
    "right to left": "Left",

    "right": "Right",
    "right-sided": "Right",
    "right sided": "Right",
    "to the right": "Right",
    "rightward": "Right",
    "towards the right": "Right",
    "on the right": "Right",
    "left to right": "Right",

    "bilateral": "Bilateral",
    "bilaterally": "Bilateral",

}

SDH_SIDE_NAMES = {"L": "Left", "R": "Right", "B": "Bilateral"}

HERNIATION_MAP = {

    "subfalcine": "herniation_subfalcine",
    "subfalcine herniation": "herniation_subfalcine",
    "uncal": "herniation_uncal",
    "uncal herniation": "herniation_uncal",
    "early uncal herniation": "herniation_uncal",
    "early uncal": "herniation_uncal",
    "uncal shift": "herniation_uncal",
    "medialized uncus": "herniation_uncal",
    "transtentorial herniation": "herniation_transtentorial",
    "transtentorial": "herniation_transtentorial",
    "descending transtentorial herniation": "herniation_transtentorial",
    "tonsillar": "herniation_tonsillar",
    "tonsillar herniation": "herniation_tonsillar",
    "cerebellar tonsillar herniation": "herniation_tonsillar",
    "downward tonsillar herniation": "herniation_tonsillar",

}

# "no brain herniation" negates every subtype
GENERIC_HERNIATION = {
    "herniation", "brain herniation", "intracranial herniation",
}

# effects that squeeze a ventricle / cistern (not "shift" or "deviation")
SQUEEZE_EFFECTS = {
    "narrowing", "decreasing the size of", "compressive effects", "compressive effect",
    "compressed", "local mass effect", "localized mass effect",
    "effaced", "partially effaced",
    "effacement", "compression", "obliteration", "obliterated", "crowding",
    "medial positioning", "distortion", "distorsion", "displacement",
    "mass effect", "indentation",
}

SULCAL_EFFECT_WORDS = {"sulcal blunting", "cortical buckling"}

SPECIFIC_CISTERNS = {
    "suprasellar": "cistern_suprasellar",
    "suprasellar cistern": "cistern_suprasellar",
    "ambient": "cistern_ambient",
    "ambient cistern": "cistern_ambient",
    "ambient cisterns": "cistern_ambient",
    "interpeduncular": "cistern_interpeduncular",
    "quadrigeminal": "cistern_quadrigeminal",
    "prepontine": "cistern_prepontine",
}


def is_midline_shift(effect_name, location_texts):

    if "midline" in effect_name and effect_name != "off the midline":
        return True

    # "5 mm shift of the septum pellucidum to the right"
    if "septum pellucidum" in effect_name:
        return True

    # "deviation of the midline structures", "shift ... midline structures"
    if effect_name in ("shift", "deviation", "shifted", "displacement"):
        return any("midline" in text for text in location_texts)

    return False


def effect_name_sorter(
    association,
    sdh_side=None
):

    mapped_variables = []
    direct_variables = {}

    effect_name = association[
        "effect_name"
    ][0].lower()

    # -------------------------
    # Extract magnitude
    # -------------------------

    severity = "Yes NS"

    if "magnitude" in association:

        severity = magnitude_function(
            association["magnitude"]
        )

    # "obliteration" of a ventricle is severe effacement
    if effect_name in ("obliteration", "obliterated") and severity == "Yes NS":
        severity = "Severe"

    # -------------------------
    # Extract location info
    # -------------------------

    location_buckets = set()
    location_texts = []

    location_laterality = None

    ventricle_type = None

    cistern_variables = []

    for text, bucket in association.get("location", []):

        text = text.lower()

        location_buckets.add(
            bucket
        )

        location_texts.append(text)

        if bucket == "laterality":

            if text == "ipsilateral" and sdh_side:
                location_laterality = SDH_SIDE_NAMES.get(sdh_side)

            elif text == "contralateral" and sdh_side in ("L", "R"):
                location_laterality = "Right" if sdh_side == "L" else "Left"

            elif location_laterality is None:
                location_laterality = LATERALITY_MAP.get(text)

        elif bucket == "ventricle":

            if text in ("lateral ventricle", "anterior horn", "posterior horn",
                        "occipital horn", "temporal horn", "body",
                        "lateral and third ventricles", "ventricular trigone", "trigone"):

                ventricle_type = ventricle_type or "lateral"

            elif text in ("lateral ventricles", "lateral ventricles bilaterally"):

                ventricle_type = "lateral"

                # -------------------------
                # Bilateral override
                # -------------------------

                location_laterality = (
                    "Bilateral"
                )

            elif text == "third ventricle":

                ventricle_type = ventricle_type or "third"

            elif text == "fourth ventricle":

                ventricle_type = ventricle_type or "fourth"

            else:

                ventricle_type = ventricle_type or "system"

        elif bucket == "cistern":

            cistern_variables.append(
                SPECIFIC_CISTERNS.get(text, "effacement_cisterns")
            )

    # "effacement ... in both cerebral hemispheres"
    if "both" in " ".join(location_texts) or association.get("both"):
        location_laterality = "Bilateral"

    midline = is_midline_shift(effect_name, location_texts)

    # -------------------------
    # Mass Effect
    # -------------------------

    if effect_name in ("mass effect", "mass effects", "local mass effect",
                       "localized mass effect", "mass-effect",
                       "compressive effects", "compressive effect"):

        mapped_variables.append(
            f"mass_effect_present_{severity}"
        )

    # -------------------------
    # Uncal Herniation (medialized uncus)
    # -------------------------

    if "uncus" in location_buckets and not midline:

        mapped_variables.append(
            f"herniation_uncal_{severity}"
        )

    # -------------------------
    # Midline Shift
    # -------------------------

    if midline:

        mapped_variables.append(
            f"midline_shift_present_{severity}"
        )

        direction = None

        if "numeric_mass_effect" in association:

            shift_data = association[
                "numeric_mass_effect"
            ]

            value_text = (shift_data.get("value") or "").lower()

            mm, cm = measurement_to_mm_cm(value_text)

            if mm is not None:

                direct_variables[
                    "midline_shift_mm"
                ] = mm

                direct_variables[
                    "midline_shift_cm"
                ] = cm

            if shift_data.get("direction"):

                direction = LATERALITY_MAP.get(
                    shift_data["direction"].lower()
                )

        if direction is None:
            direction = location_laterality

        if direction in ("Left", "Right"):

            direct_variables[
                "midline_shift_direction"
            ] = direction

    # -------------------------
    # Herniation
    # ------------------------

    elif effect_name in HERNIATION_MAP:

        mapped_variables.append(
            f"{HERNIATION_MAP[effect_name]}_{severity}"
        )

    elif effect_name in GENERIC_HERNIATION:

        mapped_variables.append(
            f"herniation_generic_{severity}"
        )

    # -------------------------
    # Sulcal Effacement
    # -------------------------

    if (
        ("sulci" in location_buckets and not midline)
        or effect_name in SULCAL_EFFECT_WORDS
    ):

        # "diffuse sulcal effacement" is recorded as "Diffuse"
        sulci_value = severity
        if severity == "Yes NS" and any(
            t.lower() in ("diffuse", "global") for t, _ in association.get("magnitude", [])
        ):
            sulci_value = "Diffuse"

        mapped_variables.append(
            f"effacement_sulci_{sulci_value}"
        )

        # only a side that is written ("left sulcal", "ipsilateral",
        # "both hemispheres"); the human sheet does not assume the SDH side
        sulci_laterality = location_laterality

        if sulci_laterality:

            mapped_variables.append(
                f"effacement_sulci_laterality_{sulci_laterality}"
            )

    # -------------------------
    # Ventricular Effects (only effects that squeeze the ventricle)
    # -------------------------

    squeezes = effect_name in SQUEEZE_EFFECTS and not midline

    if squeezes and ventricle_type == "lateral":

        mapped_variables.append(
            f"effacement_lateral_ventricle_{severity}"
        )

        if location_laterality:

            mapped_variables.append(
                f"l_ventricle_laterality_{location_laterality}"
            )

    elif squeezes and ventricle_type == "third":

        mapped_variables.append(
            f"effacement_third_ventricle_{severity}"
        )

    elif squeezes and ventricle_type == "fourth":

        mapped_variables.append(
            f"effacement_forth_ventricle_{severity}"
        )

    elif squeezes and ventricle_type == "system":

        mapped_variables.append(
            f"effacement_ventricular_system_{severity}"
        )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Negation

```python
def negation_sorter(
    negation,
    sdh_side=None
):

    # -------------------------
    # Negated pathology
    # -------------------------

    if "pathology" in negation:

        mapped_variables, direct_variables = (
            pathology_function(
                negation["pathology"],
                negation.get(
                    "modifiers",
                    []
                ),
                negation.get(
                    "anatomy",
                    []
                ),
                negated=True
            )
        )

    # -------------------------
    # Negated effects
    # -------------------------

    elif "effect_name" in negation:

        mapped_variables, direct_variables = (
            effect_name_sorter(
                negation,
                sdh_side
            )
        )

        # a negated shift has no size / direction
        direct_variables = {}

    # -------------------------
    # Negated bleeds ("no acute parenchymal or subarachnoid hemorrhage")
    # -------------------------

    elif "bleed" in negation:

        text, btype = negation["bleed"]

        mapped_variables = hemorrhage_variables(btype)
        direct_variables = {}

        anatomy_buckets = {b for _, b in negation.get("anatomy", [])}

        # "no hemorrhage in the basilar cisterns" -> no subarachnoid hemorrhage
        if btype == "generic" and "cistern" in anatomy_buckets:
            mapped_variables.append("subarachnoid_hemorrhage")

        # "No acute intracranial hemorrhage" -> no intraparenchymal hemorrhage
        elif btype == "generic":
            mapped_variables.append("intraparenchymal_hemorrhage")

        # "parenchymal" saved in front of the head negates IPH as well
        for anatomy_text, anatomy_bucket in negation.get("anatomy", []):
            if anatomy_bucket == "intraparenchymal" and "parenchym" in anatomy_text.lower():
                mapped_variables.append("intraparenchymal_hemorrhage")
            if anatomy_bucket == "ventricle" and "intraventricular" in anatomy_text.lower():
                mapped_variables.append("intraventricular_hemorrhage")

    # -------------------------
    # Negated outcomes
    # -------------------------

    elif "outcome" in negation:

        mapped_variables, direct_variables = (
            outcome_function(
                negation["outcome"]
            )
        )

    else:

        return (
            [],
            {}
        )

    # -------------------------
    # Negate mapped variables
    # -------------------------
    mapped_variables = [
        variable if variable.startswith("N_") else f"N_{variable}"
        for variable in mapped_variables

    ]

    return (
        mapped_variables,
        direct_variables
    )
```

#### Pathology

```python
INFARCT_VARIABLES = {
    "territorial_ischemic",
    "unspecified_infarct",
    "territorial_infarct",
    "lacunar_infarct",
    "cortical_infarct",
}

SEVERITY_VARIABLES = {
    "encephalomalacia",
    "cerebral_atrophy",
    "microangiopathy",
    "hydrocephalus",
}

ACUTE_WORDS = {"acute", "new", "subacute"}
CHRONIC_WORDS = {"chronic", "old", "remote", "prior"}


def infarct_qualifier(modifier_tokens):
    """
    "No acute infarct" -> Acute, "Old infarct" -> Chronic, otherwise NS.
    ("recent" / "interval" stay NS, as the human sheet does)
    """

    for text, bucket in modifier_tokens:

        if bucket != "acuity":
            continue

        text = text.lower()

        if text in ACUTE_WORDS:
            return "Acute"

        if text in CHRONIC_WORDS:
            return "Chronic"

    return "NS"


def pathology_function(
    pathology_list,
    modifier_tokens=None,
    anatomy_tokens=None,
    negated=False
):

    modifier_tokens = modifier_tokens or []

    pathology_text = []

    for text, bucket in pathology_list:

        pathology_text.append(
            text
        )

    mapped_variables = pathology_filter(
        pathology_text
    )

    # -------------------------
    # Anatomy-based mappings
    # -------------------------

    if anatomy_tokens:

        anatomy_text = []

        for text, bucket in anatomy_tokens:

            anatomy_text.append(
                text
            )

        anatomy_variables = (
            pathology_anatomy_filter(
                anatomy_text
            )
        )

        mapped_variables.extend(
            anatomy_variables
        )

    # -------------------------
    # Generic abnormality
    # -------------------------

    if "generic_abnormality" in mapped_variables:

        mapped_variables.remove(
            "generic_abnormality"
        )

    # -------------------------
    # Infarct type from the saved modifiers
    # ("large artery territory ischemic infarct", "cortical infarction")
    # -------------------------

    modifier_buckets = {bucket for _, bucket in modifier_tokens}

    updated_variables = []

    qualifier_tokens = modifier_tokens + [
        (word, "acuity")
        for text, _ in pathology_list
        for word in text.lower().split()
    ]

    pathology_words = " ".join(pathology_text).lower()

    for variable in mapped_variables:

        if variable in ("unspecified_infarct", "territorial_infarct") and "cortical" in pathology_words:
            variable = "cortical_infarct"

        if variable == "unspecified_infarct":

            if "cortical" in modifier_buckets:
                variable = "cortical_infarct"

            elif "territorial" in modifier_buckets:
                variable = "territorial_infarct"

        if variable in INFARCT_VARIABLES:
            variable = f"{variable}_{infarct_qualifier(qualifier_tokens)}"

        updated_variables.append(variable)

    mapped_variables = updated_variables


    anatomy_words = " ".join(t.lower() for t, _ in (anatomy_tokens or []))
    if "edema" in mapped_variables and any(
        w in anatomy_words for w in ("soft tissue", "scalp", "subcutaneous", "orbit", "facial", "prefacial")
    ):
        mapped_variables = ["soft_tissue_lesion" if v == "edema" else v for v in mapped_variables]

    # -------------------------
    # Add severity for selected
    # pathology variables
    # -------------------------

    if not negated:

        # "There is general atrophy." -> Yes NS (not just Yes)
        severity = magnitude_function(
            modifier_tokens,
            allow_age=True
        )

        updated_variables = []

        for variable in mapped_variables:

            if variable in SEVERITY_VARIABLES:

                updated_variables.append(
                    f"{variable}_{severity}"
                )

            else:

                updated_variables.append(
                    variable
                )

        mapped_variables = (
            updated_variables
        )

    direct_variables = {}

    return (
        list(set(mapped_variables)),
        direct_variables
    )
```

#### Outcome 

```python
def outcome_function(
    outcome
):

    # -------------------------
    # Outcome pathology: one item per pathology head,
    # each with the modifiers that were said before it
    # -------------------------

    if "items" in outcome:

        mapped_variables = []

        for item in outcome["items"]:

            item_variables, _ = pathology_function(
                item["pathology"],
                item.get("modifiers", []),
                outcome.get("anatomy") or []
            )

            mapped_variables.extend(item_variables)

        return (
            list(set(mapped_variables)),
            {}
        )

    if "pathology" in outcome:

        result = pathology_function(
            outcome["pathology"],
            outcome.get(
                "modifiers",
                []
            ),
            outcome.get(
                "anatomy",
                []
            )
        )

        return result

    # -------------------------
    # Outcome normal state
    # -------------------------

    elif "normal_state" in outcome:

        result = normal_state_function(
            outcome
        )

        return result

    return (
        [],
        {}
    )
```

#### Normal State

```python
def normal_state_function(
    outcome
):
    anatomy_text = []

    if outcome.get("anatomy"):

        for text, bucket in outcome["anatomy"]:

            anatomy_text.append(
                text
            )

    mapped_variables = (
        normal_state_anatomy_filter(
            anatomy_text
        )
    )

    mapped_variables = [

        f"N_{variable}"

        for variable in mapped_variables

    ]

    direct_variables = {}

    return (
        mapped_variables,
        direct_variables
    )
```

#### Prominence 

```python
def prominence_sorter(
    association
):

    mapped_variables = []
    direct_variables = {}

    prominence_text = association[
        "prominence"
    ][0].lower()

    location_buckets = set()
    laterality = None

    if "location" in association:

        for text, bucket in association["location"]:

            location_buckets.add(
                bucket
            )

            if bucket == "laterality" and laterality is None:
                laterality = LATERALITY_MAP.get(text.lower())

            if bucket == "ventricle" and text.lower() in ("lateral ventricles", "temporal horns"):
                laterality = laterality or "Bilateral"

    severity = magnitude_function(association.get("magnitude", []))

    # ex vacuo dilatation of the ventricles is also ventricular prominence
    if prominence_text == "ex vacuo dilatation":

        mapped_variables.append(
            f"ex_vacuo_dilatation_{severity}"
        )

        location_buckets.add("ventricle")

    if "ventricle" in location_buckets:

        mapped_variables.append(
            f"ventricular_prominence_{severity}"
        )

        if laterality:
            mapped_variables.append(
                f"l_v_prominence_laterality_{laterality}"
            )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Magnitude Mapper

```python
MAGNITUDE_MAP = {

    "mild": "Mild",
    "partial": "Mild",
    "partially": "Mild",
    "mildly": "Mild",
    "minimal": "Mild",
    "some": "Mild",
    "slight": "Mild",
    "slightly": "Mild",
    "subtle": "Mild",
    "trace": "Mild",
    "minor": "Mild",
    "relatively mild": "Mild",
    "mild-early": "Mild",
    "moderate": "Moderate",
    "moderate sized": "Moderate",
    "moderate size": "Moderate",
    "severe": "Severe",
    "marked": "Severe",
    "significant": "Severe",
    "large": "Severe",
    "moderately severe": "Severe",
    "complete": "Severe",
    "near complete": "Severe",

    # ranges keep both ends, like the human sheet ("Moderate/ Severe")
    "moderate to severe": "Moderate/ Severe",
    "mild to moderate": "Mild/ Moderate",

    # "diffuse" / "generalized" say how widespread, not how severe
    "diffuse": "Yes NS",
    "generalized": "Yes NS",
    "global": "Yes NS",

}

AGE_WORDS = {"age appropriate", "age-appropriate", "age-related", "age related", "in keeping with age"}

SEVERITY_RANK = {
    "Yes NS": 0,
    "Mild": 1,
    "Mild/ Moderate": 2,
    "Moderate": 3,
    "Moderate/ Severe": 4,
    "Severe": 5,
}


def magnitude_function(
    modifier_tokens,
    allow_age=False
):
    """
    The STRONGEST grade said wins ("severe generalized" -> Severe,
    not the last word). Words that are not grades are ignored.
    "Age-related / age-appropriate" (atrophy) is only used when no
    real grade was said.
    """

    severity = "Yes NS"
    age_appropriate = False

    for text, bucket in modifier_tokens:

        if bucket != "magnitude":
            continue

        text = text.lower()

        if text in AGE_WORDS:
            age_appropriate = True
            continue

        grade = MAGNITUDE_MAP.get(text)

        if grade and SEVERITY_RANK[grade] > SEVERITY_RANK[severity]:
            severity = grade

    if allow_age and age_appropriate and severity == "Yes NS":
        return "Age Appropriate"

    return severity
```

### SDH Characteristics

```python
def side_of(text):
    """
    "left", "left-sided", "on the left" -> "L"   "bilateral" -> "B"
    "rightward", "to the right" etc are directions of shift, not a side,
    but inside an SDH sentence they still name the side.
    """

    text = text.lower()

    if "bilateral" in text:
        return "B"

    if text in ("left to right", "right to left"):
        return None

    if "left" in text:
        return "L"

    if "right" in text:
        return "R"

    return None


def new_side_record():
    return {
        "cerebral_locations": [],
        "density": [],
        "acuity": [],
        "magnitude": [],
        "morphology": [],
        "thickness": [],
        "density_negated": [],
        "density_dominant": [],
    }


# "predominantly hypo to isodense", "largely isodense", "background of
# low-density": the density words right after these are the dominant ones
DOMINANCE_WORDS = {
    "predominantly", "predominately", "predominant", "largely", "mostly",
    "mainly", "primarily", "chiefly", "background of", "majority of",
}

DOMINANCE_VALUES = {
    frozenset(["hypo_dense"]): "Hypodense",
    frozenset(["iso_dense"]): "Isodense",
    frozenset(["hyper_dense"]): "Hyperdense",
    frozenset(["hypo_dense", "iso_dense"]): "Hypo/Isodense",
    frozenset(["iso_dense", "hyper_dense"]): "Iso/Hyperdense",
}


ATTRIBUTE_BUCKETS = {
    "cerebral_location": "cerebral_locations",
    "density": "density",
    "acuity": "acuity",
    "magnitude": "magnitude",
    "morphology": "morphology",
}

ACUITY_PRIORITY = [
    "acuity_acute_on_chronic",
    "acuity_subacute_on_chronic",
    "acuity_acute_subacute",
    "acuity_acute",
    "acuity_subacute",
    "acuity_chronic",
]

MORPHOLOGY_PRIORITY = ["septations", "dependent", "layering"]


def SDH_characteristics(children):
    """
    Walk the SDH children IN ORDER and give every characteristic to
    the side it was said about.

      - a laterality switches the side ("right ... while the left ...")
      - words said before the first laterality of a sentence are saved
        and given to that laterality ("large mixed density LEFT ...",
        "isodense SDH along RIGHT convexity")
      - with no laterality in the sentence they go to the side of the
        previous sentence
      - "bilateral" gives them to both sides
      - words about another bleed (IPH, SAH) are dropped
      - falx / tentorium locations open a compartment
        (parafalcine_hematoma / tentorium_hematoma)
    """

    sides = {"L": new_side_record(), "R": new_side_record()}

    compartments = {}           # "parafalcine" -> {"sides": set(), ...}

    mapped_variables = []
    direct_variables = {}

    tree_side = None            # side of the previous sentence
    side_before_sentence = None
    previous_sentence_compartment = None
    continuing_compartment = None
    sent_side = None            # side named in this sentence
    sent_compartment = None
    pending = []                # (record_key, value) waiting for a side
    compartment_acuity = []     # (side, acuity) said about the falx / tentorium
    falx_where = []             # "anterior" / "posterior" falx
    unsided = []                # never got a side in their sentence
    blocked = False             # inside a clause about another bleed
    last_laterality = None      # for "left falx", "right tentorium"
    tokens_since_laterality = 99
    dominance_open = False      # after "predominantly": density words are dominant

    sent_shared = []            # said before the first side of the sentence
    sent_sides_named = set()
    sent_bleed_mentions = 0

    sent_log = []               # (target, key, value) given in this sentence
    since_mention = []          # (target, key, value) given after the last bleed word / side

    def give(side, key, value):
        targets = ["L", "R"] if side == "B" else [side]
        for target in targets:
            sides[target][key].append(value)
            sent_log.append((target, key, value))
            if key != "thickness":
                since_mention.append((target, key, value))

    def take_back():
        """
        Words given after the last bleed word belong to the NEXT one:
        "... the left tentorium and there is thin subdural blood along
        the right frontal convexity", "SDH and additional acute left SDH"
        """
        moved = []
        for target, key, value in since_mention:
            if value in sides[target][key]:
                sides[target][key].remove(value)
                moved.append((key, value))
        since_mention.clear()
        return moved

    def flush(side):
        nonlocal pending
        if side:
            for key, value in pending:
                give(side, key, value)
        else:
            unsided.extend(pending)
        pending = []

    def end_sentence():
        nonlocal sent_shared, tree_side
        """
        One SDH named on both sides in one breath
        ("mixed attenuation SDH along right frontoparietal and left
        frontoparietal temporal convexities") -> both sides get the words.
        """
        flush(sent_side or tree_side)

        # "acute SDH involving the superior aspect of the right tentorium":
        # a sentence only about the falx / tentorium describes that bleed,
        # not the convexity SDH of that side - its words go to the
        # compartment (acuity) or nowhere, and the side is not changed
        if sent_compartment and not any(
            key == "cerebral_locations" for _, key, _ in sent_log
        ):
            for target, key, value in sent_log:
                if value in sides[target][key]:
                    sides[target][key].remove(value)
                if key == "acuity":
                    compartments[sent_compartment]["acuity"].append(value)
            tree_side = side_before_sentence

        if {"L", "R"} <= sent_sides_named and sent_bleed_mentions <= 1:
            for key, value in sent_shared:
                if key == "thickness":
                    continue
                for target in ("L", "R"):
                    if value not in sides[target][key]:
                        sides[target][key].append(value)

    def note_compartment(text):
        nonlocal tokens_since_laterality
        if text.lower() in ("falco tentorial", "falcotentorial"):
            note_compartment("tentorium")
            text = "falx"
        name = COMPARTMENT_WORDS.get(text.lower())
        if not name:
            return None
        record = compartments.setdefault(
            name, {"sides": set(), "subregion": None, "thickness": None, "acuity": []}
        )
        # only a laterality said right in front: "left falx", "right side of
        # the falx"; "right tentorium and falx" carries on to the next one
        if last_laterality and tokens_since_laterality <= 1:
            record["sides"].add(last_laterality)
            tokens_since_laterality = 0
        return name

    def take_token(text, bucket, from_measurement=False):
        nonlocal sent_side, tree_side, last_laterality, tokens_since_laterality, sent_compartment
        nonlocal dominance_open

        tokens_since_laterality += 1

        if bucket == "laterality":
            side = side_of(text)
            if (
                side and sent_bleed_mentions and sent_side and side != sent_side
                and "B" not in (side, sent_side) and not from_measurement
            ):
                # "SDH and additional acute LEFT frontoparietal SDH"
                pending.extend(take_back())
            if side:
                since_mention.clear()
                last_laterality = side
                tokens_since_laterality = 0
                if not sent_sides_named:
                    sent_shared.extend(pending)
                sent_side = side
                tree_side = side
                sent_sides_named.add(side)
                flush(side)
            return

        if bucket in ("cerebral_location", "bone"):
            compartment = note_compartment(text)
            if compartment:
                sent_compartment = compartment
                return

        key = ATTRIBUTE_BUCKETS.get(bucket)

        if bucket == "bone":
            key = "cerebral_locations"

        if key is None or blocked:
            return

        if sent_compartment and key == "acuity":
            compartments[sent_compartment]["acuity"].append(text)
            compartment_acuity.append((sent_side or tree_side, text))
            return

        # "predominantly" opens the dominance window; it is not a density itself
        if key == "density" and text.lower() in DOMINANCE_WORDS:
            dominance_open = True
            return

        keys = [key]
        if key == "density" and dominance_open:
            keys.append("density_dominant")

        for key in keys:
            if sent_side:
                give(sent_side, key, text)
            else:
                pending.append((key, text))

    for child in children:

        # the dominance window stays open only over density words and
        # "and" ("predominantly hypo to isodense", "mostly iso and hyperdense")
        if dominance_open and not (
            isinstance(child, tuple) and child[1] in ("density", "sandwich_and")
        ):
            dominance_open = False

        # -------------------------
        # new primary sentence
        # -------------------------
        if isinstance(child, dict) and "sentence_start" in child:
            end_sentence()
            previous_sentence_compartment = (
                sent_compartment
                if sent_compartment and not any(k == "cerebral_locations" for _, k, _ in sent_log)
                else None
            )
            side_before_sentence = tree_side
            # "This hematoma extends to ... the falx" after a tentorial bleed:
            # still the tentorial bleed
            continuing_compartment = previous_sentence_compartment if child.get("anaphor") else None
            sent_side = None
            sent_compartment = None
            blocked = False
            last_laterality = None
            sent_shared = []
            sent_log.clear()
            since_mention.clear()
            sent_sides_named = set()
            sent_bleed_mentions = 0
            continue

        if isinstance(child, dict) and "falx_subregion" in child:
            falx_where.append(child["falx_subregion"])
            continue

        # -------------------------
        # "extension along the right side of the falx" (secondary sentence)
        # -------------------------
        if isinstance(child, dict) and "compartment_location" in child:

            saved = (sent_side, tree_side, last_laterality, tokens_since_laterality)
            for text, bucket in child["compartment_location"]:
                if bucket == "laterality" and side_of(text):
                    last_laterality = side_of(text)
                    tokens_since_laterality = 0
                    continue
                tokens_since_laterality += 1
                note_compartment(text)
            sent_side, tree_side, last_laterality, tokens_since_laterality = saved
            continue

        # -------------------------
        # which bleed is talked about
        # -------------------------
        if isinstance(child, dict) and "bleed_mention" in child:
            if child["bleed_mention"] in ("subdural", "generic"):
                blocked = False
                if sent_bleed_mentions and since_mention:
                    # a second bleed in the sentence: what was said since the
                    # last one waits for this bleed's side
                    pending.extend(take_back())
                    sent_side = None
                since_mention.clear()
                sent_bleed_mentions += 1
                # "Parafalcine hematoma measuring 0.5 cm"
                if "parafalcine" in child.get("text", "").lower():
                    sent_compartment = note_compartment("parafalcine")
            else:
                blocked = True
                pending = []
            continue

        # -------------------------
        # SDH Negation branch
        # -------------------------
        if isinstance(child, dict) and "negation" in child:

            for negation in child["negation"]:

                if "density" in negation:
                    side = sent_side or tree_side
                    if side:
                        for text, bucket in negation["density"]:
                            give(side, "density_negated", text)
                    continue

                negation_variables, negation_direct = (
                    negation_sorter(
                        negation
                    )
                )

                mapped_variables.extend(
                    negation_variables
                )

                direct_variables.update(
                    negation_direct
                )

            continue

        # -------------------------
        # Direct tuples
        # -------------------------
        if isinstance(child, tuple):
            take_token(*child)
            continue

        # -------------------------
        # Location branch
        # -------------------------
        if isinstance(child, dict) and "location" in child:
            for text, bucket in child["location"]:
                take_token(text, bucket)
            continue

        # -------------------------
        # Thickness branch
        # -------------------------
        if isinstance(child, dict) and "thickness" in child:

            thickness = child["thickness"]

            # "maximum thickness left frontal 2.3 cm": the side and place
            # said inside the measurement are SDH words too
            if thickness.get("side") and side_of(thickness["side"]):
                take_token(thickness["side"], "laterality", from_measurement=True)

            side = (
                side_of(thickness["side"]) if thickness.get("side") else None
            ) or sent_side or tree_side

            for text, bucket in thickness.get("locations", []):
                take_token(text, bucket)

            compartment = thickness.get("compartment") or sent_compartment

            if compartment and compartment in compartments:
                if compartments[compartment]["thickness"] is None:
                    compartments[compartment]["thickness"] = thickness["value"]

            # falx / tentorium bleeds are their own variables: their size
            # never goes to L_ / R_ thickness
            if compartment:
                continue

            if side:
                give(side, "thickness", dict(thickness, compartment=compartment))
            else:
                unsided.append(("thickness", dict(thickness, compartment=compartment)))

            continue

    end_sentence()

    # words that never had a side in their sentence go to the side the
    # SDH is named on elsewhere ("there is subdural hematoma predominantly
    # isodense ... over the right frontoparietal convexity")
    named = {k for k, record in sides.items() if any(record.values())}
    if len(named) == 1 and unsided:
        target = next(iter(named))
        for key, value in unsided:
            sides[target][key].append(value)

    # -------------------------
    # Per side: run filters, prefix L_ / R_
    # -------------------------
    for prefix, record in sides.items():

        if not any(record.values()):
            continue

        mapped_variables.append(f"sdh_side_{prefix}")

        sdh_variables = []

        sdh_variables.extend(cerebral_location_filter(record["cerebral_locations"]))
        sdh_variables.extend(density_filter(record["density"]))

        # one acuity: compound ("acute on chronic") beats simple ("acute")
        acuity_words = {a.lower() for a in record["acuity"]}
        density_words = " ".join(d.lower() for d in record["density"])
        has_chronic_part = (
            "chronic" in acuity_words
            or any(w in density_words for w in ("hypo", "low"))
        )

        # "high density ... in keeping with acute hemorrhage" inside an
        # "intermediate to low density" collection, "RECENT COMPONENTS UPON
        # ... LOW-DENSITY CHRONIC COMPONENT" -> acute on chronic
        if has_chronic_part and acuity_words & {"acute", "recent", "new"}:
            record["acuity"] = record["acuity"] + ["acute on chronic"]
        elif "chronic" in acuity_words and "subacute" in acuity_words:
            record["acuity"] = record["acuity"] + ["subacute on chronic"]

        acuity_variables = acuity_filter(record["acuity"])
        for candidate in ACUITY_PRIORITY:
            if candidate in acuity_variables:
                sdh_variables.append(candidate)
                break

        # one description: the first size word said
        magnitude_variables = []
        for text in record["magnitude"]:
            magnitude_variables = magnitude_filter([text])
            if magnitude_variables:
                break
        sdh_variables.extend(magnitude_variables)

        # one morphology: septations > dependent > layering
        morphology_variables = morphology_filter(record["morphology"])
        for candidate in MORPHOLOGY_PRIORITY:
            if candidate in morphology_variables:
                sdh_variables.append(candidate)
                break

        for variable in sdh_variables:
            mapped_variables.append(f"{prefix}_{variable}")

        # density dominance: only when the report says which density dominates
        dominant = frozenset(density_filter(record["density_dominant"])) - {"mixed_density"}
        if dominant in DOMINANCE_VALUES:
            mapped_variables.append(f"{prefix}_density_dominance_{DOMINANCE_VALUES[dominant]}")

        for variable in density_filter(record["density_negated"]):
            mapped_variables.append(f"N_{prefix}_{variable}")

        # -------------------------
        # Thickness: the first measurement of a side wins
        # (later ones are "previously 1.2 cm", impression repeats ...)
        # -------------------------
        if record["thickness"]:

            # a measurement along the falx / tentorium only counts for the
            # side when nothing was measured over the convexity
            convexity = [t for t in record["thickness"] if not t.get("compartment")]
            thickness = (convexity or record["thickness"])[0]
            mm, cm = measurement_to_mm_cm(thickness["value"])

            if mm is not None:

                direct_variables[f"{prefix}_thickness_mm"] = mm
                direct_variables[f"{prefix}_thickness_cm"] = cm

                if thickness["max"]:
                    direct_variables[f"{prefix}_thickness_max"] = "Yes"

    # -------------------------
    # Compartments (falx / tentorium)
    # -------------------------
    side_names = {"L": "Left", "R": "Right", "B": "Bilateral"}

    for name, record in compartments.items():

        named = record["sides"]

        if "B" in named or ("L" in named and "R" in named):
            value = "Bilateral"
        elif named:
            value = side_names[next(iter(named))]
        else:
            value = "Yes NS"

        mapped_variables.append(f"{name}_hematoma_{value}")

        if name == "parafalcine" and falx_where:
            mapped_variables.append(f"parafalcine_location_{falx_where[0]}")

        if record["thickness"]:
            mm, cm = measurement_to_mm_cm(record["thickness"])
            if mm is not None:
                direct_variables[f"{name}_thickness_mm"] = mm
                direct_variables[f"{name}_thickness_cm"] = cm

        acuity_variables = acuity_filter(record["acuity"])
        for candidate in ACUITY_PRIORITY:
            if candidate in acuity_variables:
                mapped_variables.append(f"{name}_{candidate}")
                break

    return (
        mapped_variables,
        direct_variables
    )


def measurement_to_mm_cm(value_text):

    value_text = value_text.lower().replace(" ", "")

    try:
        if value_text.endswith("cm"):
            cm = float(value_text[:-2])
            return round(cm * 10, 2), cm

        if value_text.endswith("mm"):
            mm = float(value_text[:-2])
            return mm, round(mm / 10, 3)

    except ValueError:
        pass

    return None, None
```

```python
ARRAY_DEFAULTS = {

    # Overall
    "laterality": "NR",

    # L Description
    "L_description": "NR",

    # L Morphology
    "L_morphology": "NR",

    # L Acuity
    "L_acuity": "NR",

    # L Density
    "L_mixed_density": "NR",
    "L_hyper_dense": "NR",
    "L_hypo_dense": "NR",
    "L_iso_dense": "NR",
    "L_density_dominance": "NR",

    # L Thickness
    "L_thickness_mm": "NR",
    "L_thickness_cm": "NR",
    "L_thickness_max": "NR",

    # L Locations
    "L_frontal_lobe": "NR",
    "L_parietal_lobe": "NR",
    "L_temporal_lobe": "NR",
    "L_occipital_lobe": "NR",
    "L_convexity_hematoma": "NR",
    "L_holohemispheric_hematoma": "NR",
    "L_vertex": "NR",

    # R Description
    "R_description": "NR",

    # R Morphology
    "R_morphology": "NR",

    # R Acuity
    "R_acuity": "NR",

    # R Density
    "R_mixed_density": "NR",
    "R_hyper_dense": "NR",
    "R_hypo_dense": "NR",
    "R_iso_dense": "NR",
    "R_density_dominance": "NR",

    # R Thickness
    "R_thickness_mm": "NR",
    "R_thickness_cm": "NR",
    "R_thickness_max": "NR",

    # R Locations
    "R_frontal_lobe": "NR",
    "R_parietal_lobe": "NR",
    "R_temporal_lobe": "NR",
    "R_occipital_lobe": "NR",
    "R_convexity_hematoma": "NR",
    "R_holohemispheric_hematoma": "NR",
    "R_vertex": "NR",

    # Falx / tentorium compartments
    "parafalcine_hematoma": "NR",
    "parafalcine_location": "NR",
    "parafalcine_thickness_mm": "NR",
    "parafalcine_thickness_cm": "NR",
    "parafalcine_acuity": "NR",
    "tentorium_hematoma": "NR",
    "tentorium_thickness_mm": "NR",
    "tentorium_thickness_cm": "NR",
    "tentorium_acuity": "NR",

    # Mass Effect
    "mass_effect_present": "NR",

    # Midline Shift
    "midline_shift_present": "NR",
    "midline_shift_mm": "NR",
    "midline_shift_cm": "NR",
    "midline_shift_direction": "NR",

    # Sulcal Effacement
    "effacement_sulci": "NR",
    "effacement_sulci_laterality": "NR",

    # Ventricular Effacement
    "effacement_ventricular_system": "NR",
    "effacement_lateral_ventricle": "NR",
    "l_ventricle_laterality": "NR",
    "effacement_third_ventricle": "NR",
    "effacement_forth_ventricle": "NR",

    # Ventricular Prominence
    "ventricular_prominence": "NR",
    "l_v_prominence_laterality": "NR",
    "ex_vacuo_dilatation": "NR",

    # Herniation
    "herniation_subfalcine": "NR",
    "herniation_uncal": "NR",
    "herniation_transtentorial": "NR",
    "herniation_tonsillar": "NR",

    # Other Pathologies
    "encephalomalacia": "NR",
    "hydrocephalus": "NR",
    "microangiopathy": "NR",
    "cerebral_atrophy": "NR",
    "white_dark_differentiation": "NR",
    "mass": "NR",
    "posterior_fossa_abnormality": "NR",
    "unspecified_infarct": "NR",
    "territorial_infarct": "NR",
    "lacunar_infarct": "NR",
    "cortical_infarct": "NR",
    "territorial_ischemic": "NR",
    "hyperdense_vessel": "NR",

    # Trauma
    "soft_tissue_lesion": "NR",
    "scalp_hematoma": "NR",
    "contusion": "NR",
    "edema": "NR",
    "aneurysm": "NR",

    # Other hemorrhage
    "subarachnoid_hemorrhage": "NR",
    "intraparenchymal_hemorrhage": "NR",
    "intraventricular_hemorrhage": "NR",
    "epidural_hemorrhage": "NR",

    # Fracture
    "orbital_lesion": "NR",
    "bony_lesion": "NR",
    "fracture_location": "NR",
    "fracture_type": "NR",

    # Sinuses / mastoids
    "paranasal_sinuses": "NR",
    "maxillary_sinus": "NR",
    "sphenoid_sinus": "NR",
    "ethmoid_sinus": "NR",
    "mastoid_air_cells": "NR",

}

# variables written as "<prefix><value>" e.g. "mass_effect_present_Mild"
# (longest prefixes first so "effacement_sulci_laterality_" wins over
#  "effacement_sulci_")
VALUE_PREFIXES = [
    "effacement_sulci_laterality_",
    "effacement_sulci_",
    "effacement_ventricular_system_",
    "effacement_lateral_ventricle_",
    "effacement_third_ventricle_",
    "effacement_forth_ventricle_",
    "l_ventricle_laterality_",
    "l_v_prominence_laterality_",
    "mass_effect_present_",
    "midline_shift_present_",
    "herniation_subfalcine_",
    "herniation_uncal_",
    "herniation_transtentorial_",
    "herniation_tonsillar_",
    "cerebral_atrophy_",
    "microangiopathy_",
    "hydrocephalus_",
    "parafalcine_hematoma_",
    "parafalcine_location_",
    "tentorium_hematoma_",
    "ventricular_prominence_",
    "ex_vacuo_dilatation_",
    "subarachnoid_hemorrhage_",
]

INFARCT_KEYS = [
    "territorial_ischemic",
    "unspecified_infarct",
    "territorial_infarct",
    "lacunar_infarct",
    "cortical_infarct",
]

ACUITY_VALUES = {
    "acuity_acute": "Acute",
    "acuity_subacute": "Subacute",
    "acuity_chronic": "Chronic",
    "acuity_acute_on_chronic": "Acute on Chronic",
    "acuity_subacute_on_chronic": "Subacute on Chronic",
    "acuity_acute_subacute": "Acute/Subacute",
}

MORPHOLOGY_VALUES = {
    "dependent": "Dependent",
    "layering": "Layering",
    "septations": "Septations",
}

DESCRIPTION_VALUES = {
    "description_minimal": "Minimal",
    "description_small": "Small",
    "description_moderate": "Medium",
    "description_large": "Large",
}


def value_rank(value):
    """
    How much a value says. A later mention only overwrites an earlier
    one if it says more:  NR < No... < Yes / Yes NS < a grade / a value
    ("Cerebral atrophy." in the impression must not wipe out "Severe"
    from the findings). Equal rank -> the first one said is kept.
    """

    if value is None or value == "NR":
        return 0

    if isinstance(value, str) and value.startswith("No"):
        return 1

    if value in ("Yes", "Yes NS"):
        return 2

    return 3


FRACTURE_LOCATION_ORDER = ["Calvarial", "Skull base", "Facial"]
FRACTURE_TYPE_ORDER = ["Nondisplaced", "Displaced", "Depressed", "Comminuted"]


def put(final_variables, key, value):

    old = final_variables.get(key, "NR")

    # fractures in different sentences add up: "Calvarial" + "Facial"
    if key == "fracture_location" and value_rank(old) == 3 and value_rank(value) == 3:
        parts = set(old.split(" + ")) | set(value.split(" + "))
        final_variables[key] = " + ".join(p for p in FRACTURE_LOCATION_ORDER if p in parts)
        return

    # the most severe fracture type is kept
    if key == "fracture_type" and old in FRACTURE_TYPE_ORDER and value in FRACTURE_TYPE_ORDER:
        if FRACTURE_TYPE_ORDER.index(value) > FRACTURE_TYPE_ORDER.index(old):
            final_variables[key] = value
        return

    if value_rank(value) > value_rank(old):
        final_variables[key] = value


def final_array_function(
    mapped_variables,
    direct_variables,
    final_variables
):

    # -------------------------
    # Populate defaults
    # -------------------------

    for key, value in ARRAY_DEFAULTS.items():

        if key not in final_variables:

            final_variables[key] = value

    # -------------------------
    # Apply mapped variables
    # -------------------------

    for variable in mapped_variables:

        negated = variable.startswith("N_")
        target = variable[2:] if negated else variable

        # -------------------------
        # Which sides have an SDH (for "laterality")
        # -------------------------

        if variable == "sdh_side_L":
            final_variables["_sdh_left"] = True
            continue

        if variable == "sdh_side_R":
            final_variables["_sdh_right"] = True
            continue

        # -------------------------
        # White / gray differentiation
        # -------------------------

        if target in ("white_dark_differentiation", "loss_white_dark_differentiation"):

            if negated or target == "white_dark_differentiation":
                put(final_variables, "white_dark_differentiation", "Preserved")
            else:
                put(final_variables, "white_dark_differentiation", "Loss")

            continue

        # -------------------------
        # "No brain herniation" -> every subtype
        # -------------------------

        if target.startswith("herniation_generic_"):

            if negated:
                for key in ("herniation_subfalcine", "herniation_uncal",
                            "herniation_transtentorial", "herniation_tonsillar"):
                    put(final_variables, key, "No")

            continue

        # -------------------------
        # Infarcts: "No Acute", "No NS", "Yes Chronic", ...
        # -------------------------

        infarct_key = next(
            (key for key in INFARCT_KEYS if target.startswith(key + "_")),
            None
        )

        if infarct_key:

            qualifier = target[len(infarct_key) + 1:]
            put(
                final_variables,
                infarct_key,
                f"{'No' if negated else 'Yes'} {qualifier}"
            )
            continue

        # -------------------------
        # L_ / R_ acuity, morphology, description
        # -------------------------

        if target[:2] in ("L_", "R_") and not negated:

            prefix = target[:2]
            rest = target[2:]

            if rest in ACUITY_VALUES:
                put(final_variables, f"{prefix}acuity", ACUITY_VALUES[rest])
                continue

            if rest in MORPHOLOGY_VALUES:
                put(final_variables, f"{prefix}morphology", MORPHOLOGY_VALUES[rest])
                continue

            if rest in DESCRIPTION_VALUES:
                put(final_variables, f"{prefix}description", DESCRIPTION_VALUES[rest])
                continue

            if rest.startswith("density_dominance_"):
                put(final_variables, f"{prefix}density_dominance", rest[len("density_dominance_"):])
                continue

        # compartment acuity ("parafalcine_acuity_acute")
        if (
            target.startswith(("parafalcine_acuity_", "tentorium_acuity_"))
            and not negated
        ):
            key = "parafalcine_acuity" if target.startswith("parafalcine") else "tentorium_acuity"
            rest = target[len(key) + 1:]
            put(final_variables, key, ACUITY_VALUES.get(rest, ACUITY_VALUES.get("acuity_" + rest, rest)))
            continue

        # -------------------------
        # Simple Yes / No variables
        # -------------------------

        if target in ARRAY_DEFAULTS:

            put(final_variables, target, "No" if negated else "Yes")
            continue

        # -------------------------
        # "<prefix><value>" variables
        # -------------------------

        prefix = next(
            (p for p in VALUE_PREFIXES if target.startswith(p)),
            None
        )

        if prefix:

            key = prefix[:-1]
            value = target[len(prefix):]

            if negated:
                # a negated laterality says nothing
                if "laterality" not in key:
                    put(final_variables, key, "No")
            else:
                put(final_variables, key, value)

            continue

    # -------------------------
    # Apply direct variables (first value said wins)
    # -------------------------

    for key, value in direct_variables.items():

        put(final_variables, key, value)

    return final_variables
```

## BLEED TREE FUNCTION

```python
MEASUREMENT_BUCKETS = (
    "numerical",
    "measurement_linker",
    "measurement_linker_max",
    "measurement_max",
    "measurement_type"
)

CLAUSE_STOP = {
    "period", "comma", "semicolon", "colon",
    "association_linker", "stop_and",
}

# anatomy that means a density / morphology word is NOT about the SDH
NON_SDH_ANATOMY = {
    "cistern", "ventricle", "intraparenchymal", "extracranial_anatomy",
    "bone", "sulci", "tonsils", "sinus",
}

ANAPHOR_WORDS = {"it", "this", "these"}


def number_belongs_to_effect(sent_tokens, i):
    """
    Is the number / measurement at i describing an effect
    ("2 cm midline shift", "midline shift measuring 15 mm")?
    Looks at the clause the number sits in, not a fixed window.
    """

    # back to the start of the clause (max 4 tokens)
    k = i - 1
    steps = 0
    while k >= 0 and steps < 4:
        bucket = sent_tokens[k][1]
        if bucket in CLAUSE_STOP:
            break
        if bucket in EFFECT_FAMILY:
            return True
        if bucket == "bleed_pathology":
            return False
        k -= 1
        steps += 1

    # forward to the end of the clause
    k = i + 1
    while k < len(sent_tokens):
        bucket = sent_tokens[k][1]
        if bucket in CLAUSE_STOP:
            break
        if bucket in EFFECT_FAMILY:
            return True
        if bucket in ("bleed_pathology", "bleed_reference"):
            return False
        k += 1

    return False


def laterality_belongs_to_effect(sent_tokens, i):

    k = i + 1

    while k < len(sent_tokens) and sent_tokens[k][1] in (
        "magnitude", "numerical", "measurement_unit", "uncertainty_linker"
    ):
        k += 1

    return k < len(sent_tokens) and sent_tokens[k][1] in EFFECT_FAMILY


def secondary_thickness(sent_tokens):
    """
    Thickness in a sentence without a bleed word is only about the SDH
    when the sentence points back to it ("It measures 8 mm",
    "This has a maximal depth of 1 cm", "The hematoma measures ...",
    "Maximal thickness 14 mm").
    """

    words = {text.lower() for text, _ in sent_tokens}

    points_back = (
        bool(words & ANAPHOR_WORDS)
        or sentence_has(sent_tokens, {"bleed_reference", "measurement_max"})
        or bool(words & {"thickness", "depth", "width", "thick", "thickest"})
    )

    if not points_back or sentence_has(sent_tokens, {"ignore"}):
        return []

    results = []
    i = 0

    while i < len(sent_tokens):

        bucket = sent_tokens[i][1]

        if bucket in MEASUREMENT_BUCKETS and not number_belongs_to_effect(sent_tokens, i):

            result, next_i = parse_thickness_branch(sent_tokens, i)

            if result:
                said_after_now = any(
                    t.lower() == "now" for t, _ in sent_tokens[max(0, i - 4):i]
                )
                results.append({"thickness": result, "now": said_after_now})

            i = max(next_i, i + 1)
            continue

        i += 1

    # "It was 14 mm and is now 15 mm thick": keep the current size
    if any(r["now"] for r in results):
        results = [r for r in results if r["now"]]

    return [{"thickness": r["thickness"]} for r in results]


def respectively_pairs(sent_tokens):

    words = [t.lower() for t, _ in sent_tokens]

    if "respectively" not in words:
        return []

    sides = [t for t, b in sent_tokens if b == "laterality" and side_of(t) in ("L", "R")]

    sizes = []
    for k, (t, b) in enumerate(sent_tokens):
        if b == "numerical":
            for t2, b2 in sent_tokens[k + 1:k + 4]:
                if b2 == "measurement_unit":
                    sizes.append(f"{t}{t2.lower()}")
                    break

    if len(sides) < 2 or len(sides) != len(sizes):
        return []

    is_max = any(b in ("measurement_max",) for _, b in sent_tokens)

    return [
        {"thickness": {"value": size, "max": is_max, "side": side,
                       "locations": [], "compartment": None}}
        for side, size in zip(sides, sizes)
    ]


EXTENSION_WORDS = {"extension", "extends", "extending", "tracking", "layering", "outlines"}


def compartment_mentions(sent_tokens):
    """
    Laterality + falx / tentorium words in a sentence that talks about
    the SDH extending there ("minor extension along the right side of
    the falx cerebri and right tentorium cerebelli").
    """

    words = {t.lower() for t, _ in sent_tokens}

    if not words & EXTENSION_WORDS:
        return []

    found = [
        (t, b) for t, b in sent_tokens
        if b == "laterality" or (b == "cerebral_location" and t.lower() in COMPARTMENT_WORDS)
    ]

    if not any(b == "cerebral_location" for _, b in found):
        return []

    return found


def falx_subregion(sent_tokens):
    """
    "along the anterior falx", "the falx posteriorly", "posterior aspect of
    the falx" -> parafalcine_location
    """

    text = " ".join(t.lower() for t, _ in sent_tokens)
    match = re.search(
        r"(anterior|posterior)\w*\s+(aspect\s+of\s+the\s+|part\s+of\s+the\s+)?(falx|falcine|interhemispheric)"
        r"|(falx|falcine|interhemispheric)\w*(\s+\w+){0,2}\s+(anterior|posterior)",
        text,
    )

    if not match:
        return []

    where = match.group(1) or match.group(6)

    return [{"falx_subregion": where.capitalize()}]


def build_bleed_tree(sent_tokens, is_bleed_sentence, existing_tree=None):

    # ============================
    # Initialize / reuse tree
    # ============================
    if existing_tree is not None:
        tree = existing_tree
    else:
        tree = {"pathology": None, "children": []}

        for text, bucket in sent_tokens:
            if bucket == "bleed_pathology":
                tree["pathology"] = text
                break

        if not tree["pathology"]:
            return None

    # -----------------------------------
    # SECONDARY SENTENCE
    #   only the association (effects, density, ...) and an
    #   anaphoric thickness. Locations / lateralities in these
    #   sentences describe the effect, not the SDH.
    # -----------------------------------
    if not is_bleed_sentence:

        assoc, _ = parse_association(sent_tokens, 0)

        # density / acuity / morphology words next to other anatomy
        # ("hyperdensity within the basal cisterns") are not the SDH
        words = {t.lower() for t, _ in sent_tokens}
        points_back = (
            sentence_has(sent_tokens, {"bleed_reference"})
            or bool(words & ANAPHOR_WORDS)
            or bool(words & {"component", "components", "blood", "compartment", "hematoma"})
        )

        # "Hypoattenuation within the left frontal lobe", "hypodense nodules in
        # both thyroid lobes", "hyperdensity within the basal cisterns"
        if (
            (sentence_has(sent_tokens, NON_SDH_ANATOMY | {"cerebral_location", "ignore"})
             or not points_back)
            and not points_back
        ):
            assoc = [
                a for a in assoc
                if not any(k in a for k in ("density", "acuity", "morphology"))
            ]

        if assoc:
            tree["children"].append({"association": assoc})

        tree["children"].extend(
            secondary_thickness(sent_tokens)
        )

        tree["children"].extend(falx_subregion(sent_tokens))

        # "Mild extension along the anterior falx" -> compartment only
        compartment_tokens = compartment_mentions(sent_tokens)
        if compartment_tokens:
            tree["children"].append(
                {"compartment_location": compartment_tokens}
            )

        return tree

    # -----------------------------------
    # PRIMARY SENTENCE
    # -----------------------------------
    tree["children"].append({"sentence_start": True})

    # "right frontoparietal and left ... convexities, measuring 1.2 cm and
    # 2.3 cm in thickness respectively" -> pair sides and sizes in order
    # (added after the sentence is read, so the sides are said first)
    paired = respectively_pairs(sent_tokens)

    i = 0
    while i < len(sent_tokens):

        text, bucket = sent_tokens[i]

        # -----------------------------------
        # which bleed is being talked about now
        # -----------------------------------
        if bucket == "bleed_pathology":

            tree["children"].append(
                {
                    "bleed_mention": bleed_type(text, sent_tokens, i),
                    "text": text
                }
            )

            i += 1
            continue

        if bucket == "negation":

            branches, next_i = parse_negation_branch(
                sent_tokens,
                i
            )

            if branches:
                tree["children"].append(
                    {"negation": branches}
                )

            i = max(next_i, i + 1)
            continue

        # -----------------------------------
        # ASSOCIATION (primary bleed sentence)
        # -----------------------------------
        if bucket == "association_linker":

            next_bucket = sent_tokens[i+1][1] if i+1 < len(sent_tokens) else None

            # "with 5 mm rightward midline shift" / "with left-sided midline shift"
            opens_numeric_effect = (
                next_bucket in ("numerical", "laterality")
                and (
                    number_belongs_to_effect(sent_tokens, i + 1)
                    if next_bucket == "numerical"
                    else laterality_belongs_to_effect(sent_tokens, i + 1)
                )
            )

            # "with mild mass effect" opens an association, but
            # "with the large intermediate density right SDH" does not:
            # the grade has to lead to an effect
            k = i + 1
            while k < len(sent_tokens) and sent_tokens[k][1] == "magnitude":
                k += 1
            grade_leads_to_effect = (
                next_bucket == "magnitude"
                and k < len(sent_tokens)
                and sent_tokens[k][1] in EFFECT_FAMILY
            )

            # Open association ONLY if next bucket is magnitude or mass effect
            if (
                grade_leads_to_effect
                or next_bucket in EFFECT_FAMILY
                or opens_numeric_effect
            ):
                assoc, next_i = parse_association(sent_tokens, i)
                if assoc:
                    tree["children"].append({"association": assoc})
                i = next_i
                continue

            # Otherwise → SKIP association_linker
            i += 1
            continue

        # -----------------------------------
        # DIRECT ATTACH (after association logic)
        # -----------------------------------
        if bucket in DIRECT_ATTACH:

            # "small volume of ... blood", "small amount of", "small component":
            # the size of a part, not of the whole SDH
            if bucket == "magnitude" and " ".join(
                t.lower() for t, _ in sent_tokens[i + 1:i + 3]
            ) in ("volume of", "amount of", "component", "components", "foci of", "areas of"):
                i += 1
                continue

            # "with left-sided midline shift": a side in front of an effect
            # is the effect's direction, not the side of the SDH
            if bucket == "laterality" and laterality_belongs_to_effect(sent_tokens, i):
                i += 1
                continue

            tree["children"].append((text, bucket))

            # "acute blood components" / "acute hemorrhagic components"
            # are the hyperdense part of the SDH
            following = " ".join(
                t.lower() for t, _ in sent_tokens[i + 1:i + 3]
            )
            if (
                bucket == "acuity"
                and text.lower() == "acute"
                and following.startswith(("component", "blood", "hemorrhagic component"))
            ):
                tree["children"].append(("hyperdense", "density"))

            i += 1
            continue

        # -----------------------------------
        # THICKNESS
        # -----------------------------------
        if bucket in MEASUREMENT_BUCKETS:

            if paired:
                i += 1
                continue

            if not number_belongs_to_effect(sent_tokens, i):

                result, next_i = parse_thickness_branch(
                    sent_tokens,
                    i
                )

                if result:
                    tree["children"].append(
                        {"thickness": result}
                    )

                i = max(next_i, i + 1)
                continue

        # -----------------------------------
        # LOCATION BRANCH
        # -----------------------------------
        if bucket == "location_linker":
            branch, next_i = parse_location_branch(sent_tokens, i)
            if branch:
                tree["children"].append({"location": branch})
            i = next_i
            continue

        # -----------------------------------
        # Direct Mass Effect Opener
        # -----------------------------------
        if bucket in EFFECT_FAMILY:

            assoc, next_i = parse_association(
                sent_tokens,
                i
            )

            if assoc:
                tree["children"].append(
                    {"association": assoc}
                )

            i = next_i
            continue

        # -----------------------------------
        # DEFAULT
        # -----------------------------------
        i += 1

    tree["children"].extend(paired)
    tree["children"].extend(falx_subregion(sent_tokens))

    return tree
```

#### Thickness Helper

```python
def get_measurement_branch(
    tokens,
    start_i
):

    branch = []
    i = start_i

    while i < len(tokens):

        _, bucket = tokens[i]

        if bucket in (
            "association_linker",
            "stop_and",
            "period",
            "comma",
            "semicolon",
        ):
            break

        branch.append(
            tokens[i]
        )

        i += 1

    return branch
```

```python
def measurement_branch_contains_effect(
    branch
):

    return any(
        bucket in EFFECT_FAMILY
        for _, bucket in branch
    )
```

### AND function

```python
def classify_and(sent_tokens, idx):

    if idx <= 0 or idx >= len(sent_tokens) - 1:
        return "unknown_and"

    left_text, left_bucket = sent_tokens[idx - 1]
    right_text, right_bucket = sent_tokens[idx + 1]

    # -----------------------------
    # 1. Paired measurements
    # -----------------------------
    if "respectively" in right_text.lower():
        return "paired_and"

    # -----------------------------
    # 1.5 Directional laterality
    # -----------------------------
    if right_text.lower() in (
        "rightward",
        "leftward"
    ):
        return "stop_and"

    # -----------------------------
    # 2. Anatomy families → sandwich_and
    # -----------------------------
    ANATOMY_FAMILY = {
        "intraparenchymal",
        "ventricle",
        "cistern",
        "cerebral_location",
        "extracranial_anatomy",
        "subregional_anatomy",
        "laterality",
        "sulci",
        "sinus",
        "bone",
    }

    if left_bucket in ANATOMY_FAMILY and right_bucket in ANATOMY_FAMILY:
        return "sandwich_and"

    # -----------------------------
    # 3. Pathology families → stop_and
    # -----------------------------
    PATHOLOGY_FAMILY = {
        "bleed_pathology",
        "mass_effect",
        "effacement",
        "compression",
        "midline_shift",
        "pathology",
        "bone_pathology",
    }

    if left_bucket in PATHOLOGY_FAMILY and right_bucket in PATHOLOGY_FAMILY:
        return "stop_and"

    # -----------------------------
    # 4. Mixed categories → stop_and
    # -----------------------------
    if left_bucket != right_bucket:
        return "stop_and"

    # -----------------------------
    # 5. Same bucket → sandwich_and
    # -----------------------------
    return "sandwich_and"
```

### Thickness Parser

```python
COMPARTMENT_WORDS = {
    "para-falcine": "parafalcine",
    "falcotentorial": "parafalcine",
    "falco tentorial": "parafalcine",
    "interhemispheric fissure": "parafalcine",
    "falx": "parafalcine",
    "falx cerebri": "parafalcine",
    "parafalcine": "parafalcine",
    "falcine": "parafalcine",
    "interhemispheric": "parafalcine",
    "tentorium": "tentorium",
    "tentorium cerebelli": "tentorium",
    "tentorial": "tentorium",
}


def clause_laterality_before(tokens, start_i):
    """
    Laterality said earlier in the same clause
    ("On the right this measures up to 1.4 cm").
    """

    k = start_i - 1

    while k >= 0:

        text, bucket = tokens[k]

        # a comma right after an opening "On the left," does not end the clause
        if bucket == "comma" and k <= 4 and any(b == "laterality" for _, b in tokens[:k]):
            k -= 1
            continue

        if bucket in ("period", "comma", "semicolon", "colon", "stop_and"):
            break

        if bucket in ("numerical",):
            break

        if bucket == "laterality":
            return text

        k -= 1

    return None


def parse_thickness_branch(
    tokens,
    start_i
):
    """
    Save tokens from a measurement word until the clause ends.
    Returns ({"value": "2.3cm", "max": bool, "side": "left",
              "locations": [...], "compartment": "tentorium"}, next_i)

    - stops at a SECOND number, so "2.5 cm ... left frontal region and
      4 mm ... right frontal region" gives two measurements
    - "11.0 x 2.5 cm" keeps the smallest dimension (the thickness)
    - laterality / location words met inside the clause tell which side
      and where the measurement was taken
    """

    has_thickness_context = any(
        bucket in (
            "measurement_linker_max",
            "measurement_max",
            "measurement_type",
        )
        for _, bucket in tokens
    )

    # "Right subdural, 1.0 cm"
    if not has_thickness_context and tokens[start_i][1] == "numerical":
        before = tokens[max(0, start_i - 3):start_i]
        has_thickness_context = (
            any(b in ("bleed_pathology", "bleed_reference") for _, b in before)
            and start_i + 1 < len(tokens)
            and tokens[start_i + 1][1] == "measurement_unit"
        )

    # "a 16 mm left parafalcine subdural hematoma" / "a hypodense 9 mm collection"
    if not has_thickness_context and tokens[start_i][1] == "numerical":

        window = tokens[start_i + 1: start_i + 6]

        has_thickness_context = (
            any(b == "measurement_unit" for _, b in window[:1])
            and any(b in ("bleed_pathology", "bleed_reference") for _, b in window)
        )

    if not has_thickness_context:

        return None, start_i+1

    i = start_i
    numbers = []
    unit = None
    is_max = False
    side = None
    locations = []

    STOP_BUCKETS = {
        "mass_effect",
        "herniation_subtype",
        "association_linker",
        "comma",
        "period",
        "semicolon",
        "colon",
        "stop_and",
        "negation",
    }

    SKIP_BUCKETS = {
        "noticing",
        "location_linker",
        "measurement_type",
        "measurement_linker",
        "uncertainty_linker",
        "subregional_anatomy",
        "magnitude",
        "bleed_pathology",
        "bleed_reference",
        "intraparenchymal",
        "extracranial_anatomy",
        "bone",
        "sandwich_and",
        "acuity",
        "density",
        None,
    }

    while i < len(tokens):
        text, bucket = tokens[i]

        if bucket in STOP_BUCKETS:
            break

        if bucket == "numerical":

            previous = tokens[i - 1][0].lower() if i > 0 else ""

            # second number that is not "A x B" -> next measurement
            if numbers and unit is not None and previous != "x":
                break

            if numbers and previous != "x" and unit is None:
                numbers = []

            numbers.append(text)
            i += 1
            continue

        if bucket == "measurement_unit":
            if numbers and unit is None:
                unit = text.lower()
            i += 1
            continue

        if bucket in ("measurement_max", "measurement_linker_max"):
            is_max = True
            i += 1
            continue

        if bucket == "laterality":
            if side is None:
                side = text
            i += 1
            continue

        if bucket == "cerebral_location":
            locations.append((text, bucket))
            i += 1
            continue

        if bucket in SKIP_BUCKETS or text.lower() == "x":
            i += 1
            continue

        break

    if not numbers or not unit:
        return None, i

    try:
        value_number = min(float(n) for n in numbers)
    except ValueError:
        return None, i

    if side is None:
        side = clause_laterality_before(tokens, start_i)

    compartment = None
    for text, bucket in locations:
        compartment = COMPARTMENT_WORDS.get(text.lower(), compartment)

    result = {
        "value": f"{value_number:g}{unit}",
        "max": is_max,
        "side": side,
        "locations": locations,
        "compartment": compartment,
    }

    return result, i
```

### Mass Effect Parser

```python
def parse_numeric_mass_effect(tokens, start_i):

    i = start_i

    value = None
    unit = None
    modifier = None
    direction = None

    # ============================
    # Step 1: numeric value
    # ============================
    if tokens[i][1] == "numerical":
        value = tokens[i][0]
        i += 1
    else:
        return None, start_i

    # ============================
    # Step 2: measurement unit
    # ============================
    if i < len(tokens) and tokens[i][1] == "measurement_unit":
        unit = tokens[i][0]
        i += 1

    # ============================
    # Step 3: optional modifier
    # ============================
    if i < len(tokens) and tokens[i][1] == "modifier":
        modifier = tokens[i][0]
        i += 1

    # ============================
    # Step 4: scan forward until STOP
    # ============================
    while i < len(tokens):
        text, bucket = tokens[i]

        # STOP CONDITIONS
        if bucket in MASS_NUM_STOP:
            break

        # Laterality (direction)
        if bucket == "laterality":
            direction = text
            i += 1
            continue

        # Skip harmless joiners
        if bucket in ("sandwich_and",):
            i += 1
            continue

        # Unknown → STOP
        break

    # ============================
    # Return clean measurement block
    # ============================
    measurement = {
        "value": value,
        "unit": unit,
        "modifier": modifier,
        "direction": direction
    }

    return measurement, i
```

### Negation Parser

```python
def parse_negation_branch(tokens, start_i):
    """
    Read a negated LIST, the way a reader does:

        "No evidence of acute infarct, mass or hydrocephalus"
        "No acute parenchymal or subarachnoid hemorrhage"
        "no midline shift or brain herniation"

    Every head (pathology / bleed / effect / density) after the negation
    becomes its own negated branch. Tokens are saved until a head is hit:
      - magnitude / laterality / territorial words  -> the NEXT head only
      - acuity ("acute", "recent")                   -> every head in the list
      - anatomy before a head                        -> that head
      - anatomy after a head (via "of the", "in the") -> that head
    The list ends at a period / semicolon / colon or at "but".
    """

    branches = []
    pending_modifiers = []
    pending_anatomy = []
    list_acuity = []
    current = None

    i = start_i + 1

    EFFECTS = {
        "mass_effect",
        "effacement",
        "compression",
        "herniation_subtype"
    }

    STOP_BUCKETS = {
        "period",
        "semicolon",
        "colon",
        "title",
    }

    STOP_WORDS = {"but", "however", "although", "whereas"}

    ANATOMY = {
        "extracranial_anatomy", "bone", "sinus", "intraparenchymal",
        "sulci", "ventricle", "cistern", "cerebral_location", "tonsils",
    }

    PASS_THROUGH = {
        None, "comma", "logical_connector", "sandwich_and", "stop_and",
        "noticing", "association_linker", "uncertainty_linker",
        "subregional_anatomy", "negation", "normal_state",
        "ignore", "spine",
    }

    def open_head(head):
        head["modifiers"] = pending_modifiers + list_acuity
        head["anatomy"] = list(pending_anatomy)
        pending_modifiers.clear()
        pending_anatomy.clear()
        branches.append(head)
        return head

    while i < len(tokens):

        text, bucket = tokens[i]

        if bucket in STOP_BUCKETS or text.lower() in STOP_WORDS:
            break

        # -----------------------------
        # anatomy AFTER a head ("fractures of the calvarium")
        # -----------------------------
        if bucket == "location_linker":

            k = i + 1
            found = []

            while k < len(tokens):

                t_text, t_bucket = tokens[k]

                if t_bucket in ANATOMY or t_bucket == "laterality":
                    found.append((t_text, t_bucket))
                    k += 1
                    continue

                if t_bucket in (None, "subregional_anatomy", "location_linker", "sandwich_and"):
                    k += 1
                    continue

                break

            if current is not None:
                if "effect_name" in current:
                    current.setdefault("location", []).extend(found)
                else:
                    current["anatomy"].extend(found)
            else:
                pending_anatomy.extend(found)

            i = k
            continue

        # -----------------------------
        # saved modifiers
        # -----------------------------
        if bucket == "acuity":
            list_acuity.append((text, bucket))
            i += 1
            continue

        if bucket in ("magnitude", "laterality", "territorial", "cortical"):
            pending_modifiers.append((text, bucket))
            i += 1
            continue

        if bucket in ANATOMY:
            pending_anatomy.append((text, bucket))
            i += 1
            continue

        # -----------------------------
        # heads
        # -----------------------------
        if bucket in ("pathology", "bone_pathology"):
            current = open_head({"pathology": [(text, bucket)]})
            i += 1
            continue

        if bucket == "bleed_pathology":

            btype = bleed_type(text, tokens, i)

            # "no intraventricular extension of hemorrhage": the generic
            # word repeats the head before it
            if btype == "generic" and current is not None:
                i += 1
                continue

            # "no new or expanding intracranial hematoma / collection" is
            # about the SDH, not about an intraparenchymal bleed
            if btype == "generic" and text.lower().split()[-1] in ("hematoma", "collection", "hematomas", "collections"):
                btype = "subdural"

            current = open_head({
                "bleed": (text, btype)
            })
            i += 1
            continue

        if (
            bucket in EFFECTS
            and text.lower() in ("herniation", "brain herniation")
            and current is not None
            and current.get("effect_name", ("", ""))[1] == "herniation_subtype"
        ):
            i += 1
            continue

        if bucket in EFFECTS:

            magnitude = [
                (t, b) for t, b in pending_modifiers
                if b == "magnitude"
            ]
            location = [
                (t, b) for t, b in pending_modifiers + pending_anatomy
                if b != "magnitude"
            ]

            current = open_head({"effect_name": (text, bucket)})

            if magnitude:
                current["magnitude"] = magnitude
            if location:
                current["location"] = location

            i += 1
            continue

        if bucket == "density":
            current = open_head({"density": [(text, bucket)]})
            i += 1
            continue

        if bucket in PASS_THROUGH:
            i += 1
            continue

        # anything else (numbers, measurements, sinus findings...) ends the list
        break

    return branches, i
```

### Location Parser

```python
def parse_location_branch(sent_tokens, start_idx):

    branch = []
    i = start_idx + 1

    while i < len(sent_tokens):
        text, bucket = sent_tokens[i]

        # --- CONDITIONAL STOP: association_linker ---
        if bucket == "association_linker":
            # Look ahead safely
            if i + 1 < len(sent_tokens):
                next_bucket = sent_tokens[i + 1][1]

                # If next token is a location linker → continue scanning
                if next_bucket == "location_linker":
                    i += 1
                    continue

            # Otherwise → STOP
            break

        # --- NORMAL STOP TOKENS ---
        if bucket in LOCATION_STOP:
            break

        # --- ACCEPTABLE LOCATION TOKENS ---
        if bucket in LOCATION_ACCEPTABLE:
            branch.append((text, bucket))
            i += 1
            continue

        # --- OTHERWISE IGNORE AND CONTINUE ---
        i += 1

    return branch, i
```

### Association Parser

```python
# what may sit between a grade and its effect: "significant 1.4 cm midline
# shift", "mild leftward midline shift", "partial effacement"
BETWEEN_GRADE_AND_EFFECT = {
    "magnitude", "numerical", "measurement_unit", "laterality",
    "uncertainty_linker", "sulci", "ventricle", "uncus",
}


def adjacent_magnitude(tokens):
    """
    Walk back from the end of `tokens` (the words just before an effect)
    and keep the grades until something else is hit, so "marked
    progression in the midline shift" does not grade the shift.
    """

    found = []

    for text, bucket in reversed(tokens):

        if bucket not in BETWEEN_GRADE_AND_EFFECT:
            break

        if bucket == "magnitude":
            found.append((text, bucket))

    return list(reversed(found))


def parse_association(tokens, start_i):

    # 1. Slice tokens into segments using STOP buckets
    segments = []
    current_segment = []

    negation_groups = []

    i = start_i
    while i < len(tokens):
        text, bucket = tokens[i]

        # NEGATION → read the whole "no A or B" list on the full sentence,
        # before "or" / commas cut it into segments
        if bucket == "negation":
            if current_segment:
                segments.append(current_segment)
                current_segment = []

            neg_branches, next_i = parse_negation_branch(tokens, i)

            if neg_branches:
                negation_groups.append({"negation": neg_branches})

            i = max(next_i, i + 1)
            continue

        # "secondary to left-sided midline shift, measuring 1.6 cm":
        # a comma followed by a measurement keeps the clause going
        if (
            bucket == "comma"
            and i + 1 < len(tokens)
            and tokens[i + 1][1] in ("measurement_type", "measurement_linker_max")
        ):
            i += 1
            continue

        # HARD STOP → end entire association function
        if bucket in ASSOCIATION_HARD_STOP:
            if current_segment:
                segments.append(current_segment)
                current_segment = []
            break

        # "severe associated mass effect": the magnitude is still waiting
        # for its effect, so the linker does not cut the segment
        if (
            bucket == "association_linker"
            and current_segment
            and current_segment[-1][1] == "magnitude"
        ):
            i += 1
            continue

        # BRANCH STOP → end current segment, start new one
        if bucket in ASSOCIATION_BRANCH_STOP:
            if current_segment:
                segments.append(current_segment)
                current_segment = []
            i += 1
            continue

        current_segment.append((text, bucket))
        i += 1

    # Add last segment if any
    if current_segment:
        segments.append(current_segment)

    # 2. Process each segment
    associations = list(negation_groups)

    for seg in segments:
               
        # -------------------------
        # Numeric Mass Effect
        # -------------------------

        if (
            any(b == "numerical" for _, b in seg)
            and
            any(b in EFFECT_FAMILY for _, b in seg)
        ):

            associations.append(
                num_mass(seg)
            )

        current = None
        pre_effect_tokens = []
        j = 0

        while j < len(seg):

            text, bucket = seg[j]

            # ⭐ NEGATION HANDLING
            if bucket == "negation":

                neg_branch, next_j = (
                    parse_negation_branch(
                        seg,
                        j
                    )
                )

                associations.append(
                    {"negation": [neg_branch]}
                )

                j = next_j
                continue

            # EFFECT OPENER
            if bucket in EFFECT_FAMILY:
               
                if current:
                    associations.append(current)

                current = {
                    "effect_name": (text, bucket)
                }

                for t_text, t_bucket in pre_effect_tokens:

                    if t_bucket in (
                        "laterality",
                        "cerebral_location",
                        "sulci",
                        "ventricle",
                        "uncus",
                    ):

                        current.setdefault(
                            "location",
                            []
                        ).append(
                            (t_text, t_bucket)
                        )

                for t_text, t_bucket in adjacent_magnitude(pre_effect_tokens):

                    current.setdefault(
                        "magnitude",
                        []
                    ).append(
                        (t_text, t_bucket)
                    )

                pre_effect_tokens = []
                j += 1
                continue

            # DENSITY OPENER
            if bucket == "density":
            
                if current:
                    associations.append(current)
            
                current = {
                    "density": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # ACUITY OPENER
            if bucket == "acuity":
            
                if current:
                    associations.append(current)
            
                current = {
                    "acuity": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # MORPHOLOGY OPENER
            if bucket == "morphology":
            
                if current:
                    associations.append(current)
            
                current = {
                    "morphology": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # PROMINENCE OPENER
            if bucket == "prominence":
            
                if current:
                    associations.append(current)
            
                current = {
                    "prominence": (text, bucket)
                }

                grades = adjacent_magnitude(pre_effect_tokens)
                if grades:
                    current["magnitude"] = grades
            
                for t_text, t_bucket in pre_effect_tokens:
            
                    if t_bucket in (
                        "laterality",
                        "ventricle",
                    ):
            
                        current.setdefault(
                            "location",
                            []
                        ).append(
                            (t_text, t_bucket)
                        )
                        
                pre_effect_tokens = []
                j += 1
                continue

            # LOCATION LINKER
            if bucket == "location_linker" and current:

                # "effacement in both cerebral hemispheres"
                if "both" in text.lower():
                    current["both"] = True

                loc_list = []
                k = j + 1

                while k < len(seg):

                    t_text2, t_bucket2 = seg[k]

                    if t_bucket2 in ASSO_LOCATION_STOP:
                        break

                    if t_bucket2 == "sandwich_and":
                        k += 1
                        continue

                    if t_bucket2 in (
                        None,
                        "location_linker",
                        "subregional_anatomy"
                    ):
                        k += 1
                        continue

                    loc_list.append(
                        (t_text2, t_bucket2)
                    )

                    k += 1

                if loc_list:

                    current.setdefault(
                        "location",
                        []
                    ).extend(
                        loc_list
                    )

                j = k 
                continue

            # Store pre-effect tokens
            pre_effect_tokens.append(
                (text, bucket)
            )

            j += 1

        # location words AFTER the effect with no linker
        # ("Mild effacement left lateral ventricle")
        if current and pre_effect_tokens:

            # "sulcal effacement in both cerebral hemispheres"
            if any(t.lower().startswith("both") for t, _ in pre_effect_tokens):
                current["both"] = True

            trailing = [
                (t_text, t_bucket)
                for t_text, t_bucket in pre_effect_tokens
                if t_bucket in (
                    "laterality", "sulci", "ventricle", "cistern",
                    "uncus", "intraparenchymal", "cerebral_location",
                )
            ]

            if trailing and "location" not in current:
                current["location"] = trailing

            elif trailing and not any(
                b == "laterality" for _, b in current.get("location", [])
            ):
                current["location"] = current["location"] + [
                    (t, b) for t, b in trailing if b == "laterality"
                ]

        if current:
            associations.append(current)

    return associations, i
```

### Mass Effect Num Collector

```python
def num_mass(collected):

    value = None
    unit = None
    direction = None
    effect_name = None
    magnitude = []

    # Extract components from collected tokens
    for item in collected:

        if (
            isinstance(item, tuple)
            and isinstance(item[1], str)
            and item[1] in (
                "mass_effect",
                "herniation_subtype"
            )
        ):

            text = item[0].lower()

            # Always prefer midline shift
            if "midline" in text and text != "off the midline":

                effect_name = item

            # Only use off the midline
            # if nothing better found
            elif (
                text == "off the midline"
                and effect_name is None
            ):

                effect_name = (
                    "midline shift",
                    "mass_effect"
                )

            elif effect_name is None:

                effect_name = item

        # numeric value
        if (
            isinstance(item, tuple)
            and item[1] == "numerical"
        ):
            value = item[0]

        # measurement unit
        if (
            isinstance(item, tuple)
            and item[1] == "measurement_unit"
        ):
            unit = item[0]


        # laterality
        if (
            isinstance(item, tuple)
            and item[1] == "laterality"
        ):
            direction = item[0]

    # Combine number + unit into one string
    combined_value = None

    if value:

        if unit:
            combined_value = f"{value} {unit}"
        else:
            combined_value = value

    branch = {
        "effect_name": effect_name,
        "numeric_mass_effect": {
            "value": combined_value,
            "direction": direction
        }
    }

    # magnitude said directly in front of the effect
    # ("significant 1.4 cm midline shift")
    effect_index = next(
        (k for k, item in enumerate(collected)
         if isinstance(item, tuple) and item[0].lower() == str(effect_name[0]).lower()),
        None
    ) if effect_name else None

    if effect_index is not None:
        magnitude = adjacent_magnitude(collected[:effect_index])

    if magnitude:
        branch["magnitude"] = magnitude

    return branch
```

```python
def parse_association_location(tokens, start_i):
   
    i = start_i + 1
    locations = []

    while i < len(tokens):
        text, bucket = tokens[i]

        # STOP CONDITIONS
        if bucket in ASSO_LOCATION_STOP:
            break

        # ACCEPTABLE LOCATION TOKENS
        if bucket in LOCATION_ACCEPTABLE:
            locations.append(
                (text, bucket)
            )
            i += 1
            continue


        # JOINERS (sandwich_and only)
        if bucket == "sandwich_and":
            i += 1
            continue

        # UNKNOWN → STOP
        break

    return locations, i
```

## Short Sentences Function

### Other Pathology

```python
def parse_other_pathology(sent_tokens):

    negation_branches = []
    normal_state_token = None

    anatomy_tokens = []
    items = []              # one item per pathology head
    pending_modifiers = []  # saved until the next pathology head

    i = 0
    while i < len(sent_tokens):

        text, bucket = sent_tokens[i]

        # ============================
        # 1. NEGATION (reads the whole "No A, B or C" list)
        # ============================
        if bucket == "negation":
            branches, next_i = parse_pathology_negation_branch(
                sent_tokens,
                i
            )
            negation_branches.extend(branches)
            i = next_i
            continue

        # ============================
        # 2. NORMAL STATE / OUTCOME
        # ============================
        if (
            bucket in {"normal_state", "outcome"}
            or text.lower() in ("clear", "otherwise clear")
        ):
            normal_state_token = (text, bucket)
            i += 1
            continue

        # ============================
        # 3. ANATOMY
        # ============================
        if bucket in {
            "extracranial_anatomy",
            "sinus",
            "intraparenchymal",
            "sulci",
            "cistern",
            "ventricle",
            "cerebral_location",
            "bone",
            "tonsils",
        }:
            anatomy_tokens.append((text, bucket))
            i += 1
            continue

        # ============================
        # 4. PATHOLOGY (modifiers saved so far belong to this head)
        # ============================
        if bucket in {
            "pathology",
            "bone_pathology",
            "general_pathology"
        }:
            items.append({
                "pathology": [(text, bucket)],
                "modifiers": pending_modifiers,
            })
            pending_modifiers = []
            i += 1
            continue

        # ============================
        # 5. MODIFIERS
        # ============================
        if bucket in {
            "magnitude",
            "acuity",
            "territorial",
            "cortical",
        }:
            pending_modifiers.append((text, bucket))
            i += 1
            continue

        # ============================
        # 6. LOCATION LINKER
        # ============================
        if bucket == "location_linker":
            loc_branch, next_i = parse_association_location(
                sent_tokens,
                i
            )

            anatomy_tokens.extend(loc_branch)
            i = max(next_i, i + 1)
            continue

        i += 1

    # modifiers after the last head ("... which may represent mild chronic
    # microangiopathic disease" is handled above; "infarct, old" here)
    if items and pending_modifiers:
        items[-1]["modifiers"] = items[-1]["modifiers"] + pending_modifiers

    # ============================
    # BUILD FINAL OBJECT
    # ============================
    result = {}

    # NEGATION OVERRIDES EVERYTHING, but a normal anatomy said before it
    # ("... maintained with no evidence of acute infarct") is kept too
    if negation_branches:
        result["negation"] = negation_branches
        if normal_state_token and anatomy_tokens:
            result["negation"] = negation_branches + [{
                "outcome": {
                    "normal_state": normal_state_token,
                    "anatomy": anatomy_tokens,
                }
            }]
        return result

    # NORMAL OUTCOME
    if normal_state_token and not items:
        result["outcome"] = {
            "normal_state": normal_state_token,
            "anatomy": anatomy_tokens if anatomy_tokens else None
        }
        return result

    # POSITIVE PATHOLOGY
    if items:
        result["outcome"] = {
            "items": items,
            "anatomy": anatomy_tokens if anatomy_tokens else None,
        }

    return result
```

### Other Path Negation

```python
def parse_pathology_negation_branch(tokens, start_i):
    """
    Negated pathology sentences use the same list reader as the bleed tree,
    so "No intracranial herniation or hydrocephalus" negates BOTH the
    herniation (an effect) and the hydrocephalus (a pathology).
    Returns (list_of_branches, next_index).
    """

    return parse_negation_branch(tokens, start_i)
```

### Sinus / Mastoid

```python
def sinus_column(text):

    text = text.lower()

    if "mastoid" in text:
        return "mastoid_air_cells"
    if "maxillary" in text:
        return "maxillary_sinus"
    if "sphenoid" in text:
        return "sphenoid_sinus"
    if "ethmoid" in text:
        return "ethmoid_sinus"
    if "frontal sinus" in text:
        return None
    return "paranasal_sinuses"


def sinus_finding_value(text, magnitude_words):

    text = text.lower()
    mild = bool(magnitude_words & {"mild", "trace", "minimal", "small", "partial", "slight", "minor"})
    moderate = "moderate" in magnitude_words
    severe = bool(magnitude_words & {"severe", "complete", "extensive", "marked"})

    if text in ("good aeration", "aeration",
                "clear", "otherwise clear", "aerated", "well aerated", "well-aerated",
                "normally aerated", "well pneumatized", "well-pneumatized"):
        return "Well aerated"

    if text in ("unremarkable", "normal", "normal limits"):
        return "Unremarkable"

    if "mucosal thickening" in text:
        if moderate:
            return "Moderate mucosal thickening"
        return "Mild mucosal thickening" if mild else "Mucosal thickening NS"

    if "retention cyst" in text:
        return "Retention cyst"

    if "opacification" in text:
        if severe:
            return "Severe opacification"
        return "Mild opacification" if (mild or "partial" in magnitude_words) else "Opacification NS"

    if text in ("fluid", "air-fluid levels"):
        return "Fluid"

    return None


def parse_sinus_sentence(sent_tokens):
    """
    "Visualized paranasal sinuses and mastoid air cells are clear."
    "Retention cyst left maxillary sinus."
    "Mild mucosal thickening in bilateral maxillary sinuses.
     Paranasal sinuses ... are otherwise clear."

    Save every sinus named in the sentence, then give it the finding
    (the first abnormal finding wins over "clear").
    Returns {column: value}.
    """

    columns = []
    findings = []
    magnitude_words = set()

    for text, bucket in sent_tokens:

        if bucket == "sinus":
            column = sinus_column(text)
            if column and column not in columns:
                columns.append(column)

        elif bucket in ("sinus_finding", "normal_state"):
            value = sinus_finding_value(text, magnitude_words)
            if value:
                findings.append(value)

        elif bucket == "magnitude":
            magnitude_words.add(text.lower())

        elif bucket == "negation" and not findings:
            return {}

    if not columns or not findings:
        return {}

    # "fluid in the right sphenoid sinus ... inflammatory sinus disease":
    # the generic word repeats the specific sinus
    if len(columns) > 1 and "paranasal_sinuses" in columns:
        specific = [c for c in columns if c != "paranasal_sinuses"]
        if not any("paranasal" in t.lower() for t, b in sent_tokens if b == "sinus"):
            columns = specific

    # "opacification of the left sphenoid sinus (blood?)", "hemorrhage in the sinus"
    words = {t.lower().strip("?") for t, _ in sent_tokens}
    if words & {"blood", "hemorrhage", "hemorrhagic", "hemosinus"}:
        findings.insert(0, "Hemorrhage")

    abnormal = [f for f in findings if f not in ("Well aerated", "Unremarkable")]
    value = abnormal[0] if abnormal else findings[0]

    result = {}

    for column in columns:
        result[column] = value

    return result
```

### Fractures

```python
# ============================
# FRACTURES  (three variables)
#   bony_lesion        any fracture = Yes; "No ... fracture" = No
#   fracture_location  Calvarial / Skull base / Facial (joined with " + ");
#                      "No skull fracture", "calvarium intact" = No;
#                      "No fracture" with no bone named = NR (cleaned downstream)
#   fracture_type      most severe said: Comminuted > Depressed > Displaced > Nondisplaced
# ============================

LOCATION_NAMES = {"calvarial": "Calvarial", "skullbase": "Skull base", "facial": "Facial"}

FRACTURE_BONES = {
    "facial": [
        "nasal", "orbit", "orbital", "maxilla", "maxillary", "zygom", "mandib",
        "lamina papyracea", "facial", "pterygoid", "infraorbital",
    ],
    "skullbase": [
        "skull base", "skullbase", "petrous", "clivus", "condyle", "foramen magnum",
        "cranial fossa", "sphenoid", "basilar skull", "temporal bone",
    ],
    "calvarial": [
        "calvari", "skull", "parietal", "frontal bone", "occipital bone", "squamous",
        "vertex", "cranial vault", "frontal", "occipital",
    ],
}

INTACT_WORDS = ("intact", "unremarkable", "normal")

# "intact / unremarkable" only says "no fracture" when it is about BONE:
# "orbital rims are intact" yes, "the orbits are unremarkable" no
INTACT_BONE_WORDS = (
    "bone", "bony", "calvari", "skull", "rim", "wall", "arch", "mandible",
    "maxilla", "condyle", "clivus", "petrous", "lamina papyracea",
)


def fracture_bone_groups(text):
    """
    Which bone groups a sentence names. 'skull base' is not the calvarium,
    and 'temporal bone' / 'petrous' is skull base.
    """

    groups = []
    rest = text

    for group in ("facial", "skullbase", "calvarial"):
        for word in FRACTURE_BONES[group]:
            if word in rest:
                groups.append(group)
                rest = rest.replace(word, " ")
                break

    return groups


def parse_fracture_sentence(sent_tokens):
    """
    Returns {variable: "Yes" / "No"} for one sentence.
    """

    words = [t.lower() for t, _ in sent_tokens]
    text = " ".join(words)

    has_fracture = "fractur" in text
    is_intact = any(w in words for w in INTACT_WORDS)

    if not has_fracture and not is_intact:
        return {}

    groups = fracture_bone_groups(text)
    result = {}

    if has_fracture:

        fracture_at = next(i for i, w in enumerate(words) if "fractur" in w)

        negated = any(
            b == "negation" for _, b in sent_tokens[:fracture_at]
        ) and not any(w in ("but", "however") for w in words[:fracture_at])

        if negated:
            result["bony_lesion"] = "No"
            if groups:
                result["fracture_location"] = "No"
            return result

        result["bony_lesion"] = "Yes"

        if groups:
            result["fracture_location"] = " + ".join(
                LOCATION_NAMES[g] for g in ("calvarial", "skullbase", "facial") if g in groups
            )

        # the type is only kept for the calvarium (calvarial_fracture_type)
        if "calvarial" not in groups:
            return result

        if re.search(r"(?<!non-)(?<!non)comminut", text):
            result["fracture_type"] = "Comminuted"
        elif re.search(r"(?<!non-)(?<!non)depressed", text):
            result["fracture_type"] = "Depressed"
        elif re.search(r"non-?displaced|undisplaced", text):
            result["fracture_type"] = "Nondisplaced"
        elif "displaced" in text:
            result["fracture_type"] = "Displaced"

        return result

    # "The bony calvarium is intact", "Orbital rims are intact"
    if not any(w in text for w in INTACT_BONE_WORDS):
        return {}

    if groups and not any(b == "negation" for _, b in sent_tokens):
        result["bony_lesion"] = "No"
        result["fracture_location"] = "No"

    return result
```

## Detect Pathologies Function

```python
def detect_pathology_root(sent_tokens):
    for text, bucket in sent_tokens:
        if bucket in ("pathology", "bone_pathology"):
            return bucket
    for text, bucket in sent_tokens:
        if bucket == "bleed_pathology":
            return "bleed_pathology"
    return None
```

```python
def detect_other_pathology_root(sent_tokens):

    has_target_anatomy = any(bucket in {
        "extracranial_anatomy",
        "sinus",
        "intraparenchymal",
        "sulci",
        "cistern",
        "ventricle",
        "cerebral_location",
        "bone",
        "tonsils",
    } for _, bucket in sent_tokens)

    has_state = any(
        bucket in {"normal_state", "outcome"}
        or text.lower() in ("clear", "otherwise clear")
        for text, bucket in sent_tokens
    )

    if has_target_anatomy and has_state:
        return True

    return False
```

```python
def detect_relevant_sentence(sent_tokens):

    RELEVANT_BUCKETS = {

        "bleed_pathology",
        "pathology",
        "bone_pathology",

        "mass_effect",
        "herniation_subtype",

        "density",
        "acuity",

        "cerebral_location",
        "intraparenchymal",
        "ventricle",
        "cistern",
        "sulci",
        "extracranial_anatomy",
        "bone",

        "normal_state",
        "negation",

        "prominence",
        "morphology",
        "bleed_reference",
        "measurement_max",

    }

    return any(
        bucket in RELEVANT_BUCKETS
        for _, bucket in sent_tokens
    )
```

### Bleed routing helpers

```python
def bleed_type(text, sent_tokens=None, idx=None):
    """
    Which compartment a bleed_pathology token refers to.
    subdural / subarachnoid / intraparenchymal / intraventricular / epidural / generic

    A generic word ("hematoma", "collection", "bleed") takes its type from
    an anatomy qualifier directly in front of it ("parenchymal hemorrhage").
    """

    text = text.lower()

    if "subdural" in text or "sdh" in text or "extra-axial" in text or "extra- axial" in text:
        return "subdural"

    if "parafalcine" in text:
        return "subdural"

    if "subarachnoid" in text:
        return "subarachnoid"

    if "intraventricular" in text:
        return "intraventricular"

    if "parenchymal" in text or "intra-axial" in text or "intra- axial" in text:
        return "intraparenchymal"

    if "epidural" in text:
        return "epidural"

    if sent_tokens is not None and idx is not None:

        # look back a couple of tokens for a qualifier
        for back in range(1, 3):

            if idx - back < 0:
                break

            prev_text, prev_bucket = sent_tokens[idx - back]
            prev_text = prev_text.lower()

            if prev_text in ("subdural", "extra-axial"):
                return "subdural"

            if prev_text == "subarachnoid":
                return "subarachnoid"

            if prev_text in ("intraparenchymal", "parenchymal", "intra-axial"):
                return "intraparenchymal"

            if prev_text == "intraventricular":
                return "intraventricular"

            if prev_bucket not in ("magnitude", "acuity", "laterality",
                                   "cerebral_location", "density", None):
                break

    return "generic"


def bleed_is_negated(sent_tokens, idx):
    """
    A bleed token is negated when a negation word appears before it
    in the same sentence with no 'but' in between
    ("No acute intracranial hemorrhage", "There is no evidence of
    parenchymal hemorrhage").
    """

    for back in range(idx - 1, -1, -1):

        text, bucket = sent_tokens[back]

        if text.lower() in ("but", "however", "although"):
            return False

        if bucket == "negation":
            return True

    return False


def bleed_mentions(sent_tokens):
    """
    [(idx, text, type, negated), ...] for every bleed token in a sentence
    """

    mentions = []

    for idx, (text, bucket) in enumerate(sent_tokens):

        if bucket != "bleed_pathology":
            continue

        mentions.append(
            (
                idx,
                text,
                bleed_type(text, sent_tokens, idx),
                bleed_is_negated(sent_tokens, idx)
            )
        )

    return mentions


ANAPHORA_WORDS = {"lesion", "collection", "hematoma", "abnormality"}


def resolve_anaphora(sent_tokens):
    """
    "The lesion ..." / "This collection ..." after an SDH has been
    described refers back to the SDH, not to a new pathology.
    """

    resolved = list(sent_tokens)

    for idx in range(1, len(resolved)):

        text, bucket = resolved[idx]
        prev_text = resolved[idx - 1][0].lower()

        if (
            text.lower() in ANAPHORA_WORDS
            and bucket in ("pathology", "bleed_pathology")
            and prev_text in ("the", "this")
        ):
            resolved[idx] = (text, "bleed_reference")

    return resolved


def sentence_has(sent_tokens, buckets):
    return any(
        bucket in buckets
        for _, bucket in sent_tokens
    )
```

## Orchestrator Function

```python
def orchestrator(sentence_lists):
    """
    Route every sentence:

      sinus / mastoid sentence           -> parse_sinus_sentence
      positive subdural (or generic) bleed -> THE one SDH tree (all sides,
                                            all wordings: collection, SDH,
                                            hematoma, extra-axial ...)
      positive SAH / IPH / IVH           -> {"hemorrhage": ...}
      negated bleed / other pathology    -> parse_other_pathology
      anything else relevant             -> secondary sentence of the SDH tree
                                            (effects seen before the SDH is
                                            described are saved in `pending`)
    """

    bleed_trees = []
    sdh_tree = None
    pending = []
    last_bleed_type = None

    def add_secondary(sent_tokens):

        if sdh_tree is not None:
            build_bleed_tree(
                sent_tokens,
                is_bleed_sentence=False,
                existing_tree=sdh_tree
            )
            return

        # no SDH yet: keep the effects until it shows up
        holder = {"pathology": None, "children": []}

        build_bleed_tree(
            sent_tokens,
            is_bleed_sentence=False,
            existing_tree=holder
        )

        pending.extend(
            child for child in holder["children"]
            if isinstance(child, dict) and "association" in child
        )

    for sent_tokens in sentence_lists:

        sent_tokens = resolve_anaphora(sent_tokens)

        # "The patient has a history of a left orbital floor fracture"
        # describes the past, not this scan
        if sentence_has(sent_tokens, {"history"}):
            continue

        # cervical spine / prevertebral findings are not head findings
        if (
            sentence_has(sent_tokens, {"spine"})
            and not sentence_has(sent_tokens, {"bleed_pathology"})
        ):
            continue

        # ============================
        # FRACTURES (own variables, any sentence)
        # ============================
        fracture = parse_fracture_sentence(sent_tokens)

        if fracture:
            bleed_trees.append({"fracture": fracture})

        # ============================
        # CASE 0: SINUS / MASTOID
        # ============================
        if sentence_has(sent_tokens, {"sinus"}):

            sinus = parse_sinus_sentence(sent_tokens)

            if sinus:
                bleed_trees.append({"sinus": sinus})

            remaining = [
                (text, bucket) for text, bucket in sent_tokens
                if bucket not in ("sinus", "sinus_finding")
            ]

            if not detect_relevant_sentence(remaining):
                continue

        # ============================
        # CASE 1: BLEEDS
        # ============================
        mentions = bleed_mentions(sent_tokens)
        positive = [m for m in mentions if not m[3]]

        sdh_positive = False
        other_types = []

        for idx, text, btype, negated in positive:

            if btype == "generic":
                btype = last_bleed_type or "subdural"

            if btype == "subdural":
                sdh_positive = True
            elif btype not in other_types:
                other_types.append(btype)

            last_bleed_type = btype

        for btype in other_types:
            bleed_trees.append(
                {"hemorrhage": {"type": btype, "grade": hemorrhage_grade(sent_tokens) if btype == "subarachnoid" else None}}
            )

        if sdh_positive:

            if sdh_tree is None:

                sdh_tree = {
                    "pathology": "subdural",
                    "children": list(pending)
                }
                pending = []
                bleed_trees.append(sdh_tree)

            build_bleed_tree(
                sent_tokens,
                is_bleed_sentence=True,
                existing_tree=sdh_tree
            )
            continue

        if other_types:

            # "There is a large midline shift" style effects in a sentence
            # about another bleed still describe the brain
            if sentence_has(sent_tokens, EFFECT_FAMILY):
                add_secondary(sent_tokens)
            continue

        # "the hematoma ..." -> still about the SDH
        if (
            sentence_has(sent_tokens, {"bleed_reference"})
            and last_bleed_type not in (None, "subdural")
        ):
            continue

        # "The hematoma underlies the right frontal parietal convexity
        # measuring up to 9 mm": where and how big, read like a bleed
        # sentence; the effects in it are read over the whole sentence
        if (
            sentence_has(sent_tokens, {"bleed_reference"})
            and sdh_tree is not None
            and not sentence_has(sent_tokens, {"pathology", "bone_pathology"})
        ):
            holder = {"pathology": None, "children": []}

            build_bleed_tree(sent_tokens, is_bleed_sentence=True, existing_tree=holder)

            # mark it: "This hematoma ..." continues the bleed of the last sentence
            for child in holder["children"]:
                if isinstance(child, dict) and "sentence_start" in child:
                    child["anaphor"] = True

            sdh_tree["children"].extend(
                child for child in holder["children"]
                if not (isinstance(child, dict) and "association" in child)
            )

            assoc, _ = parse_association(sent_tokens, 0)
            assoc = [a for a in assoc if not any(k in a for k in ("density", "acuity", "morphology"))]

            if assoc:
                sdh_tree["children"].append({"association": assoc})

            continue

        # ============================
        # CASE 2: OTHER PATHOLOGY / NEGATED BLEED / NORMAL ANATOMY
        # ============================
        root_bucket = detect_pathology_root(sent_tokens)

        if (
            root_bucket in ("pathology", "bone_pathology")
            or mentions
            or detect_other_pathology_root(sent_tokens)
        ):

            other_result = parse_other_pathology(sent_tokens)

            if other_result:
                bleed_trees.append(other_result)

            # mixed sentence: "atrophy with ex vacuo dilatation of the
            # ventricles", "fullness ... concerning for early trapping"
            if sentence_has(sent_tokens, EFFECT_FAMILY | {"prominence"}):
                add_secondary(sent_tokens)

            continue

        # ============================
        # NOT RELEVANT
        # ============================
        # "This measures 11.0 x 2.5 cm" points back to the SDH
        if not detect_relevant_sentence(sent_tokens) and not secondary_thickness(sent_tokens):
            continue

        # ============================
        # CASE 3: NO PATHOLOGY -> secondary sentence of the SDH
        # ============================
        add_secondary(sent_tokens)

    if sdh_tree is None and pending:
        bleed_trees.append(
            {"pathology": None, "children": pending}
        )

    return bleed_trees
```

```python
# ============================
# FINAL OUTPUT VARIABLES
#   The column names, order and value wording of the study dataset.
#   Left = column in SDH_results.csv, right = the variable the pipeline
#   fills. Edit names here; the extraction code does not need to change.
# ============================

FINAL_COLUMNS = [

    # Overall
    ("laterality", "laterality"),

    # Left hematoma
    ("L_description", "L_description"),
    ("L_acuity", "L_acuity"),
    ("L_internal_appearance", "L_morphology"),
    ("L_mixed_density", "L_mixed_density"),
    ("L_hyperdense", "L_hyper_dense"),
    ("L_hypodense", "L_hypo_dense"),
    ("L_isodense", "L_iso_dense"),
    ("L_density_dominance", "L_density_dominance"),
    ("L_thickness_mm", "L_thickness_mm"),
    ("L_thickness_cm", "L_thickness_cm"),
    ("L_convexity_hematoma", "L_convexity_hematoma"),
    ("L_holohemispheric_hematoma", "L_holohemispheric_hematoma"),
    ("L_vertex", "L_vertex"),
    ("L_frontal_lobe", "L_frontal_lobe"),
    ("L_parietal_lobe", "L_parietal_lobe"),
    ("L_temporal_lobe", "L_temporal_lobe"),
    ("L_occipital_lobe", "L_occipital_lobe"),

    # Right hematoma
    ("R_description", "R_description"),
    ("R_acuity", "R_acuity"),
    ("R_internal_appearance", "R_morphology"),
    ("R_mixed_density", "R_mixed_density"),
    ("R_hyperdense", "R_hyper_dense"),
    ("R_hypodense", "R_hypo_dense"),
    ("R_isodense", "R_iso_dense"),
    ("R_density_dominance", "R_density_dominance"),
    ("R_thickness_mm", "R_thickness_mm"),
    ("R_thickness_cm", "R_thickness_cm"),
    ("R_convexity_hematoma", "R_convexity_hematoma"),
    ("R_holohemispheric_hematoma", "R_holohemispheric_hematoma"),
    ("R_vertex", "R_vertex"),
    ("R_frontal_lobe", "R_frontal_lobe"),
    ("R_parietal_lobe", "R_parietal_lobe"),
    ("R_temporal_lobe", "R_temporal_lobe"),
    ("R_occipital_lobe", "R_occipital_lobe"),

    # Other locations
    ("parafalcine_hematoma", "parafalcine_hematoma"),
    ("parafalcine_location", "parafalcine_location"),
    ("parafalcine_thickness_mm", "parafalcine_thickness_mm"),
    ("parafalcine_thickness_cm", "parafalcine_thickness_cm"),
    ("parafalcine_acuity", "parafalcine_acuity"),
    ("tentorium_hematoma", "tentorium_hematoma"),
    ("tentorium_thickness_mm", "tentorium_thickness_mm"),
    ("tentorium_thickness_cm", "tentorium_thickness_cm"),
    ("tentorium_acuity", "tentorium_acuity"),

    # Mass effect
    ("mass_effect_present", "mass_effect_present"),
    ("midline_shift_present", "midline_shift_present"),
    ("midline_shift_mm", "midline_shift_mm"),
    ("midline_shift_cm", "midline_shift_cm"),
    ("midline_shift_direction", "midline_shift_direction"),

    # Effacement
    ("effacement_sulci", "effacement_sulci"),
    ("effacement_sulci_laterality", "effacement_sulci_laterality"),
    ("effacement_lateral_ventricle", "effacement_lateral_ventricle"),
    ("effacement_lateral_ventricle_laterality", "l_ventricle_laterality"),
    ("effacement_ventricular_system", "effacement_ventricular_system"),
    ("ventricular_prominence", "ventricular_prominence"),
    ("l_v_prominence_laterality", "l_v_prominence_laterality"),
    ("ex_vacuo_ventriculomegaly", "ex_vacuo_dilatation"),

    # Herniation
    ("herniation_subfalcine", "herniation_subfalcine"),
    ("herniation_uncal", "herniation_uncal"),
    ("herniation_transtentorial", "herniation_transtentorial"),
    ("herniation_tonsillar", "herniation_tonsillar"),

    # Other findings
    ("encephalomalacia", "encephalomalacia"),
    ("hydrocephalus", "hydrocephalus"),
    ("microangiopathy", "microangiopathy"),
    ("cerebral_atrophy", "cerebral_atrophy"),
    ("white_dark_differentiation", "white_dark_differentiation"),
    ("mass", "mass"),
    ("unspecified_infarct", "unspecified_infarct"),
    ("territorial_infarct", "territorial_infarct"),
    ("lacune_infarct", "lacunar_infarct"),
    ("soft_tissue_lesion", "soft_tissue_lesion"),
    ("scalp_hematoma", "scalp_hematoma"),
    ("bony_lesion", "bony_lesion"),
    ("orbital_lesion", "orbital_lesion"),
    ("fracture_location", "fracture_location"),
    ("calvarial_fracture_type", "fracture_type"),
    ("parenchymal_contusion_present", "contusion"),
    ("parenchymal_edema", "edema"),
    ("parenchymal_hemorrhage", "intraparenchymal_hemorrhage"),
    ("subarachnoid_hemorrhage", "subarachnoid_hemorrhage"),
    ("intraventricular_hemorrhage", "intraventricular_hemorrhage"),
    ("aneurysm", "aneurysm"),

    # Sinuses / mastoids
    ("paranasal_sinus", "paranasal_sinuses"),
    ("mastoid_air_cells", "mastoid_air_cells"),
    ("sphenoid_sinus", "sphenoid_sinus"),
    ("ethmoid_sinus", "ethmoid_sinus"),
    ("maxillary_sinus", "maxillary_sinus"),
]

# value wording of the final dataset (applies to every column)
FINAL_VALUE_WORDING = {
    "Acute on Chronic": "Acute on chronic",
    "Subacute on Chronic": "Subacute on chronic",
    "Age Appropriate": "Age appropriate",
    "Dependent": "Dependant",
    "Minimal": "Small",
    "Diffuse": "Yes NS",
    "Moderate/ Severe": "Severe",      # "moderate to severe" -> the higher grade
    "Mild/ Moderate": "Moderate",
    "Nondisplaced": "Non-displaced",
    "Preserved": "Yes",                # gray-white differentiation present
    "Loss": "No",
    "Mild Opacification": "Mild opacification",
}

# a plain "Yes" is "Yes NS" in these graded columns
YES_NS_COLUMNS = {
    "mass_effect_present", "midline_shift_present", "effacement_sulci",
    "effacement_lateral_ventricle", "effacement_ventricular_system",
    "ventricular_prominence", "ex_vacuo_ventriculomegaly",
    "herniation_subfalcine", "herniation_uncal", "herniation_transtentorial",
    "herniation_tonsillar", "encephalomalacia", "hydrocephalus",
    "microangiopathy", "cerebral_atrophy", "parafalcine_hematoma", "tentorium_hematoma",
}


def final_fracture_location(value):
    # "Calvarial + Skull base" -> "Calvarial & Skull base" in the order of the list
    if value in ("NR", "No", None):
        return value
    parts = set(value.split(" + "))
    order = ["Calvarial", "Facial", "Skull base"]
    return " & ".join(p for p in order if p in parts)


def final_output(variables):
    """
    Pipeline variables -> one row of the final dataset (ordered dict).
    """

    row = {}

    # infarct types without their own column fold into the nearest one
    variables = dict(variables)
    if variables.get("unspecified_infarct", "NR") == "NR":
        variables["unspecified_infarct"] = variables.get("cortical_infarct", "NR")
    if variables.get("territorial_infarct", "NR") == "NR":
        variables["territorial_infarct"] = variables.get("territorial_ischemic", "NR")

    for column, source in FINAL_COLUMNS:

        value = variables.get(source, "NR")

        if column == "fracture_location":
            value = final_fracture_location(value)

        if isinstance(value, str):
            value = FINAL_VALUE_WORDING.get(value, value)

            if value == "Yes" and column in YES_NS_COLUMNS:
                value = "Yes NS"

        row[column] = value

    return row
```

# RUN ALL REPORTS (SDH_reports.csv -> SDH_results.csv)

```python
# ============================
# RUN ALL REPORTS
#   Input  SDH_reports.csv : 3 columns, in this order
#          1. randomized report number   2. number 1..n   3. report text
#          (with or without a header row - columns are read by position)
#   Output SDH_results.csv : report_num, report_index, report, tree,
#          then every variable in FINAL_COLUMNS, then error
# ============================

import contextlib
import io
import pprint
import pandas as pd


def tree_text(bleed_trees):
    """
    The tree before variable mapping, as readable text for one cell.
    """
    return pprint.pformat(bleed_trees, width=110, sort_dicts=False)


def run_csv(input_path="SDH_reports.csv", output_path="SDH_results.csv"):

    reports = pd.read_csv(input_path, header=None, dtype=str, keep_default_na=False)

    # drop a header row if there is one (its 2nd column is not a number)
    if len(reports) and not str(reports.iloc[0, 1]).strip().isdigit():
        reports = reports.iloc[1:]

    rows = []

    for _, report in reports.iterrows():

        report_num = str(report.iloc[0]).strip()
        report_index = str(report.iloc[1]).strip()
        text = str(report.iloc[2])

        row = {
            "report_num": report_num,
            "report_index": report_index,
            "report": text,
        }

        try:
            # the parsers print debugging output - hide it for batch runs
            with contextlib.redirect_stdout(io.StringIO()):
                sentence_lists = tokenize_report(text)
                bleed_trees = [orchestrator(sentence_lists)]
                variables = master_variable_mapping_function(bleed_trees)

            row["tree"] = tree_text(bleed_trees)
            row.update(final_output(variables))
            row["error"] = ""

        except Exception as error:
            # keep going: this report gets NR everywhere and the error is recorded
            row["tree"] = ""
            row.update(final_output({}))
            row["error"] = f"{type(error).__name__}: {error}"

        rows.append(row)

    results = pd.DataFrame(rows)

    # utf-8-sig so Excel opens it with the right characters
    results.to_csv(output_path, index=False, encoding="utf-8-sig")

    failed = (results["error"] != "").sum()
    print(f"{len(results)} reports -> {output_path}  ({failed} with errors)")

    return results
```
