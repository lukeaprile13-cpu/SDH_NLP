```python
import spacy
from spacy.matcher import PhraseMatcher
from spacy.language import Language
from spacy.util import filter_spans     # removes overlapping phrases (there is, there is a) to get the longest phrase

nlp = spacy.load("en_core_web_sm")
nlp.disable_pipes(["parser"])
```

```python
# --- plural named bleeds: "subarachnoid hemorrhages" (not a bare "hematomas") ---
for _key, _bucket in list(BLEED_PATHOLOGY.items()):
    if " " in _key and _key.endswith(("hemorrhage", "hematoma", "haemorrhage", "haematoma", "bleed", "contusion")) and _key + "s" not in BLEED_PATHOLOGY:
        BLEED_PATHOLOGY[_key + "s"] = _bucket
        TOKEN_BUCKETS.setdefault(_key + "s", _bucket)

# --- Collect all bucket dictionaries into one list ---
all_dicts = [
    BLEED_PATHOLOGY,
    OTHER_PATHOLOGIES,
    INTRACRANIAL_ANATOMY,
    EXTRACRANIAL_ANATOMY,
    BONE,
    NOTICING,
    LINKERS,
    UNCERTAINTY,
    CEREBRAL_LOCATIONS,
    SUBREGIONAL_ANATOMY,
    MASS_EFFECT,
    HERNIATION_SUBTYPES,
    ACUITY,
    DENSITY,
    PROMINENCE,
    MAGNITUDE,
    MORPHOLOGY,
    NORMAL_STATE,
    MEASUREMENT,
    NEGATION,
    LATERALITY,
    IGNORE,
    SINUS,
    REPORT_TITLES,
    PUNCTUATION,
    NUMERICAL
]


# --- Extract multi-word keys ---
multiword_phrases = []

for d in all_dicts:
    for tok in d.keys():
        if " " in tok or "-" in tok or "/" in tok:   # contains a space, - or / (multi-word)
            multiword_phrases.append(tok)

patterns = [nlp.make_doc(text) for text in multiword_phrases]
matcher = PhraseMatcher(nlp.vocab, attr="LOWER")
matcher.add("TOKEN_BUCKET_PHRASES", patterns)

@Language.component("merge_phrases")
def merge_phrases(doc):
    matches = matcher(doc)
    spans = [doc[start:end] for _, start, end in matches]
    filtered = filter_spans(spans)
    with doc.retokenize() as retokenizer:
        for span in filtered:
            retokenizer.merge(span)
    return doc

nlp.add_pipe("merge_phrases", first=True)

# --- Force sentence splitting ONLY by punctuation ---
@Language.component("punct_only_sentences")
def punct_only_sentences(doc):

    for token in doc:
        token.is_sent_start = False

    for i, token in enumerate(doc):

        if i == 0:
            token.is_sent_start = True
            continue

        # normal punctuation
        if doc[i-1].text in {".", ";", ":"}:
            token.is_sent_start = True
            continue

        # implicit sentence break:
        # Capitalized word but NOT all caps
        if (
            token.text[0].isupper()
            and not token.text.isupper()
        ):
            token.is_sent_start = True

    return doc

# GUARANTEED to run after parser
nlp.add_pipe("punct_only_sentences", last=True)
```

## Preprocessing, sections and run_report

```python
import re

# ============================
# REPORT PREPROCESSING
# ============================

# every word that appears in a dictionary key, plus common report words
KNOWN_WORDS = set()

for key in TOKEN_BUCKETS:
    for word in re.findall(r"[a-z]+", key.lower()):
        KNOWN_WORDS.add(word)

KNOWN_WORDS.update({
    "hemorrhage", "dimension", "differentiation", "intraventricular", "excluded",
    "measuring", "measures", "significant", "subdural", "hematoma", "hemorrhage",
    "herniation", "ventricle", "ventricles", "right", "left", "side", "which",
    "blood", "image", "parietal", "lateral", "microangiopathic", "subarachnoid",
    "lateral", "sulci", "interpretation", "consistent", "hemorrhages",
})


def repair_split_words(text):
    """
    Dictated / OCR reports often contain words broken by a space
    ("subd ural", "herni ation", "m easuring", "righ t").
    Join two fragments when neither is a known word but the joined
    word is.
    """

    tokens = text.split(" ")
    repaired = []
    i = 0

    while i < len(tokens):

        if i + 1 < len(tokens):

            left = tokens[i]
            right = tokens[i + 1]

            left_core = left.lower().strip(".,;:()")
            right_core = right.lower().strip(".,;:()")
            joined = (left_core + right_core)

            if (
                left_core.isalpha()
                and right_core.isalpha()
                and left_core not in KNOWN_WORDS
                and (right_core not in KNOWN_WORDS or len(left_core) >= 4 or len(right_core) == 1)
                and joined in KNOWN_WORDS
            ):
                repaired.append(left + right)
                i += 2
                continue

        repaired.append(tokens[i])
        i += 1

    return " ".join(repaired)


GLUE_FUNCTION_WORDS = {"of", "the", "and", "for", "are", "is", "in", "on", "at", "to", "with", "a", "an", "no"}
GLUE_SIDE_WORDS = {"left", "right"}
_glue_words = None


def glue_words():
    """
    Single words the dictionary knows ("thickness", "acute", "paranasal").
    """
    global _glue_words
    if _glue_words is None:
        words = set()
        for key in TOKEN_BUCKETS:
            for word in re.split(r"[\s/-]+", key.lower()):
                if word.isalpha():
                    words.add(word)
        words |= {"amount", "findings", "stable", "unchanged", "measurement", "thickness", "areas",
                  "material", "seen", "noted", "blood", "products", "fluid", "identified", "present"}
        _glue_words = words
    return _glue_words


def split_glued(match):
    """
    "thicknessof" -> "thickness of", "foracute" -> "for acute",
    "Stableright" -> "Stable right", "paranasalsinuses" -> "paranasal sinuses".
    Only an unknown word is split, and only into two known words.
    """
    token = match.group(0)
    low = token.lower()
    known = glue_words()
    if low in known:
        return token
    for i in range(2, len(low) - 1):
        a, b = low[:i], low[i:]
        if (
            (a in GLUE_FUNCTION_WORDS and b in known and len(b) >= 4)
            or (a in known and (len(a) >= 5 or a in ("seen", "blood")) and (b in GLUE_FUNCTION_WORDS or (b in known and len(b) >= 4)))
            or (len(a) >= 4 and b in GLUE_SIDE_WORDS)
        ):
            return token[:i] + " " + token[i:]
    return token


def preprocess_report(text):

    text = str(text)

    # abbreviations, before the glued-word splitter turns "cSDH" into "c SDH"
    text = re.sub(r"\b(?:ac|aoc|AoC)SDHs?\b", "acute on chronic SDH", text)
    text = re.sub(r"\bcSDHs?\b", "chronic SDH", text)
    text = re.sub(r"\baSDHs?\b", "acute SDH", text)
    text = re.sub(r"\bsa?SDHs?\b", "subacute SDH", text)

    # words run together by the export: "thicknessof", "foracute", "Stableright"
    text = re.sub(r"[A-Za-z]{7,}", split_glued, text)

    # "flow of CSF.Subacute to chronic": split sentences before any rewrite
    # below changes the capital letter
    text = re.sub(r"(?<=[A-Za-z\)])\.(?=[A-Z])", ". ", text)

    # "|" is used as a line break inside sentences
    text = text.replace("|", " ")

    # glued words: "injuryTechnique", "2022There", "reviewFINDINGS", "2026.FINDINGS"
    text = re.sub(r"(?<=[a-z0-9])(?=[A-Z][a-z])", " ", text)
    text = re.sub(r"(?<=[a-z])(?=[A-Z]{3,})", " ", text)
    text = re.sub(r"(?<=\d)\.(?=[A-Z][A-Za-z])", ". ", text)

    # names put in by de-identification: "MERWIN-white differentiation",
    # "BASSITI-white matter differentiation" -> gray-white
    text = re.sub(r"\b[A-Za-z]+-white( matter)? diff ?erentiation", r"gray-white\1 differentiation", text)

    # glued side words: "rightinterhemispheric fissure", "lefttentorium"
    text = re.sub(
        r"(?i)\b(right|left)(?=(interhemispheric|tentorium|tentorial|falx|frontal|parietal|temporal|occipital|convexity|hemispheric))",
        r"\1 ", text)

    # "There are subacute and chronic components" = subacute on chronic
    text = re.sub(r"(?i)\b(acute|subacute|recent)\s+and\s+chronic\s+component", r"\1 on chronic component", text)

    # "subacute to chronic subdural hemorrhage" = subacute on chronic
    text = re.sub(r"(?i)\bsubacute\s+to\s+chronic\b", "subacute on chronic", text)

    # ---- general variations (audit) ----
    # units: "12 millimeters", "1.2 centimeters", "12-mm", "1.2-cm"
    text = re.sub(r"(?i)\bmillimet(?:er|re)s?\b", "mm", text)
    text = re.sub(r"(?i)\bcentimet(?:er|re)s?\b", "cm", text)
    text = re.sub(r"(?i)\b(\d+(?:\.\d+)?)\s*-\s*(mm|cm)\b", r"\1 \2", text)

    # abbreviations: "MLS", "cSDH", "aSDH", "acSDH"
    text = re.sub(r"\bMLS\b", "midline shift", text)

    # "left-to-right midline shift" = rightward
    text = re.sub(r"(?i)\b(?:left|right)\s*-?\s*to\s*-?\s*(left|right)\s+(midline\s+shift|shift)",
                  lambda m: m.group(1).lower() + "ward " + m.group(2).lower(), text)

    # upper-case units: "2 MM", "1.2 CM"
    text = re.sub(r"\b(\d+(?:\.\d+)?)\s*(MM|Mm|CM|Cm)\b", lambda m: m.group(1) + " " + m.group(2).lower(), text)

    # "No clear calvarial fracture": the hedge word is not a finding
    text = re.sub(r"(?i)\b(no|without)\s+((?:other|additional|further)\s+)?(?:clear|clearly|obvious|convincing)\s+(?=\w)", r"\1 \2", text)

    # bare "No shift" = no midline shift
    text = re.sub(r"(?i)\b(no)\s+(?:significant\s+)?shift\b(?!\s+of\s+the\s+midline)", r"\1 midline shift", text)

    # "acute on subacute/chronic" = acute on chronic; "acute-subacute" = acute on subacute
    text = re.sub(r"(?i)\bacute\s+on\s+subacute\s*/\s*chronic\b", "acute on chronic", text)
    text = re.sub(r"(?i)\bacute\s*-\s*subacute\b", "acute on subacute", text)

    # "there has been no interval change in the size of bilateral SDHs":
    # the "no" is about change, not the bleed
    text = re.sub(r"(?i)\b(?:there\s+(?:has|have)\s+been\s+)?no\s+(?:significant\s+)?(?:interval\s+)?change\s+in\s+(?:the\s+)?(?:size|appearance|thickness|extent|configuration)\s+of\s+(?:the\s+)?",
                  "stable ", text)

    # "the 3.1 cm in greatest depth subacute subdural hematoma": the size
    # said before the bleed goes after it
    text = re.sub(r"(?i)\b(\d+(?:\.\d+)?\s*(?:mm|cm))\s+in\s+((?:greatest|maximal|maximum|max)\s+)?(depth|thickness|width)\s+((?:[\w/-]+\s+){0,3}?)(subdural\s+(?:hematoma|hemorrhage|collection)s?)",
                  r"\4\5 measuring \1 in \2\3", text)

    # "small acute (high attenuation) component and larger subacute/chronic
    # (low attenuation) component": the larger low-density part dominates
    text = re.sub(r"(?i)\b(larger|large|major|dominant)\s+([\w/-]+\s+)?\(\s*(?:low\s+(?:attenuation|density)|hypodense|hypoattenuating)\s*\)\s*(components?)",
                  r"\1 predominantly hypodense \2\3", text)
    text = re.sub(r"(?i)\s*\(\s*(?:high|low|intermediate|mixed)\s+(?:attenuation|density)\s*\)", "", text)

    # "decreased attenuation along the left frontoparietal lobe ... in keeping
    # with the subdural hematoma" = hypodensity
    text = re.sub(r"(?i)\b(?:decreased|reduced|lower)\s+(?:density|attenuation)\b", "hypodensity", text)

    # "as seen on the recent MRI": "recent" is about the study, not the bleed's age
    text = re.sub(r"(?i)\b(?:more\s+)?recent\s+(?=(?:outside\s+)?(?:MRI|MR|CT|CTA|study|studies|exam\w*|scans?|imaging|images|comparison|head\s+CT)\b)", "prior ", text)

    # "No hyperdense areas noted" = no hyperdense component
    text = re.sub(r"(?i)\b(no)\s+(hyperdense|hypodense|isodense|high density|low density)\s+(?:areas?|foci|focus|regions?)\b",
                  r"\1 \2 component", text)

    # "no active bleeding" is not "no subarachnoid / intraventricular ... hemorrhage"
    text = re.sub(r"(?i)\bactive\s+(?:bleeding|hemorrhage|bleed|extravasation)\b", "active extravasation", text)

    # "flattening of the underlying cortical gyri" = sulcal effacement
    # ("underlying": the side is not said, as for "underlying sulcal effacement")
    text = re.sub(r"(?i)\b(?:flattening|effacement)\s+of\s+the\s+adjacent\s+(?:cortical\s+)?gyri\b", "adjacent sulcal effacement", text)
    text = re.sub(r"(?i)\b(?:flattening|effacement)\s+of\s+the\s+underlying\s+(?:cortical\s+)?gyri\b", "sulcal effacement", text)
    text = re.sub(r"(?i)\b(?:flattening|effacement)\s+of\s+the\s+(?:cortical\s+)?gyri\b|\bgyral\s+flattening\b", "sulcal effacement", text)

    # "Mild scalp swelling" is soft tissue swelling, not a scalp hematoma
    text = re.sub(r"(?i)\bscalp\s+(?:soft\s+tissue\s+)?(swelling|edema|thickening)\b|\b(swelling|edema|thickening)\s+of\s+the\s+scalp\b",
                  "soft tissue swelling", text)

    # "Abnormal texture of the skull bones (renal osteodystrophy)": not a fracture
    text = re.sub(r"(?i)\babnormal\s+(texture|appearance|mineralization|trabecular pattern)\b", r"altered \1", text)

    # "sinuses and mastoid air cells are clear (aside from trace mucosal
    # thickening in the ethmoid air cells)": clear, and only the named sinus
    # has the finding
    text = re.sub(r"(?i)\b(clear|well[\s-]aerated|unremarkable)\s*,?\s*\(?\s*(?:aside|apart)\s+from\s+([^.)]*)\)?",
                  r"\1. There is \2", text)
    text = re.sub(r"(?i)\b(clear|well[\s-]aerated)\s*,?\s*\(?\s*except\s+(?:for\s+)?([^.)]*)\)?", r"\1. There is \2", text)

    # "No significant abnormality is seen within the bony structures, orbits
    # and paranasal sinuses": each is unremarkable
    def each_unremarkable(m):
        items = re.split(r"\s*,\s*(?:and\s+)?|\s+and\s+", m.group(1).strip())
        items = [re.sub(r"(?i)^(?:the\s+)", "", i).strip() for i in items if i.strip()]
        return " ".join(f"The {i} are unremarkable." for i in items) + " "
    text = re.sub(r"(?i)\bno\s+(?:significant\s+|acute\s+)?abnormalit(?:y|ies)\s+(?:is\s+|are\s+)?(?:seen\s+|identified\s+|noted\s+)?(?:with)?in\s+the\s+((?:[\w-]+\s*,?\s*){1,12}?)(?:\s+as\s+seen[^.]*)?\.",
                  each_unremarkable, text)

    # "nonacute infarct" = chronic
    text = re.sub(r"(?i)\bnon-?\s?acute\b", "chronic", text)

    # "increased density along the anterior falx" = hyperdensity
    text = re.sub(r"(?i)\b(?:increased|higher)\s+(?:density|attenuation)\b", "hyperdensity", text)

    # both sides: "bifrontal", "bihemispheric", "on both sides", "both convexities"
    text = re.sub(r"(?i)\bbi-?(frontal|parietal|temporal|occipital|hemispheric|convexity)\b", r"bilateral \1", text)
    text = re.sub(r"(?i)\bon\s+both\s+sides\b", "bilaterally", text)
    text = re.sub(r"(?i)\bboth\s+(?=(?:cerebral\s+)?(?:convexit|hemispher|frontal|parietal|temporal|occipital))", "bilateral ", text)

    # size words: "sizable" = large
    text = re.sub(r"(?i)\bsize?able\b", "large", text)

    # "heterogeneous left subdural", "heterogeneously hyperdense": mixed density
    text = re.sub(r"(?i)\bheterogeneously\s+(?=(?:hyper|hypo|iso))", "mixed density ", text)
    text = re.sub(r"(?i)\bheterogeneous\s+(?=(?:left|right|bilateral|subdural|extra-?axial|collection|hematoma))", "heterogeneous density ", text)

    # "effacement of the adjacent sulci" = adjacent sulcal effacement (SDH side)
    text = re.sub(r"(?i)\b(?:(mild|moderate|severe|marked|minimal|significant|slight|extensive)\s+)?effacement\s+of\s+the\s+adjacent\s+(?:cerebral\s+)?sulci\b(\s*(?:,|as\s+well\s+as|and)\s+(?:of\s+)?the\s+)?",
                  lambda m: "adjacent " + (m.group(1) + " " if m.group(1) else "") + "sulcal effacement"
                  + ((" and " + (m.group(1) + " " if m.group(1) else "") + "effacement of the ") if m.group(2) else ""), text)

    # "mild adjacent sulcal effacement": the grade goes next to the effect
    text = re.sub(r"(?i)\b(mild|moderate|severe|marked|minimal|significant|slight|extensive)\s+(adjacent|underlying|regional|local)\s+(?=(?:cerebral\s+)?(?:sulc|effacement|mass))",
                  r"\2 \1 ", text)

    # "mild frontal and temporal sulcal effacement": the grade is the effacement's
    text = re.sub(r"(?i)\b(mild|moderate|severe|marked|minimal|significant|slight|extensive)\s+((?:(?:left|right|bilateral|frontal|parietal|temporal|occipital|frontoparietal|temporoparietal|and|,)\s*){1,6}?)(sulcal\s+effacement)",
                  r"\2 \1 \3", text)

    # earlier sizes are not the current size: "(previously 12 mm)",
    # "previously 1.3 cm", "increased from 8 mm to 12 mm"
    text = re.sub(r"(?i),?\s*\(?\s*(?:previously|formerly|prior|compared (?:to|with))\s+(?:measuring\s+|measured\s+)?\d+(?:\.\d+)?\s*(?:mm|cm)\s*\)?", "", text)
    text = re.sub(r"(?i)\bfrom\s+\d+(?:\.\d+)?\s*(?:mm|cm)?\s+to\s+(?=\d)", "to ", text)

    # negation said after the finding: "Hydrocephalus is not present",
    # "Midline shift is not seen", "Mass effect is absent" -> "No ..."
    text = re.sub(
        r"(\A|[.;:]\s+)([A-Z][\w-]*(?:\s+[\w-]+){0,5}?)\s+(?:is|are)\s+(?:(?:not|no\s+longer)\s+(?:present|seen|identified|demonstrated|evident|appreciated|visualized|visible|noted)|absent)\b",
        lambda m: m.group(1) + "No " + m.group(2)[0].lower() + m.group(2)[1:], text)

    # a finding that has resolved is not there = No: "Previous left subdural
    # hematoma has resolved" -> "No left subdural hematoma", "Resolved
    # midline shift", "Interval resolution of the midline shift" (but not
    # "partially resolved")
    def resolved_to_no(m):
        sentence = m.group(0)
        if re.search(r"(?i)partial|near|residual|incomplete|largely|mostly|decreas|persist|remain", sentence):
            return sentence
        finding = m.group("finding").split(",")[0]
        finding = re.sub(r"(?i)^\s*(?:the\s+)?(?:(?:previously|previous|prior|earlier)\s+(?:seen\s+|noted\s+|described\s+|identified\s+)?)?(?:the\s+)?", "", finding).strip()
        if not finding:
            return " "
        return " No " + finding[0].lower() + finding[1:] + "."
    text = re.sub(r"(?i)(?P<finding>[^.:]*?)\s+(?:has|have)\s+(?:now\s+)?(?:completely\s+)?resolved\b[^.]*\.", resolved_to_no, text)
    text = re.sub(r"(?i)(?<=[.:]\s)(?:interval\s+|complete\s+)?(?:resolution\s+of|resolved)\s+(?P<finding>[^.]*)\.", resolved_to_no, text)

    # British spelling
    text = re.sub(r"(?i)\boedema\b", "edema", text)

    # "No intra-axial hemorrhage": inside the brain = parenchymal and
    # ventricular, not subarachnoid
    text = re.sub(r"(?i)\b(no|without)\s+((?:\w+\s+){0,2}?)intra-?\s?axial\s+(hemorrhages?|hematomas?|bleed\w*|blood)",
                  r"\1 \2intraparenchymal \3 or intraventricular \3", text)

    # "subarachnoid and subdural hemorrhage": both are bleeds
    text = re.sub(r"(?i)\b(subarachnoid|intraventricular|intraparenchymal|epidural|extradural|subdural)\s+(and|or|/)\s+(subdural|subarachnoid|intraventricular|intraparenchymal|epidural|extradural)\s+(hemorrhages?|hematomas?|blood|bleed\w*)",
                  r"\1 \4 \2 \3 \4", text)

    # "predominantly acute subdural hematoma": the acute part dominates, so
    # it is predominantly hyperdense too
    text = re.sub(r"(?i)\b(predominantly|predominately|mostly|largely|mainly)\s+(hyper-?acute|acute)\b(?!\s+(?:on|to|and)\b)",
                  r"\1 hyperdense \2", text)

    # "small acute superimposed components": the components' size
    text = re.sub(r"(?i)\b(acute|hyperdense)\s+superimposed\s+(components?)", r"\1 \2", text)

    # "There is a right subdural hematoma. It tracks along the right temporal
    # lobe": a pronoun sentence about where the SDH just named goes
    sentences = re.split(r"(?<=\.)\s+", text)
    for k in range(1, len(sentences)):
        before = sentences[k - 1].lower()
        if (
            re.search(r"subdural|\bsdh|extra-?\s?axial", before)
            and not re.search(r"epidural|subarachnoid|\bsah\b|contusion|intraparenchymal|intraventricular|scalp|subgaleal", before)
        ):
            sentences[k] = re.sub(
                r"^(?:It|This|This collection|This hematoma)\s+(?=(?:extends|tracks|layers|lies|spans|courses|wraps|overlies|involves|has moved|is located|is centered|is seen)\b)",
                "The hematoma ", sentences[k])
    text = " ".join(sentences)

    # findings outside the head: a parotid / lung / thyroid mass is not an
    # intracranial mass; an epidural hematoma in the spinal canal is not
    # the head's
    extracranial = r"(?:parotid|lung|thyroid|neck|nasopharyn\w*|oropharyn\w*|cervical lymph|scalp|subcutaneous)"
    text = re.sub(r"(?i)\bmass(?:es)?\b(?![\s-]+effect)(?=(?:\s+[\w-]+){0,5}?\s+" + extracranial + r"\b)", "extracranial lesion", text)
    text = re.sub(r"(?i)(\b" + extracranial + r"\b(?:\s+[\w-]+){0,3}?\s+)mass(?:es)?\b(?![\s-]+effect)", r"\1extracranial lesion", text)
    text = re.sub(r"[^.]*\b(?:spinal canal|cervical spine|spinal|vertebra\w*|prevertebral|cord)\b[^.]*",
                  lambda m: re.sub(r"(?i)\bepidural\s+(hematoma|hemorrhage|collection)s?\b", "spinal canal finding", m.group(0)), text)

    # "more likely a contusion than subarachnoid hemorrhage": the option
    # not chosen is not recorded
    text = re.sub(r"(?i)(\b(?:more|most)\s+likely\b[^.]*?)\s+(?:rather\s+)?than\s+[^.,;]*", r"\1", text)
    text = re.sub(r"(?i)\s+rather\s+than\s+[^.,;]*", "", text)

    # "No other extra-axial hemorrhage" (an SDH is described): no SAH / EDH
    text = re.sub(r"(?i)\bno\s+(?:other|additional|further)\s+extra-?\s?axial\s+(?:or\s+(intraparenchymal|intra-?\s?axial)\s+)?(?:hemorrhages?|blood|collections?|hematomas?)",
                  lambda m: "No subarachnoid hemorrhage or epidural hematoma"
                  + (" or " + m.group(1) + " hemorrhage" if m.group(1) else ""), text)

    # "No orbital abnormality", "No concerning finding within the visualized orbits"
    text = re.sub(r"(?i)\bno\s+(?:acute\s+)?orbital\s+(?:abnormalit(?:y|ies)|findings?|injury|lesions?)\b", "The orbits are unremarkable", text)
    text = re.sub(r"(?i)\bno\s+(?:acute\s+|concerning\s+|significant\s+)?(?:abnormalit(?:y|ies)|findings?)\s+(?:with)?in\s+the\s+(?:visualized\s+|imaged\s+)?orbits\b", "The orbits are unremarkable", text)

    # "large artery territory ischemic cortical infarct": one infarct
    text = re.sub(r"(?i)\bischemic\s+(?=(?:cortical\s+|territorial\s+)?infarct)", "", text)

    # "moderate atrophy and microangiopathic disease": the grade is said
    # once for both graded findings (a second grade of its own still wins)
    graded = (r"(?:generalized\s+|diffuse\s+|cerebral\s+|brain\s+|global\s+|parenchymal\s+)?(?:atrophy|volume\s+loss)"
              r"|(?:chronic\s+)?(?:periventricular\s+)?(?:white\s+matter\s+)?(?:microangiopathic|small\s+vessel(?:\s+ischemic)?)\s+(?:changes?|disease)"
              r"|hydrocephalus")
    text = re.sub(r"(?i)\b(mild|moderate|severe|marked|minimal|extensive|significant)\s+(" + graded + r")\s+and\s+(?=(?:" + graded + r"))",
                  r"\1 \2 and \1 ", text)

    # differential lists: "Periventricular hypodensities ... may be due to
    # infarctions, small vessel ischemic or other white matter disease":
    # microangiopathy named as a possible cause counts (Yes NS); an infarct
    # that is only one option in the list is not recorded
    infarct_option = r"(?i)\b(?:(?:acute|chronic|old|ischemic|lacunar)\s+)?infarct(?:ion)?s?\b,?"
    def differential(m):
        before, cue, options = m.group(1), m.group(2), m.group(3)
        options = re.sub(r"(?i)\bsmall\s+vessel\s+ischemic(?!\s+(?:disease|changes?))", "small vessel ischemic disease", options)
        return before + cue + re.sub(infarct_option, "", options)
    text = re.sub(r"(?i)([^.]*?)(\b(?:may|might|could)\s+(?:be\s+due\s+to|represent|reflect)\b)([^.]*\bor\b[^.]*\.)", differential, text)
    text = re.sub(r"(?i)([^.]*?)(\b(?:differential(?:\s+diagnosis)?\s+includes?)\s)([^.]*\.)", differential, text)
    # "infarct versus artifact", "artifact versus infarct"
    text = re.sub(r"(?i)\b(?:(?:acute|chronic|old|ischemic|lacunar)\s+){0,2}infarct(?:ion)?s?\s+(?=(?:versus|vs\.?)\s)", "", text)
    text = re.sub(r"(?i)\b((?:versus|vs\.?)\s+)(?:(?:acute|chronic|old|ischemic|lacunar)\s+){0,2}infarct(?:ion)?s?\b", r"\1", text)

    # orbital fat / muscle "herniation" (orbital floor fracture) is not brain herniation
    text = re.sub(r"[^.]*\b(?:orbit\w*|rectus|globe|fat|muscle)\b[^.]*",
                  lambda m: re.sub(r"(?i)\bherniat(?:ion|ing|ed)\b", "protrusion", m.group(0)), text)

    # "low/intermediate density" = low density and intermediate density
    text = re.sub(r"(?i)\b(low|high|intermediate|iso)\s*/\s*(low|high|intermediate|iso)\s+(density|attenuation)\b",
                  r"\1 \3 and \2 \3", text)

    # "heterogeneously dense" = heterogeneous density (mixed)
    text = re.sub(r"(?i)\bheterogeneously\s+(dense|attenuating)\b", "heterogeneous density", text)

    # "crescentic abnormality ... within this abnormality": an extra-axial collection
    text = re.sub(r"(?i)\bcrescentic\s+(abnormality|lesion|density|fluid)\b", "crescentic collection", text)

    # "Acute-hyperacute" -> acute (not a density word)
    text = re.sub(r"(?i)\bacute\s*-\s*hyper\s*-?\s*acute\b", "acute", text)
    text = re.sub(r"(?i)\bhyper\s*-?\s*acute\b", "acute", text)

    # an acute part in an existing hematoma is acute on chronic:
    # "interval acute hemorrhage", "acute blood within the hematoma",
    # "a small acute and a larger subacute/chronic component"
    text = re.sub(r"(?i)\b(?:a\s+)?(?:small|smaller|large|larger|minor)\s+(acute|hyperdense)\s+and\s+(?:a\s+)?(?:larger|smaller|large|small|major)\s+(subacute\s*/\s*chronic|subacute|chronic)\s+components?\b",
                  r"acute component and \2 component", text)
    text = re.sub(r"(?i)\b(?:interval|superimposed|new)\s+acute\s+(?:hemorrhage|blood|bleeding|bleed)\b",
                  "acute on chronic component", text)
    text = re.sub(r"(?i)\bacute\s+(?:blood|hemorrhage)\s+(?=(?:with)?in\s+(?:the|this)\s+(?:hematoma|collection|subdural))",
                  "acute on chronic component ", text)

    # "Mass effect remains minimal": the grade said after the effect
    # (keep a capital first letter so the sentence still starts there)
    text = re.sub(r"(?i)\b(mass[\s-]effect|midline shift|sulcal effacement)\s+(?:remains|is|appears|is now)\s+(minimal|mild|moderate|severe|marked|slight)\b",
                  lambda m: (m.group(2).capitalize() + " " + m.group(1).lower()) if m.group(1)[0].isupper()
                  else (m.group(2) + " " + m.group(1)), text)

    # "global near complete left cerebral sulcal effacement": the grade
    # reaches the effect
    text = re.sub(r"(?i)\b(left|right|bilateral)\s+cerebral\s+(sulc\w*)", r"\1 \2", text)

    # "Please note conventional angiogram is more sensitive to assess for
    # aneurysms": a sentence about the method, not a finding
    text = re.sub(r"(?i)[^.]*\b(?:more sensitive|to assess for|to evaluate for|to exclude)\b[^.]*\.", " ", text)

    # "no circle of Willis aneurysm"
    text = re.sub(r"(?i)\bcircle\s+of\s+willis\s+(aneurysms?)", r"\1", text)

    # "extensive consequent mass-effect"
    text = re.sub(r"(?i)\b(extensive|severe|marked|significant|mild|moderate|minimal)\s+(?:consequent|resultant|resulting)\s+(mass[\s-]effects?)",
                  r"\1 \2", text)

    # "1 to 2 mm increase in right midline shift": a change, not the size
    text = re.sub(r"(?i)\b\d+(?:\.\d+)?(?:\s*(?:to|-)\s*\d+(?:\.\d+)?)?\s*(?:mm|cm)\s+((?:interval\s+)?(?:increase|decrease|change|progression))\s+in\b",
                  r"\1 in", text)

    # "the septum pellucidum is deviated to the right by 11 mm" = midline shift
    text = re.sub(r"(?i)\b(?:the\s+)?septum\s+pellucidum\s+is\s+(?:deviated|shifted|displaced)\s+(to\s+the\s+(?:left|right))\s+by\s+(\d+(?:\.\d+)?\s*(?:mm|cm))",
                  r"\2 midline shift \1", text)

    # "Ventricles: Within normal limits." / "Midline: Central."
    text = re.sub(r"(?i)\bventricles\s*:\s*(within normal limits|normal|unremarkable)", "The ventricles are normal in size", text)
    text = re.sub(r"(?i)\bmidline\s*:\s*(central|midline|normal)", "No midline shift", text)

    # "expansion of the temporal horn of the right lateral ventricle" = prominence
    text = re.sub(r"(?i)\bexpansion\s+of\s+the\s+((?:[\w-]+\s+){0,7}?(?:ventricles?|horns?))\b", r"prominence of the \1", text)

    # "C6 lateral masses": vertebral, not a mass
    text = re.sub(r"(?i)\blateral\s+masses?\b", "lateral vertebral elements", text)

    # "No tonsilar herniation"
    text = re.sub(r"(?i)\btonsilar\b", "tonsillar", text)

    # "soft tissues of the neck": not the head
    text = re.sub(r"(?i)\bsoft\s+tissues?\s+of\s+the\s+neck\b", "neck", text)

    # "Right maxillary mucous retention cyst", "thickening in the right
    # maxillary, bilateral frontal and ethmoid sinuses": the maxillary sinus
    text = re.sub(r"(?i)\b(maxillary|sphenoid)\b(?!\s+(?:sinus|sinuses|antrum|molar|tooth|teeth|bone|fracture|wall|process|division|nerve))(?=[^.]*\b(?:sinus|sinuses|retention cyst)\b)",
                  r"\1 sinus", text)

    # "5 mm shift of midline towards the right" = midline shift
    text = re.sub(r"(?i)\bshift\s+of\s+(the\s+)?midline\b(\s+structures)?", "midline shift", text)

    # "fairly low in density" = low density
    text = re.sub(r"(?i)\b(low|high|intermediate)\s+in\s+(density|attenuation)\b", r"\1 \2", text)

    # "grey white differentiation" = gray-white differentiation
    text = re.sub(r"(?i)\bgr[ae]y[\s-]+white(\s+matter)?\s+differentiation", r"gray-white\1 differentiation", text)

    # "a moderate degree of underlying sulcal effacement and mass effect":
    # the grade is said once for both
    text = re.sub(
        r"(?i)\b(mild|moderate|severe|marked|significant)\s+degree\s+of\s+(underlying\s+)?((?:\w+\s+)?effacement)\s+and\s+(mass\s+effect)",
        r"\1 \3 and \1 \4", text)
    text = re.sub(r"(?i)\b(mild|moderate|severe|marked|significant)\s+degree\s+of\s+", r"\1 ", text)


    # "acute on subacute/chronic" = acute on chronic
    text = re.sub(r"(?i)\bacute on subacute\s*/\s*chronic", "acute on chronic", text)

    # "right- sided", "extra- axial": a line break after the hyphen
    text = re.sub(r"(?i)\b(right|left|extra|intra)-\s+(sided|axial)\b", r"\1-\2", text)

    # "high left frontal parafalcine region": the lobe says where on the falx
    text = re.sub(r"(?i)\bfrontal\s+(para-?falcine|falcine|falx)\b", r"anterior \1", text)
    text = re.sub(r"(?i)\b(parietal|occipital)\s+(para-?falcine|falcine|falx)\b", r"posterior \2", text)

    # "fracture of the anterior nasal spine" is a facial bone, not the spine
    text = re.sub(r"(?i)\bnasal spine\b", "nasal process", text)

    # "The midline is intact" says there is no midline shift
    text = re.sub(r"(?i)\bthe midline is intact\b", "No midline shift", text)

    # "may reflect a more acute component", "a trace of more acute blood"
    # (inside an older collection) = acute on chronic
    text = re.sub(r"(?i)\bmore acute\b", "acute on chronic", text)

    # "an acute bleed on a pre-existing subdural hematoma" = acute on chronic
    text = re.sub(
        r"(?i)\bacute(\s+(bleed|hemorrhage|haemorrhage|blood))?\s+on\s+(an?\s+)?(pre-?\s?existing|underlying|older|background)",
        "acute on chronic", text)

    # "a trace of separate SDH", "a trace of acute blood": the trace is a
    # part of the bleed, not the size of the hematoma
    text = re.sub(
        r"(?i)\b(a\s+)?trace\s+(amount\s+)?of\s+(?=(acute|separate|hyperdense|high|new|fresh)\b)", "", text)

    # "patchy areas of hyperintense changes": patches of high density in a
    # lower density collection = mixed density
    text = re.sub(r"(?i)\bpatchy\s+(areas?\s+of\s+)?(?=(hyper|high))", r"mixed density with \1", text)

    # "mixture of high and low density" -> both densities
    text = re.sub(
        r"(?i)\b(high|low)\s+and\s+(high|low)\s+(density|attenuation)",
        r"\1 \3 and \2 \3", text)

    # "the collection is smaller compared to previous" is a change, not a
    # size ("a similar smaller collection" stays a size)
    text = re.sub(r"(?i)\b(is|are|now|has become|have become)\s+smaller\b", r"\1 decreased", text)
    text = re.sub(r"(?i)\bsmaller\s+(?=(than|compared|since|in size)\b)", "decreased ", text)

    # "left larger than right", "right greater than left": a comparison of
    # two sides already named, not a new side
    text = re.sub(
        r"(?i)\b(left|right)\s*(larger|greater|bigger|more|>)\s*(than\s*)?(left|right)\b", "", text)

    # "ex vacuo prominence / dilation / enlargement" -> one wording
    text = re.sub(r"(?i)\bex[\s-]+vacuo\s+(prominence|dilatation|dilation|enlargement|ventriculomegaly)\b",
                  "ex vacuo dilatation", text)

    # "no CT findings for acute infarct" = "no evidence of acute infarct"
    # ("findings" is also a section heading word)
    text = re.sub(r"(?i)\bno\s+(ct\s+|imaging\s+|radiographic\s+)?(findings|signs?)\s+(for|of|to suggest)\b",
                  "no evidence of", text)

    # "no worsening mass effect or midline shift": it is there, just not worse
    text = re.sub(r"(?i)\bno\s+(significant\s+)?(worsening|increase|increased|progression|interval\s+increase)\s+(in\s+|of\s+)?(the\s+)?",
                  "stable ", text)

    # "significant subfalcine, uncal and tonsillar herniation": the grade is
    # for every herniation in the list
    text = re.sub(r"(?i)\btonsiallar\b", "tonsillar", text)
    def share_grade(m):
        grade, items = m.group(1), m.group(2)
        parts = re.split(r"(\s*,\s*|\s+and\s+|\s+or\s+)", items)
        return " ".join(
            p if re.fullmatch(r"\s*,\s*|\s+and\s+|\s+or\s+", p) else f"{grade} {p.strip()}"
            for p in parts
        ).replace(" ,", ",") + " herniation"
    text = re.sub(
        r"(?i)\b(significant|severe|mild|moderate|marked|early|minimal)\s+"
        r"((?:subfalcine|uncal|transtentorial|tonsillar|descending transtentorial)"
        r"(?:(?:\s*,\s*|\s+and\s+|\s+or\s+)(?:subfalcine|uncal|transtentorial|tonsillar|descending transtentorial))+)"
        r"\s+herniations?\b",
        share_grade, text)

    # size ranges keep the larger: "moderate to large" -> large
    text = re.sub(r"(?i)\b(small|moderate|medium)(\s*-?\s*sized)?\s+to\s+large\b", "large", text)
    text = re.sub(r"(?i)\bsmall\s+to\s+(moderate|medium)(\s*-?\s*sized)?\b", "moderate", text)

    # "unremarkable for significant abnormality" = unremarkable
    text = re.sub(r"(?i)\bunremarkable\s+for\s+(any\s+)?(significant\s+)?abnormality\b", "unremarkable", text)

    # "encephalomalacia ... from remote trauma or ischemia": a cause, not an infarct
    text = re.sub(r"(?i)\b(from|due to|secondary to|related to)\s+(remote\s+|prior\s+)?(trauma\s+or\s+)?(ischemia|ischemic insult)\b",
                  r"\1 prior insult", text)

    # "Both of these subdural collections are isoattenuating" = both sides
    text = re.sub(r"(?i)\bboth\s+of\s+these\b", "bilateral", text)
    text = re.sub(r"(?i)\bboth\s+(?=(subdural\s+)?(collections|hematomas|haematomas|subdurals)\b)", "bilateral ", text)

    # "It was 14 mm and is now 15 mm thick": the size now
    text = re.sub(r"(?i)\b(it|this|the collection|the hematoma)\s+(was|measured)\s+[\d.]+\s*(mm|cm)\s+and\s+(is\s+)?now\s+",
                  "this now measures ", text)

    # "microangio pathic" (split word); "chronic ischemic microangiopathic
    # changes" are microangiopathy, not an infarct
    text = re.sub(r"(?i)\bmicroangio\s+pathic\b", "microangiopathic", text)
    text = re.sub(r"(?i)\bischemic\s+(?=(microangiopathic|small vessel|microvascular)\b)", "", text)

    # "No intra or extra-axial hemorrhage"
    text = re.sub(r"(?i)\bintra\s+or\s+extra-?\s?axial\s+(hemorrhage|bleed|blood)", r"intracranial \1", text)

    # "3rd ventricle" -> "third ventricle"
    text = re.sub(r"(?i)\b3rd\b", "third", text)
    text = re.sub(r"(?i)\b4th\b", "fourth", text)

    # "13 mm subfalcine midline shift" is subfalcine herniation
    text = re.sub(r"(?i)\bsubfalcine\s+(midline\s+)?shift\b", "subfalcine herniation", text)

    # "rightward shift of midline structures x 3 mm" = midline shift of 3 mm
    text = re.sub(r"(?i)\bshift\s+of\s+(the\s+)?midline\s+structures\b", "midline shift", text)
    text = re.sub(r"(?i)\b(midline shift)\s+x\s+(?=\d)", r"\1 of ", text)

    # "No cerebellar herniation" = no tonsillar herniation
    text = re.sub(r"(?i)\bcerebellar\s+(tonsillar\s+)?herniations?\b", "tonsillar herniation", text)

    # "... no new infarct although minimal ... changes are present":
    # the clause after "although" is not negated
    text = re.sub(r"(?i)\s+although\s+", ". ", text)

    # list items: "COMPONENT.-RIGHT: SMALL" -> "COMPONENT. RIGHT: SMALL"
    text = re.sub(r"\.\s*-\s*(?=[A-Za-z])", ". ", text)

    # "FINDINGS:There" -> "FINDINGS: There"
    text = re.sub(r":(?=[A-Za-z])", ": ", text)

    # "1 cm.There" -> "1 cm. There"   (not decimals)
    text = re.sub(r"(?<=[A-Za-z\)])\.(?=[A-Z])", ". ", text)

    # numbered impression lines "1. Large ..." -> "Large ..."
    text = re.sub(r"(?<![\d.])\b\d\.\s+(?=[A-Z])", " ", text)

    text = re.sub(r"[ \t]+", " ", text)

    # "tentor ium" -> "tentorium" first, then the rewrites that need whole words
    text = repair_split_words(text)
    # a first letter split off by the export: "c omponents", "h ematoma"
    text = re.sub(r"(?i)\b(c)\s+(omponents?)\b|\b(h)\s+(ematomas?)\b|\b(s)\s+(ubdural)\b",
                  lambda m: "".join(g for g in m.groups() if g), text)
    text = re.sub(r"(?i)\b(acute|hyperdense)\s+superimposed\s+(components?)", r"\1 \2", text)

    # "Small amount of blood in left tentorium": blood along the falx /
    # tentorium is subdural blood there (not after "subarachnoid")
    def compartment_blood(m):
        before = text[max(0, m.start() - 25):m.start()].lower()
        if re.search(r"(subarachnoid|intraventricular|sah)\s*$", before):
            return m.group(0)
        size = (m.group(1) + " ") if m.group(1) else ""
        return f"{size}subdural hematoma {m.group(2)} the {m.group(3) or ''}{m.group(4)}"
    text = re.sub(
        r"(?i)\b(?:(small|tiny|trace|minimal|mild)\s+(?:amount|volume)\s+of\s+)?blood\s+(?:is\s+)?(in|along|within|over)\s+(?:the\s+)?((?:left|right)\s+)?(tentorium|tentorial\s+leaflet|falx)\b",
        compartment_blood, text)

    return text


# ============================
# SECTIONS
# ============================

# sentences under these headings are history / technique, not findings
SKIP_SECTIONS = {
    "clinical history", "history", "clinical indication", "indication",
    "clinical information", "comparison", "comparison study", "technique",
    "procedure", "reference examination", "reference examinations",
    "exam", "contrast",
}


KEEP_SECTIONS = {
    "findings", "imaging findings", "impression", "imaging result", "imaging_result",
    "opinion", "summary", "conclusion", "interpretation",
}


def heading_text(sent_tokens):
    """
    A heading sentence ends in ':' e.g. [("FINDINGS", ...), (":", "colon")].
    It is either a known heading (possibly glued to the end of the
    previous section's text, "... 2009 Imaging Findings :") or a short
    run of words ("CT HEAD:", "Midline:").
    """

    if not sent_tokens or sent_tokens[-1][1] != "colon":
        return None

    words = [text for text, bucket in sent_tokens[:-1] if text not in ("-", "*", "|")]

    if not words:
        return None

    joined = " ".join(words).lower().strip()

    for name in SKIP_SECTIONS | KEEP_SECTIONS:
        if joined == name or joined.endswith(" " + name):
            return name

    if len(words) <= 4:
        return joined

    return None


def reports_finding(sent_tokens):

    buckets = [bucket for _, bucket in sent_tokens]

    if buckets and buckets[0] == "noticing":
        return True

    has_number = "numerical" in buckets and "measurement_unit" in buckets
    has_finding = "bleed_pathology" in buckets or any(b in EFFECT_FAMILY for b in buckets)

    return has_number and has_finding


# these sections are a single line ("COMPARISON: NONE.") - the next
# sentence is findings even when no "FINDINGS:" heading follows
ONE_SENTENCE_SECTIONS = {
    "comparison", "comparison study", "technique", "procedure", "contrast",
    "reference examination", "reference examinations",
}

SIDE_HEADINGS = {"right": "right", "left": "left", "on the right": "right", "on the left": "left"}


def filter_sections(sentence_lists):

    kept = []
    skipping = False
    one_sentence_left = False
    side_label = None           # "-RIGHT: SMALL. 10 MM THICKNESS"

    for sent_tokens in sentence_lists:

        heading = heading_text(sent_tokens)

        if heading is not None and heading.strip("- ") in SIDE_HEADINGS:
            side_label = SIDE_HEADINGS[heading.strip("- ")]
            continue

        if heading is not None:

            skipping = heading in SKIP_SECTIONS
            one_sentence_left = heading in ONE_SENTENCE_SECTIONS
            side_label = None

            continue

        if skipping and one_sentence_left:
            one_sentence_left = False
            skipping = False
            continue

        # every sentence under a side label says which side it is about
        if side_label and not skipping:
            sent_tokens = [(side_label, "laterality")] + list(sent_tokens)

        # a history / technique section ends early when a sentence
        # clearly reports a finding ("There is ...", "Largely isodense
        # left SDH ... measuring 1.7 cm") - headings are often missing
        if skipping and reports_finding(sent_tokens):
            skipping = False

        if not skipping:
            kept.append(sent_tokens)

    return kept


# ============================
# FULL PIPELINE
# ============================

def tokenize_report(report):

    doc = nlp(preprocess_report(report))

    sentence_lists = []

    for sent in doc.sents:

        sent_tokens = []

        for token in sent:
            bucket = get_bucket(token.text)
            sent_tokens.append((token.text, bucket))

        sentence_lists.append(sent_tokens)

    for sent_tokens in sentence_lists:

        for idx, (text, bucket) in enumerate(sent_tokens):

            if text.lower() == "and":

                sent_tokens[idx] = (
                    "and",
                    classify_and(sent_tokens, idx)
                )

        mark_dominance(sent_tokens)

    return filter_sections(sentence_lists)


def mark_dominance(sent_tokens):
    """
    "predominantly hypo to isodense", "mostly hyperdense with areas of
    lower density": the density words right after the cue word, joined
    only by to / and / -, are the dominant ones. They are tagged
    "dominant:<word>" (density_filter reads through the tag).
    """

    for idx, (text, bucket) in enumerate(sent_tokens):

        if not (bucket == "density" and text.lower() in DOMINANCE_WORDS):
            continue

        j = idx + 1
        marked = False

        while j < len(sent_tokens):

            t_text, t_bucket = sent_tokens[j]

            # "The majority of the subdural is low-density": a few words
            # may sit between the cue and the density
            if (
                not marked and j - idx <= 4
                and (t_bucket in (None, "bleed_pathology", "bleed_reference", "noticing")
                     or t_text.lower() in ("is", "are", "the", "collection", "hematoma"))
            ):
                j += 1
                continue

            if t_bucket == "density" and t_text.lower() not in DOMINANCE_WORDS:
                marked = True
                if not t_text.lower().startswith("dominant:"):
                    sent_tokens[j] = (f"dominant:{t_text}", t_bucket)
                j += 1
                continue

            if t_bucket == "sandwich_and" or t_text.lower() in ("to", "-", "/", "and", "or"):
                j += 1
                continue

            break


def run_report(report):

    sentence_lists = tokenize_report(report)

    bleed_trees = [orchestrator(sentence_lists)]

    return master_variable_mapping_function(bleed_trees)
```

## VARIABLE MAPPING

### Master Mapping function

```python
def master_variable_mapping_function(
    bleed_trees
):

    final_variables = {}

    # side of the SDH (used for "ipsilateral" and side-less sulcal effacement)
    sdh_side = None

    for tree_group in bleed_trees:

        for tree in tree_group:

            if (
                isinstance(tree, dict)
                and "pathology" in tree
                and tree.get("pathology")
            ):
                sdh_side = summarize_sdh_side(tree["children"])

    for tree_group in bleed_trees:

        for tree in tree_group:

            if not tree:
                continue

            all_mapped_variables = []
            all_direct_variables = {}

            # -------------------------
            # SDH Tree
            # -------------------------

            if "pathology" in tree:

                sdh_children = []
                association_children = []

                for child in tree["children"]:

                    if (
                        isinstance(child, dict)
                        and "association" in child
                    ):

                        # -------------------------
                        # Pull SDH characteristics
                        # back into sdh_children
                        # (in place, so the side they were said on is kept)
                        # -------------------------

                        previous_side = None

                        for association in child["association"]:

                            # "mass effect on the left hemisphere with
                            # effacement of the hemispheric sulci"
                            if "effect_name" in association:
                                sides_here = [
                                    t for t, b in association.get("location", [])
                                    if b == "laterality"
                                ]
                                if sides_here:
                                    previous_side = sides_here[0]
                                elif previous_side and any(
                                    b == "sulci" for _, b in association.get("location", [])
                                ):
                                    association = dict(association)
                                    association["location"] = (
                                        association["location"] + [(previous_side, "laterality")]
                                    )

                            if "density" in association:

                                sdh_children.extend(
                                    association["density"]
                                )

                            elif "acuity" in association:

                                sdh_children.extend(
                                    association["acuity"]
                                )

                            elif "morphology" in association:

                                sdh_children.extend(
                                    association["morphology"]
                                )

                            elif "thickness" in association:

                                sdh_children.append(
                                    {
                                        "thickness":
                                        association["thickness"]
                                    }
                                )

                            elif (
                                "negation" in association
                                and any("density" in n for n in association["negation"])
                            ):

                                sdh_children.append(association)

                            else:

                                association_children.append(
                                    [association]
                                )

                        continue

                    sdh_children.append(
                        child
                    )

                # -------------------------
                # SDH Characteristics
                # -------------------------

                (
                    sdh_mapped_variables,
                    sdh_direct_variables
                ) = SDH_characteristics(
                    sdh_children
                )

                # -------------------------
                # SDH Associations
                # -------------------------

                (
                    assoc_mapped_variables,
                    assoc_direct_variables
                ) = SDH_association_characteristics(
                    association_children,
                    sdh_side
                )

                all_mapped_variables.extend(
                    sdh_mapped_variables
                )

                all_mapped_variables.extend(
                    assoc_mapped_variables
                )

                all_direct_variables.update(
                    sdh_direct_variables
                )

                all_direct_variables.update(
                    assoc_direct_variables
                )

            # -------------------------
            # Negation Tree (a list of negated heads)
            # -------------------------

            elif "negation" in tree:

                negations = tree["negation"]

                if isinstance(negations, dict):
                    negations = [negations]

                for negation in negations:

                    (
                        mapped_variables,
                        direct_variables
                    ) = negation_sorter(
                        negation,
                        sdh_side
                    )

                    all_mapped_variables.extend(
                        mapped_variables
                    )

                    all_direct_variables.update(
                        direct_variables
                    )

            # -------------------------
            # Outcome Tree
            # -------------------------

            elif "outcome" in tree:

                (
                    mapped_variables,
                    direct_variables
                ) = outcome_function(
                    tree["outcome"]
                )

                all_mapped_variables.extend(
                    mapped_variables
                )

                all_direct_variables.update(
                    direct_variables
                )

            # -------------------------
            # Other hemorrhage (SAH / IPH / IVH)
            # -------------------------

            elif "mapped" in tree:

                all_mapped_variables.extend(tree["mapped"])

            elif "hemorrhage" in tree:

                all_mapped_variables.extend(
                    hemorrhage_variables(tree["hemorrhage"]["type"], tree["hemorrhage"].get("grade"))
                )

            # -------------------------
            # Sinus / mastoid
            # -------------------------

            elif "sinus" in tree:

                all_direct_variables.update(
                    tree["sinus"]
                )

            elif "fracture" in tree:

                all_direct_variables.update(
                    tree["fracture"]
                )

            else:

                continue

            # -------------------------
            # Final Mapping
            # -------------------------

            final_array_function(
                all_mapped_variables,
                all_direct_variables,
                final_variables
            )

    if not final_variables:
        final_array_function([], {}, final_variables)

    final_variables = post_process_variables(final_variables)

    # sides of epidural bleeds, for the review flag
    final_variables["_epidural_sides"] = [
        tree["hemorrhage"].get("side")
        for tree_group in bleed_trees
        for tree in tree_group
        if isinstance(tree, dict) and tree.get("hemorrhage", {}).get("type") == "epidural"
    ]

    return final_variables


def summarize_sdh_side(children):
    """
    "L" / "R" / "B" for the SDH, from the lateralities said in the
    bleed sentences.
    """

    named = set()

    for child in children:

        if isinstance(child, tuple) and child[1] == "laterality":
            side = side_of(child[0])
            if side:
                named.add(side)

        elif isinstance(child, dict) and "location" in child:
            for text, bucket in child["location"]:
                if bucket == "laterality" and side_of(text):
                    named.add(side_of(text))

    if "B" in named or {"L", "R"} <= named:
        return "B"

    if named:
        return next(iter(named))

    return None


def hemorrhage_grade(sent_tokens, idx=None):
    """
    "scattered areas of ... subarachnoid hemorrhage" -> Scattered,
    "minor / small volume / trace subarachnoid blood" -> Mild,
    "moderate" -> Moderate, "diffuse / extensive / large" -> Severe
    """

    # whole words only ("within" is not "thin"); with idx, only the words
    # said just before the SAH ("large ... hematoma with ... subarachnoid
    # hemorrhage" does not make the SAH severe)
    window = sent_tokens[max(0, idx - 4):idx + 1] if idx is not None else sent_tokens
    words = set(" ".join(t.lower() for t, _ in window).split())

    if "scattered" in words:
        return "Scattered"

    if any(w in words for w in ("diffuse", "extensive", "large", "severe", "significant")):
        return "Severe"

    if "moderate" in words:
        return "Moderate"

    if any(w in words for w in ("minor", "small", "trace", "mild", "minimal", "tiny", "thin")):
        return "Mild"

    return None


def hemorrhage_variables(btype, grade=None):

    if btype == "subarachnoid" and grade:
        return [f"subarachnoid_hemorrhage_{grade}"]


    names = {
        "subarachnoid": "subarachnoid_hemorrhage",
        "intraparenchymal": "intraparenchymal_hemorrhage",
        "intraventricular": "intraventricular_hemorrhage",
        "epidural": "epidural_hemorrhage",
    }

    if btype in names:
        return [names[btype]]

    return []


def post_process_variables(final_variables):
    """
    Values that depend on the whole report.
    """

    # -------------------------
    # Overall SDH laterality
    # -------------------------
    left = final_variables.pop("_sdh_left", False)
    right = final_variables.pop("_sdh_right", False)

    if left and right:
        final_variables["laterality"] = "Bilateral"
    elif left:
        final_variables["laterality"] = "Left"
    elif right:
        final_variables["laterality"] = "Right"

    # -------------------------
    # Two different densities in one SDH ("hyperdense and hypodense
    # components") = mixed density too
    # -------------------------
    for side in ("L", "R"):
        named = [d for d in ("hyper_dense", "iso_dense", "hypo_dense")
                 if final_variables.get(f"{side}_{d}") == "Yes"]
        if len(named) >= 2:
            final_variables[f"{side}_mixed_density"] = "Yes"

    return final_variables
```

### Association_mapping

```python
def SDH_association_characteristics(
    association_children,
    sdh_side=None
):

    mapped_variables = []
    direct_variables = {}

    for group in association_children:

        for association in group:

            # -------------------------
            # Effects
            # -------------------------

            if "effect_name" in association:

                (
                    function_mapped_variables,
                    function_direct_variables
                ) = effect_name_sorter(
                    association,
                    sdh_side
                )
                
            # -------------------------
            # Negation
            # -------------------------

            elif "negation" in association:

                for negation in association["negation"]:
            
                    (
                        function_mapped_variables,
                        function_direct_variables
                    ) = negation_sorter(
                        negation,
                        sdh_side
                    )
            
                    mapped_variables.extend(
                        function_mapped_variables
                    )
            
                    direct_variables.update(
                        function_direct_variables
                    )
            
                continue

            elif "prominence" in association:

                (
                    function_mapped_variables,
                    function_direct_variables
                ) = prominence_sorter(
                    association
                )

            else:

                continue

            mapped_variables.extend(
                function_mapped_variables
            )

            direct_variables.update(
                function_direct_variables
            )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Effect name Sorter

```python
LATERALITY_MAP = {

    "left": "Left",
    "left-sided": "Left",
    "left sided": "Left",
    "to the left": "Left",
    "leftward": "Left",
    "towards the left": "Left",
    "on the left": "Left",
    "right to left": "Left",

    "right": "Right",
    "right-sided": "Right",
    "right sided": "Right",
    "to the right": "Right",
    "rightward": "Right",
    "towards the right": "Right",
    "on the right": "Right",
    "left to right": "Right",

    "bilateral": "Bilateral",
    "bilaterally": "Bilateral",
    "both cerebral hemispheres": "Bilateral",
    "both hemispheres": "Bilateral",
    "both sides": "Bilateral",
    "both cerebral convexities": "Bilateral",
    "both convexities": "Bilateral",

}

SDH_SIDE_NAMES = {"L": "Left", "R": "Right", "B": "Bilateral"}

HERNIATION_MAP = {

    "subfalcine": "herniation_subfalcine",
    "subfalcine herniation": "herniation_subfalcine",
    "uncal": "herniation_uncal",
    "uncal herniation": "herniation_uncal",
    "early uncal herniation": "herniation_uncal",
    "early uncal": "herniation_uncal",
    "uncal shift": "herniation_uncal",
    "medialized uncus": "herniation_uncal",
    "transtentorial herniation": "herniation_transtentorial",
    "transtentorial": "herniation_transtentorial",
    "descending transtentorial herniation": "herniation_transtentorial",
    "tonsillar": "herniation_tonsillar",
    "tonsillar herniation": "herniation_tonsillar",
    "transforaminal": "herniation_tonsillar",
    "transforaminal herniation": "herniation_tonsillar",
    "cerebellar tonsillar herniation": "herniation_tonsillar",
    "downward tonsillar herniation": "herniation_tonsillar",

}

# "no brain herniation" negates every subtype
GENERIC_HERNIATION = {
    "herniation", "brain herniation", "intracranial herniation",
}

# effects that squeeze a ventricle / cistern (not "shift" or "deviation")
SQUEEZE_EFFECTS = {
    "narrowing", "decreasing the size of", "compressive effects", "compressive effect",
    "compressed", "local mass effect", "localized mass effect",
    "effaced", "partially effaced",
    "effacement", "compression", "obliteration", "obliterated", "crowding",
    "medial positioning", "distortion", "distorsion",
    "mass effect", "indentation",
}

SULCAL_EFFECT_WORDS = {"sulcal blunting", "cortical buckling"}

SPECIFIC_CISTERNS = {
    "suprasellar": "cistern_suprasellar",
    "suprasellar cistern": "cistern_suprasellar",
    "ambient": "cistern_ambient",
    "ambient cistern": "cistern_ambient",
    "ambient cisterns": "cistern_ambient",
    "interpeduncular": "cistern_interpeduncular",
    "quadrigeminal": "cistern_quadrigeminal",
    "prepontine": "cistern_prepontine",
}


def is_midline_shift(effect_name, location_texts):

    if "midline" in effect_name and effect_name != "off the midline":
        return True

    # "5 mm shift of the septum pellucidum to the right"
    if "septum pellucidum" in effect_name:
        return True

    # "deviation of the midline structures", "shift ... midline structures"
    if effect_name in ("shift", "deviation", "shifted", "displacement"):
        return any("midline" in text for text in location_texts)

    return False


def effect_name_sorter(
    association,
    sdh_side=None
):

    mapped_variables = []
    direct_variables = {}

    effect_name = association[
        "effect_name"
    ][0].lower()

    # -------------------------
    # Extract magnitude
    # -------------------------

    severity = "Yes NS"

    if "magnitude" in association:

        severity = magnitude_function(
            association["magnitude"]
        )

    # "early uncal herniation", "mild midline shift" as one phrase
    words_in_name = [(w, "magnitude") for w in effect_name.split()[:-1] if w in MAGNITUDE_MAP]
    if words_in_name:
        named = magnitude_function(words_in_name)
        if SEVERITY_RANK.get(named, 0) > SEVERITY_RANK.get(severity, 0):
            severity = named

    # "obliteration" of a ventricle is severe effacement
    if effect_name in ("obliteration", "obliterated") and severity == "Yes NS":
        severity = "Severe"

    # -------------------------
    # Extract location info
    # -------------------------

    location_buckets = set()
    location_texts = []

    location_laterality = None

    # every ventricle named gets its own variable
    # ("right lateral ventricle as well as the third ventricle")
    ventricle_types = set()

    cistern_variables = []

    for text, bucket in association.get("location", []):

        text = text.lower()

        location_buckets.add(
            bucket
        )

        location_texts.append(text)

        if bucket == "laterality":

            if text == "ipsilateral" and sdh_side:
                location_laterality = SDH_SIDE_NAMES.get(sdh_side)

            elif text == "contralateral" and sdh_side in ("L", "R"):
                location_laterality = "Right" if sdh_side == "L" else "Left"

            elif location_laterality is None:
                location_laterality = LATERALITY_MAP.get(text)

        elif bucket == "ventricle":

            if text in ("lateral ventricle", "anterior horn", "posterior horn",
                        "occipital horn", "temporal horn", "body",
                        "lateral and third ventricles", "ventricular trigone", "trigone"):

                ventricle_types.add("lateral")

                if text == "lateral and third ventricles":
                    ventricle_types.add("third")

            elif text in ("lateral ventricles", "lateral ventricles bilaterally"):

                ventricle_types.add("lateral")

                # -------------------------
                # Bilateral override
                # -------------------------

                location_laterality = (
                    "Bilateral"
                )

            elif text == "third ventricle":

                ventricle_types.add("third")

            elif text == "fourth ventricle":

                ventricle_types.add("fourth")

            else:

                ventricle_types.add("system")

        elif bucket == "cistern":

            cistern_variables.append(
                SPECIFIC_CISTERNS.get(text, "effacement_cisterns")
            )

    # "effacement ... in both cerebral hemispheres"
    if "both" in " ".join(location_texts) or association.get("both"):
        location_laterality = "Bilateral"

    # "adjacent sulcal effacement" -> the side of the SDH
    if association.get("adjacent") and location_laterality is None and sdh_side:
        location_laterality = SDH_SIDE_NAMES.get(sdh_side)

    midline = is_midline_shift(effect_name, location_texts)

    # -------------------------
    # Mass Effect
    # -------------------------

    if effect_name in ("mass effect", "mass effects", "local mass effect",
                       "localized mass effect", "mass-effect", "pressure effect", "pressure effects",
                       "compressive effects", "compressive effect"):

        mapped_variables.append(
            f"mass_effect_present_{severity}"
        )

    # -------------------------
    # Uncal Herniation (medialized uncus)
    # -------------------------

    if "uncus" in location_buckets and not midline:

        mapped_variables.append(
            f"herniation_uncal_{severity}"
        )

    # -------------------------
    # Midline Shift
    # -------------------------

    if midline:

        mapped_variables.append(
            f"midline_shift_present_{severity}"
        )

        direction = None

        if "numeric_mass_effect" in association:

            shift_data = association[
                "numeric_mass_effect"
            ]

            value_text = (shift_data.get("value") or "").lower()

            mm, cm = measurement_to_mm_cm(value_text)

            if mm is not None:

                direct_variables[
                    "midline_shift_mm"
                ] = mm

                direct_variables[
                    "midline_shift_cm"
                ] = cm

            if shift_data.get("direction"):

                direction = LATERALITY_MAP.get(
                    shift_data["direction"].lower()
                )

        if direction is None:
            direction = location_laterality

        if direction in ("Left", "Right"):

            direct_variables[
                "midline_shift_direction"
            ] = direction

    # -------------------------
    # Herniation
    # ------------------------

    elif effect_name in HERNIATION_MAP:

        mapped_variables.append(
            f"{HERNIATION_MAP[effect_name]}_{severity}"
        )

    elif effect_name in GENERIC_HERNIATION:

        mapped_variables.append(
            f"herniation_generic_{severity}"
        )

    # -------------------------
    # Sulcal Effacement
    # -------------------------

    if (
        ("sulci" in location_buckets and not midline)
        or effect_name in SULCAL_EFFECT_WORDS
    ):

        # "diffuse sulcal effacement" is recorded as "Diffuse"
        sulci_value = severity
        if severity == "Yes NS" and any(
            t.lower() in ("diffuse", "global") for t, _ in association.get("magnitude", [])
        ):
            sulci_value = "Diffuse"

        mapped_variables.append(
            f"effacement_sulci_{sulci_value}"
        )

        # only a side that is written ("left sulcal", "ipsilateral",
        # "both hemispheres"); the human sheet does not assume the SDH side
        sulci_laterality = location_laterality

        if sulci_laterality:

            mapped_variables.append(
                f"effacement_sulci_laterality_{sulci_laterality}"
            )

    # -------------------------
    # Ventricular Effects (only effects that squeeze the ventricle)
    # -------------------------

    squeezes = effect_name in SQUEEZE_EFFECTS and not midline

    # "left ventricular effacement": one side's ventricle is the lateral one
    if (
        "system" in ventricle_types
        and location_laterality in ("Left", "Right")
        and any(t in ("ventricular", "ventricle") for t in location_texts)
    ):
        ventricle_types.discard("system")
        ventricle_types.add("lateral")

    # "the ventricles" only counts when no ventricle is named
    if ventricle_types - {"system"}:
        ventricle_types.discard("system")

    if squeezes and "lateral" in ventricle_types:

        mapped_variables.append(
            f"effacement_lateral_ventricle_{severity}"
        )

        if location_laterality:

            mapped_variables.append(
                f"l_ventricle_laterality_{location_laterality}"
            )

    if squeezes and "third" in ventricle_types:

        mapped_variables.append(
            f"effacement_third_ventricle_{severity}"
        )

    if squeezes and "fourth" in ventricle_types:

        mapped_variables.append(
            f"effacement_forth_ventricle_{severity}"
        )

    if squeezes and "system" in ventricle_types:

        mapped_variables.append(
            f"effacement_ventricular_system_{severity}"
        )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Negation

```python
def negation_sorter(
    negation,
    sdh_side=None
):

    # -------------------------
    # Negated pathology
    # -------------------------

    if "pathology" in negation:

        mapped_variables, direct_variables = (
            pathology_function(
                negation["pathology"],
                negation.get(
                    "modifiers",
                    []
                ),
                negation.get(
                    "anatomy",
                    []
                ),
                negated=True
            )
        )

    # -------------------------
    # Negated effects
    # -------------------------

    elif "effect_name" in negation:

        mapped_variables, direct_variables = (
            effect_name_sorter(
                negation,
                sdh_side
            )
        )

        # a negated shift has no size / direction
        direct_variables = {}

    # -------------------------
    # Negated bleeds ("no acute parenchymal or subarachnoid hemorrhage")
    # -------------------------

    elif "bleed" in negation:

        text, btype = negation["bleed"]

        mapped_variables = hemorrhage_variables(btype)
        direct_variables = {}

        anatomy_buckets = {b for _, b in negation.get("anatomy", [])}

        # "no hemorrhage in the basilar cisterns" -> no subarachnoid hemorrhage
        if btype == "generic" and "cistern" in anatomy_buckets:
            mapped_variables.append("subarachnoid_hemorrhage")

        # "No acute intracranial hemorrhage" -> no intraparenchymal,
        # subarachnoid or intraventricular hemorrhage
        elif btype == "generic":
            mapped_variables.extend([
                "intraparenchymal_hemorrhage",
                "subarachnoid_hemorrhage",
                "intraventricular_hemorrhage",
            ])

        # "parenchymal" saved in front of the head negates IPH as well
        for anatomy_text, anatomy_bucket in negation.get("anatomy", []):
            if anatomy_bucket == "intraparenchymal" and "parenchym" in anatomy_text.lower():
                mapped_variables.append("intraparenchymal_hemorrhage")
            if anatomy_bucket == "ventricle" and "intraventricular" in anatomy_text.lower():
                mapped_variables.append("intraventricular_hemorrhage")

    # -------------------------
    # Negated prominence: only a ventricle can be "not dilated" here
    # -------------------------

    elif "prominence" in negation:

        anatomy_buckets = {b for _, b in negation.get("anatomy", [])}
        mapped_variables = ["ventricular_prominence"] if "ventricle" in anatomy_buckets else []
        direct_variables = {}

    # -------------------------
    # Negated outcomes
    # -------------------------

    elif "outcome" in negation:

        mapped_variables, direct_variables = (
            outcome_function(
                negation["outcome"]
            )
        )

    else:

        return (
            [],
            {}
        )

    # -------------------------
    # Negate mapped variables
    # -------------------------
    mapped_variables = [
        variable if variable.startswith("N_") else f"N_{variable}"
        for variable in mapped_variables

    ]

    return (
        mapped_variables,
        direct_variables
    )
```

#### Pathology

```python
INFARCT_VARIABLES = {
    "territorial_ischemic",
    "unspecified_infarct",
    "territorial_infarct",
    "lacunar_infarct",
    "cortical_infarct",
}

SEVERITY_VARIABLES = {
    "cerebral_atrophy",
    "microangiopathy",
    "hydrocephalus",
}

ACUTE_WORDS = {"acute", "new", "subacute", "recent"}
CHRONIC_WORDS = {"chronic", "old", "remote", "prior", "long-standing", "longstanding", "long standing"}


def infarct_qualifier(modifier_tokens):
    """
    "No acute infarct" -> Acute, "Old infarct" -> Chronic, otherwise NS.
    ("recent" / "interval" stay NS, as the human sheet does)
    """

    for text, bucket in modifier_tokens:

        if bucket != "acuity":
            continue

        text = text.lower()

        if text in ACUTE_WORDS:
            return "Acute"

        if text in CHRONIC_WORDS:
            return "Chronic"

    return "NS"


def pathology_function(
    pathology_list,
    modifier_tokens=None,
    anatomy_tokens=None,
    negated=False
):

    modifier_tokens = modifier_tokens or []

    pathology_text = []

    for text, bucket in pathology_list:

        pathology_text.append(
            text
        )

    mapped_variables = pathology_filter(
        pathology_text
    )

    # -------------------------
    # Anatomy-based mappings
    # -------------------------

    if anatomy_tokens:

        anatomy_text = []

        for text, bucket in anatomy_tokens:

            anatomy_text.append(
                text
            )

        anatomy_variables = (
            pathology_anatomy_filter(
                anatomy_text
            )
        )

        mapped_variables.extend(
            anatomy_variables
        )

    # -------------------------
    # Generic abnormality
    # -------------------------

    if "generic_abnormality" in mapped_variables:

        mapped_variables.remove(
            "generic_abnormality"
        )

    # -------------------------
    # Infarct type from the saved modifiers
    # ("large artery territory ischemic infarct", "cortical infarction")
    # -------------------------

    modifier_buckets = {bucket for _, bucket in modifier_tokens}

    updated_variables = []

    qualifier_tokens = modifier_tokens + [
        (word, "acuity")
        for text, _ in pathology_list
        for word in text.lower().split()
    ]

    pathology_words = " ".join(pathology_text).lower()

    for variable in mapped_variables:

        if variable in ("unspecified_infarct", "territorial_infarct") and "cortical" in pathology_words \
                and "territorial" not in modifier_buckets:
            variable = "cortical_infarct"

        if variable == "unspecified_infarct":

            # "large artery territory cortical infarct" is territorial
            if "territorial" in modifier_buckets:
                variable = "territorial_infarct"

            elif "cortical" in modifier_buckets:
                variable = "cortical_infarct"

        if variable in INFARCT_VARIABLES:
            variable = f"{variable}_{infarct_qualifier(qualifier_tokens)}"

        updated_variables.append(variable)

    mapped_variables = updated_variables


    anatomy_words = " ".join(t.lower() for t, _ in (anatomy_tokens or []))

    # "scalp soft tissue density ... stranding" is not a scalp hematoma
    if "soft_tissue_lesion" in mapped_variables and "scalp_hematoma" in mapped_variables \
            and not any("hematoma" in t.lower() for t in pathology_text):
        mapped_variables = [v for v in mapped_variables if v != "scalp_hematoma"]
    if "edema" in mapped_variables and any(
        w in anatomy_words for w in ("soft tissue", "scalp", "subcutaneous", "orbit", "facial", "prefacial")
    ):
        mapped_variables = ["soft_tissue_lesion" if v == "edema" else v for v in mapped_variables]

    # -------------------------
    # Add severity for selected
    # pathology variables
    # -------------------------

    if not negated:

        # "There is general atrophy." -> Yes NS (not just Yes)
        severity = magnitude_function(
            modifier_tokens,
            allow_age=True
        )

        updated_variables = []

        for variable in mapped_variables:

            if variable in SEVERITY_VARIABLES:

                updated_variables.append(
                    f"{variable}_{severity}"
                )

            else:

                updated_variables.append(
                    variable
                )

        mapped_variables = (
            updated_variables
        )

    direct_variables = {}

    return (
        list(set(mapped_variables)),
        direct_variables
    )
```

#### Outcome 

```python
def outcome_function(
    outcome
):

    # -------------------------
    # Outcome pathology: one item per pathology head,
    # each with the modifiers that were said before it
    # -------------------------

    if "items" in outcome:

        mapped_variables = []

        for item in outcome["items"]:

            item_variables, _ = pathology_function(
                item["pathology"],
                item.get("modifiers", []),
                outcome.get("anatomy") or []
            )

            mapped_variables.extend(item_variables)

        return (
            list(set(mapped_variables)),
            {}
        )

    if "pathology" in outcome:

        result = pathology_function(
            outcome["pathology"],
            outcome.get(
                "modifiers",
                []
            ),
            outcome.get(
                "anatomy",
                []
            )
        )

        return result

    # -------------------------
    # Outcome normal state
    # -------------------------

    elif "normal_state" in outcome:

        result = normal_state_function(
            outcome
        )

        return result

    return (
        [],
        {}
    )
```

#### Normal State

```python
def normal_state_function(
    outcome
):
    anatomy_text = []

    if outcome.get("anatomy"):

        for text, bucket in outcome["anatomy"]:

            anatomy_text.append(
                text
            )

    mapped_variables = (
        normal_state_anatomy_filter(
            anatomy_text
        )
    )

    mapped_variables = [

        f"N_{variable}"

        for variable in mapped_variables

    ]

    direct_variables = {}

    return (
        mapped_variables,
        direct_variables
    )
```

#### Prominence 

```python
def prominence_sorter(
    association
):

    mapped_variables = []
    direct_variables = {}

    prominence_text = association[
        "prominence"
    ][0].lower()

    location_buckets = set()
    laterality = None

    if "location" in association:

        for text, bucket in association["location"]:

            location_buckets.add(
                bucket
            )

            if bucket == "laterality" and laterality is None:
                laterality = LATERALITY_MAP.get(text.lower())

            if bucket == "ventricle" and text.lower() in ("lateral ventricles", "temporal horns"):
                laterality = laterality or "Bilateral"

    severity = magnitude_function(association.get("magnitude", []))

    # ex vacuo dilatation is only ex vacuo (not ventricular prominence)
    if prominence_text == "ex vacuo dilatation":

        mapped_variables.append(
            f"ex_vacuo_dilatation_{severity}"
        )

        return (
            mapped_variables,
            direct_variables
        )

    if "ventricle" in location_buckets:

        mapped_variables.append(
            f"ventricular_prominence_{severity}"
        )

        if laterality:
            mapped_variables.append(
                f"l_v_prominence_laterality_{laterality}"
            )

    return (
        mapped_variables,
        direct_variables
    )
```

#### Magnitude Mapper

```python
MAGNITUDE_MAP = {

    "mild": "Mild",
    "early": "Mild",          # "early left-sided uncal herniation"
    "partial": "Mild",
    "partially": "Mild",
    "mildly": "Mild",
    "minimal": "Mild",
    "some": "Mild",
    "slight": "Mild",
    "slightly": "Mild",
    "subtle": "Mild",
    "trace": "Mild",
    "minor": "Mild",
    "relatively mild": "Mild",
    "mild-early": "Mild",
    "moderate": "Moderate",
    "moderate sized": "Moderate",
    "moderate size": "Moderate",
    "severe": "Severe",
    "marked": "Severe",
    "significant": "Severe",
    "large": "Severe",
    "moderately severe": "Severe",
    "complete": "Severe",
    "near complete": "Severe",
    "nearly complete": "Severe",
    "extensive": "Severe",
    "substantial": "Severe",

    # ranges keep both ends, like the human sheet ("Moderate/ Severe")
    "moderate to severe": "Moderate/ Severe",
    "mild to moderate": "Mild/ Moderate",

    # "diffuse" / "generalized" say how widespread, not how severe
    "diffuse": "Yes NS",
    "generalized": "Yes NS",
    "global": "Yes NS",

}

AGE_WORDS = {"age appropriate", "age-appropriate", "age-related", "age related", "in keeping with age"}

SEVERITY_RANK = {
    "Yes NS": 0,
    "Mild": 1,
    "Mild/ Moderate": 2,
    "Moderate": 3,
    "Moderate/ Severe": 4,
    "Severe": 5,
}


def magnitude_function(
    modifier_tokens,
    allow_age=False
):
    """
    The STRONGEST grade said wins ("severe generalized" -> Severe,
    not the last word). Words that are not grades are ignored.
    "Age-related / age-appropriate" (atrophy) is only used when no
    real grade was said.
    """

    severity = "Yes NS"
    age_appropriate = False

    for text, bucket in modifier_tokens:

        if bucket != "magnitude":
            continue

        text = text.lower()

        if text in AGE_WORDS:
            age_appropriate = True
            continue

        grade = MAGNITUDE_MAP.get(text)

        if grade and SEVERITY_RANK[grade] > SEVERITY_RANK[severity]:
            severity = grade

    if allow_age and age_appropriate and severity == "Yes NS":
        return "Age Appropriate"

    return severity
```

### SDH Characteristics

```python
def side_of(text):
    """
    "left", "left-sided", "on the left" -> "L"   "bilateral" -> "B"
    "rightward", "to the right" etc are directions of shift, not a side,
    but inside an SDH sentence they still name the side.
    """

    text = text.lower()

    if "bilateral" in text or text.startswith("both "):
        return "B"

    if text in ("left to right", "right to left"):
        return None

    if "left" in text:
        return "L"

    if "right" in text:
        return "R"

    return None


def new_side_record():
    return {
        "cerebral_locations": [],
        "density": [],
        "acuity": [],
        "magnitude": [],
        "morphology": [],
        "thickness": [],
        "density_negated": [],
        "density_dominant": [],
        "named": [],            # "Left subdural hematoma." with nothing else said
    }


# "predominantly hypo to isodense", "largely isodense", "background of
# low-density": the density words right after these are the dominant ones
DOMINANCE_WORDS = {
    "predominantly", "predominately", "predominant", "largely", "mostly",
    "mainly", "primarily", "chiefly", "background of", "majority of",
}

DOMINANCE_VALUES = {
    frozenset(["hypo_dense"]): "Hypodense",
    frozenset(["iso_dense"]): "Isodense",
    frozenset(["hyper_dense"]): "Hyperdense",
    frozenset(["hypo_dense", "iso_dense"]): "Hypo/Isodense",
    frozenset(["iso_dense", "hyper_dense"]): "Iso/Hyperdense",
}


ATTRIBUTE_BUCKETS = {
    "cerebral_location": "cerebral_locations",
    "density": "density",
    "acuity": "acuity",
    "magnitude": "magnitude",
    "morphology": "morphology",
}

ACUITY_PRIORITY = [
    "acuity_acute_on_chronic",
    "acuity_subacute_on_chronic",
    "acuity_acute_subacute",
    "acuity_acute",
    "acuity_subacute",
    "acuity_chronic",
]

MORPHOLOGY_PRIORITY = ["septations", "dependent", "layering"]


def SDH_characteristics(children):
    """
    Walk the SDH children IN ORDER and give every characteristic to
    the side it was said about.

      - a laterality switches the side ("right ... while the left ...")
      - words said before the first laterality of a sentence are saved
        and given to that laterality ("large mixed density LEFT ...",
        "isodense SDH along RIGHT convexity")
      - with no laterality in the sentence they go to the side of the
        previous sentence
      - "bilateral" gives them to both sides
      - words about another bleed (IPH, SAH) are dropped
      - falx / tentorium locations open a compartment
        (parafalcine_hematoma / tentorium_hematoma)
    """

    sides = {"L": new_side_record(), "R": new_side_record()}

    compartments = {}           # "parafalcine" -> {"sides": set(), ...}

    mapped_variables = []
    direct_variables = {}

    tree_side = None            # side of the previous sentence
    side_before_sentence = None
    previous_sentence_compartment = None
    continuing_compartment = None
    sent_side = None            # side named in this sentence
    sent_compartment = None
    compartment_first = False   # falx / tentorium named before any lobe this sentence
    pending = []                # (record_key, value) waiting for a side
    compartment_acuity = []     # (side, acuity) said about the falx / tentorium
    falx_where = []             # "anterior" / "posterior" falx
    unsided = []                # never got a side in their sentence
    blocked = False             # inside a clause about another bleed
    blocked_buffer = []         # "IVH in the right horn as well as <thin acute> SDH"
    last_side_change = [None, None, False, None]   # state before the last side word
    last_laterality = None      # for "left falx", "right tentorium"
    tokens_since_laterality = 99

    sent_shared = []            # said before the first side of the sentence
    sent_sides_named = set()
    sent_bleed_mentions = 0

    sent_log = []               # (target, key, value) given in this sentence
    since_mention = []          # (target, key, value) given after the last bleed word / side

    def give(side, key, value):
        targets = ["L", "R"] if side == "B" else [side]
        for target in targets:
            sides[target][key].append(value)
            sent_log.append((target, key, value))
            if key != "thickness":
                since_mention.append((target, key, value))

    def take_back():
        """
        Words given after the last bleed word belong to the NEXT one:
        "... the left tentorium and there is thin subdural blood along
        the right frontal convexity", "SDH and additional acute left SDH"
        """
        moved = []
        for target, key, value in since_mention:
            if value in sides[target][key]:
                sides[target][key].remove(value)
                moved.append((key, value))
        since_mention.clear()
        return moved

    def flush(side):
        nonlocal pending
        if side:
            for key, value in pending:
                give(side, key, value)
        else:
            unsided.extend(pending)
        pending = []

    def end_sentence():
        nonlocal sent_shared, tree_side, compartment_first
        """
        One SDH named on both sides in one breath
        ("mixed attenuation SDH along right frontoparietal and left
        frontoparietal temporal convexities") -> both sides get the words.
        """
        flush(sent_side or tree_side)

        # "acute SDH involving the superior aspect of the right tentorium":
        # a sentence only about the falx / tentorium describes that bleed,
        # not the convexity SDH of that side - its words go to the
        # compartment (acuity) or nowhere, and the side is not changed
        # "tiny amount of SDH in the anterior left parafalcine region just
        # medial to the left frontal lobe": the falx is named first, the lobe
        # is only a landmark
        sentence_places = [str(v).lower() for _, k, v in sent_log if k == "cerebral_locations"]
        landmarks_only = (
            compartment_first
            and sent_bleed_mentions <= 1
            and not any(re.search(r"convexit|vertex|hemispher", p) for p in sentence_places)
        )
        compartment_sentence = bool(sent_compartment and (landmarks_only or not sentence_places))
        if compartment_sentence:
            for target, key, value in sent_log:
                if value in sides[target][key]:
                    sides[target][key].remove(value)
                if key in ("acuity", "density", "magnitude"):
                    if value not in compartments[sent_compartment][key]:
                        compartments[sent_compartment][key].append(value)
            tree_side = side_before_sentence

        # "acute SDH over the left frontal lobe extending into the falx": the
        # falx part is the same bleed, so it has the same age
        elif sent_compartment:
            for target, key, value in sent_log:
                if key == "acuity" and value not in compartments[sent_compartment]["acuity"]:
                    compartments[sent_compartment]["acuity"].append(value)
        compartment_first = False

        if {"L", "R"} <= sent_sides_named and sent_bleed_mentions <= 1 and not compartment_sentence:
            for key, value in sent_shared:
                if key == "thickness":
                    continue
                for target in ("L", "R"):
                    if value not in sides[target][key]:
                        sides[target][key].append(value)

    def note_compartment(text):
        nonlocal tokens_since_laterality
        if text.lower() in ("falco tentorial", "falcotentorial"):
            note_compartment("tentorium")
            text = "falx"
        name = COMPARTMENT_WORDS.get(text.lower())
        if not name:
            return None
        record = compartments.setdefault(
            name, {"sides": set(), "subregion": None, "thickness": None, "acuity": [], "density": [], "magnitude": []}
        )
        # only a laterality said right in front: "left falx", "right side of
        # the falx"; "right tentorium and falx" carries on to the next one
        if last_laterality and tokens_since_laterality <= 1:
            record["sides"].add(last_laterality)
            tokens_since_laterality = 0
        return name

    def take_token(text, bucket, from_measurement=False, in_list=False):
        nonlocal sent_side, tree_side, last_laterality, tokens_since_laterality, sent_compartment

        # inside a clause about another bleed: its side and place are its own;
        # grade / age / density words wait in case the SDH is named next
        if blocked:
            if bucket in ("laterality", "cerebral_location", "bone", "ventricle"):
                blocked_buffer.clear()
            elif ATTRIBUTE_BUCKETS.get(bucket) in ("magnitude", "acuity", "density", "morphology"):
                blocked_buffer.append((text, bucket))
            return

        tokens_since_laterality += 1

        if bucket == "laterality":
            side = side_of(text)
            # remembered so "bilateral subarachnoid hemorrhage" can give it back
            last_side_change[:] = [sent_side, tree_side, side not in sent_sides_named, side]
            if (
                side and sent_bleed_mentions and sent_side and side != sent_side
                and "B" not in (side, sent_side) and not from_measurement and not in_list
            ):
                # "SDH and additional acute LEFT frontoparietal SDH"
                pending.extend(take_back())
            if side:
                since_mention.clear()
                last_laterality = side
                tokens_since_laterality = 0
                if not sent_sides_named:
                    sent_shared.extend(pending)
                sent_side = side
                tree_side = side
                sent_sides_named.add(side)
                flush(side)
            return

        if bucket in ("cerebral_location", "bone"):
            compartment = note_compartment(text)
            if compartment:
                nonlocal compartment_first
                if not any(k == "cerebral_locations" for _, k, _ in sent_log):
                    compartment_first = True
                sent_compartment = compartment
                # "thin falcotentorial subdural": the size word right in front
                # of the falx / tentorium is that bleed's size
                if sent_log and sent_log[-1][1] == "magnitude":
                    target, key, value = sent_log.pop()
                    if value in sides[target][key]:
                        sides[target][key].remove(value)
                    compartments[compartment]["magnitude"].append(value)
                elif pending and pending[-1][0] == "magnitude":
                    compartments[compartment]["magnitude"].append(pending.pop()[1])
                return
            # "along the falx and the right frontal convexity, measuring
            # 6 mm": with a convexity named too, what follows (the
            # thickness) is the convexity's
            if "convex" in text.lower():
                sent_compartment = None

        key = ATTRIBUTE_BUCKETS.get(bucket)

        if bucket == "bone":
            key = "cerebral_locations"

        if key is None or blocked:
            return

        if sent_compartment and key == "acuity":
            compartments[sent_compartment]["acuity"].append(text)
            compartment_acuity.append((sent_side or tree_side, text))
            return

        # "predominantly" itself is not a density
        if key == "density" and text.lower() in DOMINANCE_WORDS:
            return

        # tagged by mark_dominance: "predominantly <hypo to isodense>"
        keys = [key]
        if key == "density" and text.lower().startswith("dominant:"):
            keys.append("density_dominant")

        for key in keys:
            if sent_side:
                give(sent_side, key, text)
            else:
                pending.append((key, text))

    for child in children:

        # -------------------------
        # new primary sentence
        # -------------------------
        if isinstance(child, dict) and "sentence_start" in child:
            end_sentence()
            previous_sentence_compartment = (
                sent_compartment
                if sent_compartment and not any(k == "cerebral_locations" for _, k, _ in sent_log)
                else None
            )
            side_before_sentence = tree_side
            # "This hematoma extends to ... the falx" after a tentorial bleed:
            # still the tentorial bleed
            continuing_compartment = previous_sentence_compartment if child.get("anaphor") else None
            sent_side = None
            sent_compartment = None
            blocked = False
            last_laterality = None
            sent_shared = []
            sent_log.clear()
            since_mention.clear()
            sent_sides_named = set()
            sent_bleed_mentions = 0
            continue

        if isinstance(child, dict) and "falx_subregion" in child:
            falx_where.append(child["falx_subregion"])
            continue

        # -------------------------
        # "extension along the right side of the falx" (secondary sentence)
        # -------------------------
        if isinstance(child, dict) and "compartment_location" in child:

            saved = (sent_side, tree_side, last_laterality, tokens_since_laterality)
            # only a side said inside this phrase counts ("extension along the
            # anterior falx" after a right-sided sentence is not a right falx bleed)
            last_laterality, tokens_since_laterality = None, 99
            for text, bucket in child["compartment_location"]:
                if bucket == "laterality" and side_of(text):
                    last_laterality = side_of(text)
                    tokens_since_laterality = 0
                    continue
                tokens_since_laterality += 1
                note_compartment(text)
            sent_side, tree_side, last_laterality, tokens_since_laterality = saved
            continue

        # -------------------------
        # which bleed is talked about
        # -------------------------
        if isinstance(child, dict) and "bleed_mention" in child:
            # "... consistent with chronic subdural hematomas": same bleed
            if child.get("restates") and not blocked:
                continue
            if child["bleed_mention"] in ("subdural", "generic"):
                replay = list(blocked_buffer) if blocked else []
                blocked_buffer.clear()
                blocked = False
                # the side said right in front of the SDH word has an SDH,
                # even when nothing else is said about it
                if (
                    child["bleed_mention"] == "subdural"
                    and last_laterality and tokens_since_laterality <= 2
                    and not sent_compartment
                ):
                    for target in (["L", "R"] if last_laterality == "B" else [last_laterality]):
                        if True not in sides[target]["named"]:
                            sides[target]["named"].append(True)
                if sent_bleed_mentions and since_mention:
                    # a second bleed in the sentence: what was said since the
                    # last one waits for this bleed's side
                    pending.extend(take_back())
                    sent_side = None
                since_mention.clear()
                sent_bleed_mentions += 1
                # "Subdural hematoma on the left.": the side said as the very
                # next word has an SDH, even when nothing else is said
                side_after = side_of(child.get("side_after") or "")
                if child["bleed_mention"] == "subdural" and side_after:
                    for target in (["L", "R"] if side_after == "B" else [side_after]):
                        if True not in sides[target]["named"]:
                            sides[target]["named"].append(True)
                # "Parafalcine hematoma measuring 0.5 cm"
                if "parafalcine" in child.get("text", "").lower():
                    sent_compartment = note_compartment("parafalcine")
                # "... as well as thin acute SDH": words held back while blocked
                for text, bucket in replay:
                    take_token(text, bucket)
            else:
                # "with bilateral subarachnoid hemorrhage": the side said right in
                # front of the other bleed is that bleed's side
                restored = bool(last_laterality and tokens_since_laterality <= 1 and last_side_change[3])
                if restored:
                    sent_side, tree_side = last_side_change[0], last_side_change[1]
                    if last_side_change[2]:
                        sent_sides_named.discard(last_side_change[3])
                    last_laterality = None

                # "SDH with small subarachnoid hemorrhage": the words right
                # in front of the other bleed are about that bleed
                taken_back = ("magnitude", "acuity", "density") + (("cerebral_locations",) if restored else ())
                for target, key, value in reversed(list(since_mention)):
                    if key not in taken_back:
                        break
                    if value in sides[target][key]:
                        sides[target][key].reverse()
                        sides[target][key].remove(value)
                        sides[target][key].reverse()
                since_mention.clear()
                blocked = True
                blocked_buffer.clear()
                pending = []
            continue

        # -------------------------
        # SDH Negation branch
        # -------------------------
        if isinstance(child, dict) and "negation" in child:

            for negation in child["negation"]:

                if "density" in negation:
                    side = sent_side or tree_side
                    if side:
                        for text, bucket in negation["density"]:
                            give(side, "density_negated", text)
                    continue

                negation_variables, negation_direct = (
                    negation_sorter(
                        negation
                    )
                )

                mapped_variables.extend(
                    negation_variables
                )

                direct_variables.update(
                    negation_direct
                )

            continue

        # -------------------------
        # Direct tuples
        # -------------------------
        if isinstance(child, tuple):
            take_token(*child)
            continue

        # -------------------------
        # Location branch
        # -------------------------
        if isinstance(child, dict) and "location" in child:
            after_compartment = False
            branch = child["location"]
            # "overlying the right and left frontal lobes": two sides said
            # together, then the places: every place is on both sides
            leading = []
            for t, b in branch:
                if b != "laterality":
                    break
                leading.append(t)
            leading_sides = {side_of(t) for t in leading} - {None}
            if len(leading_sides) > 1 and len(leading) < len(branch):
                for t, b in branch[len(leading):]:
                    if b == "cerebral_location" and t.lower() not in COMPARTMENT_WORDS:
                        for list_side in sorted(leading_sides):
                            give(list_side, "cerebral_locations", t)
                continue
            for j, (text, bucket) in enumerate(branch):
                # "high left frontal parafalcine region": the lobe says where on
                # the falx, so it is the falx bleed's place, not a convexity lobe
                following = next((t for t, b in branch[j + 1:] if b != "laterality"), "")
                if (
                    bucket == "cerebral_location"
                    and text.lower() not in COMPARTMENT_WORDS
                    and following.lower() in COMPARTMENT_WORDS
                    and re.search(r"frontal|parietal|occipital|temporal", text.lower())
                    and not re.search(r"lobe|region|convexit", text.lower())
                ):
                    if COMPARTMENT_WORDS[following.lower()] == "parafalcine":
                        falx_where.append("Anterior" if "frontal" in text.lower() else "Posterior")
                    continue
                # "right frontoparietal and left frontoparietal temporal
                # convexities": a plural place at the end of a list with two
                # sides belongs to both
                list_sides = {side_of(t) for t, b in branch if b == "laterality" and side_of(t)}
                if (
                    j == len(branch) - 1 and len(list_sides) > 1
                    and bucket == "cerebral_location" and text.lower().endswith(("ies", "es"))
                ):
                    for list_side in list_sides:
                        give(list_side, "cerebral_locations", text)
                    continue
                # "in the falx in the parietal region": the lobe says where
                # along the falx, it is not a convexity location
                is_compartment = text.lower() in COMPARTMENT_WORDS
                if (
                    after_compartment and bucket == "cerebral_location"
                    and not is_compartment
                    and any(lobe in text.lower() for lobe in ("frontal", "parietal", "temporal", "occipital"))
                ):
                    continue
                # inside one list each place belongs to the side said before it
                take_token(text, bucket, in_list=True)
                # "falx and right vertex": a new side starts a new location
                after_compartment = is_compartment and bucket == "cerebral_location"
            continue

        # -------------------------
        # Thickness branch
        # -------------------------
        if isinstance(child, dict) and "thickness" in child:

            thickness = child["thickness"]

            # "maximum thickness left frontal 2.3 cm": the side and place
            # said inside the measurement are SDH words too
            if thickness.get("side") and side_of(thickness["side"]):
                take_token(thickness["side"], "laterality", from_measurement=True)

            side = (
                side_of(thickness["side"]) if thickness.get("side") else None
            ) or sent_side or tree_side

            already_said = len(since_mention)

            branch = thickness.get("locations", [])
            for j, (text, bucket) in enumerate(branch):
                # "... in the high left frontal parafalcine region": the lobe is
                # where on the falx
                following = next((t for t, b in branch[j + 1:] if b != "laterality"), "")
                if (
                    bucket == "cerebral_location"
                    and text.lower() not in COMPARTMENT_WORDS
                    and following.lower() in COMPARTMENT_WORDS
                    and re.search(r"frontal|parietal|occipital|temporal", text.lower())
                    and not re.search(r"lobe|region|convexit", text.lower())
                ):
                    if COMPARTMENT_WORDS[following.lower()] == "parafalcine":
                        falx_where.append("Anterior" if "frontal" in text.lower() else "Posterior")
                    continue
                take_token(text, bucket)

            # a measurement with its own side keeps its places: a later
            # "... and the left subdural hematoma" does not take them back
            if thickness.get("side") and side_of(thickness["side"]):
                del since_mention[already_said:]

            compartment = thickness.get("compartment") or sent_compartment

            if compartment and compartment in compartments:
                if compartments[compartment]["thickness"] is None:
                    compartments[compartment]["thickness"] = thickness["value"]

            # falx / tentorium bleeds are their own variables: their size
            # never goes to L_ / R_ thickness
            if compartment:
                continue

            if side:
                give(side, "thickness", dict(thickness, compartment=compartment))
            else:
                unsided.append(("thickness", dict(thickness, compartment=compartment)))

            continue

    end_sentence()

    # words that never had a side in their sentence go to the side the
    # SDH is named on elsewhere ("there is subdural hematoma predominantly
    # isodense ... over the right frontoparietal convexity")
    named = {k for k, record in sides.items() if any(record.values())}
    if len(named) == 1 and unsided:
        target = next(iter(named))
        for key, value in unsided:
            sides[target][key].append(value)

    # -------------------------
    # Per side: run filters, prefix L_ / R_
    # -------------------------
    for prefix, record in sides.items():

        if not any(record.values()):
            continue

        mapped_variables.append(f"sdh_side_{prefix}")

        sdh_variables = []

        location_variables = cerebral_location_filter(record["cerebral_locations"])

        # "holohemispheric ... from the frontal to the occipital lobe": the
        # whole hemisphere, so the single lobes are not recorded
        if "holohemispheric_hematoma" in location_variables:
            location_variables = [
                v for v in location_variables
                if v not in ("frontal_lobe", "parietal_lobe", "temporal_lobe", "occipital_lobe")
            ]

        sdh_variables.extend(location_variables)
        sdh_variables.extend(density_filter(record["density"]))

        # one acuity: compound ("acute on chronic") beats simple ("acute")
        acuity_words = {a.lower() for a in record["acuity"]}
        density_words = " ".join(d.lower() for d in record["density"])
        has_chronic_part = (
            "chronic" in acuity_words
            or any(w in density_words for w in ("hypo", "low"))
        )

        # "high density ... in keeping with acute hemorrhage" inside an
        # "intermediate to low density" collection, "RECENT COMPONENTS UPON
        # ... LOW-DENSITY CHRONIC COMPONENT" -> acute on chronic
        if has_chronic_part and acuity_words & {"acute", "recent", "new"}:
            record["acuity"] = record["acuity"] + ["acute on chronic"]
        elif "chronic" in acuity_words and "subacute" in acuity_words:
            record["acuity"] = record["acuity"] + ["subacute on chronic"]

        acuity_variables = acuity_filter(record["acuity"])
        for candidate in ACUITY_PRIORITY:
            if candidate in acuity_variables:
                sdh_variables.append(candidate)
                break

        # one description: the first size word said
        magnitude_variables = []
        for text in record["magnitude"]:
            magnitude_variables = magnitude_filter([text])
            if magnitude_variables:
                break
        sdh_variables.extend(magnitude_variables)

        # one morphology: septations > dependent > layering
        morphology_variables = morphology_filter(record["morphology"])
        for candidate in MORPHOLOGY_PRIORITY:
            if candidate in morphology_variables:
                sdh_variables.append(candidate)
                break

        for variable in sdh_variables:
            mapped_variables.append(f"{prefix}_{variable}")

        # density dominance: only when the report says which density dominates
        dominant = frozenset(density_filter(record["density_dominant"])) - {"mixed_density"}
        if dominant in DOMINANCE_VALUES:
            mapped_variables.append(f"{prefix}_density_dominance_{DOMINANCE_VALUES[dominant]}")

        for variable in density_filter(record["density_negated"]):
            mapped_variables.append(f"N_{prefix}_{variable}")

        # -------------------------
        # Thickness: the first measurement of a side wins
        # (later ones are "previously 1.2 cm", impression repeats ...)
        # -------------------------
        if record["thickness"]:

            # a measurement along the falx / tentorium only counts for the
            # side when nothing was measured over the convexity
            convexity = [t for t in record["thickness"] if not t.get("compartment")]
            thickness = (convexity or record["thickness"])[0]
            mm, cm = measurement_to_mm_cm(thickness["value"])

            if mm is not None:

                direct_variables[f"{prefix}_thickness_mm"] = mm
                direct_variables[f"{prefix}_thickness_cm"] = cm

                if thickness["max"]:
                    direct_variables[f"{prefix}_thickness_max"] = "Yes"

    # -------------------------
    # Compartments (falx / tentorium)
    # -------------------------
    side_names = {"L": "Left", "R": "Right", "B": "Bilateral"}

    for name, record in compartments.items():

        named = record["sides"]

        if "B" in named or ("L" in named and "R" in named):
            value = "Bilateral"
        elif named:
            value = side_names[next(iter(named))]
        else:
            value = "Yes NS"

        mapped_variables.append(f"{name}_hematoma_{value}")

        if name == "parafalcine" and falx_where:
            where = falx_where[0]
            if {"Anterior", "Posterior"} <= set(falx_where):
                where = "Anterior/posterior"
            mapped_variables.append(f"parafalcine_location_{where}")

        if record["thickness"]:
            mm, cm = measurement_to_mm_cm(record["thickness"])
            if mm is not None:
                direct_variables[f"{name}_thickness_mm"] = mm
                direct_variables[f"{name}_thickness_cm"] = cm

        acuity_variables = acuity_filter(record["acuity"])
        for candidate in ACUITY_PRIORITY:
            if candidate in acuity_variables:
                mapped_variables.append(f"{name}_{candidate}")
                break

        # "scattered high density along the falx" -> parafalcine_density
        if record.get("density"):
            densities = set(density_filter(record["density"]))
            names = {"hyper_dense": "Hyperdense", "hypo_dense": "Hypodense", "iso_dense": "Isodense"}
            plain = [names[d] for d in ("hyper_dense", "iso_dense", "hypo_dense") if d in densities]
            if "mixed_density" in densities or len(plain) > 1:
                direct_variables[f"{name}_density"] = "Mixed"
            elif plain:
                direct_variables[f"{name}_density"] = plain[0]

        # "small subdural hemorrhage medial to the anterior falx" -> description
        for text in record.get("magnitude", []):
            size = magnitude_filter([text])
            if size:
                direct_variables[f"{name}_description"] = {
                    "description_small": "Small", "description_minimal": "Small",
                    "description_moderate": "Medium", "description_large": "Large",
                }[size[0]]
                break

    return (
        mapped_variables,
        direct_variables
    )


def measurement_to_mm_cm(value_text):

    value_text = value_text.lower().replace(" ", "")

    try:
        if value_text.endswith("cm"):
            cm = float(value_text[:-2])
            return round(cm * 10, 2), cm

        if value_text.endswith("mm"):
            mm = float(value_text[:-2])
            return mm, round(mm / 10, 3)

    except ValueError:
        pass

    return None, None
```

```python
ARRAY_DEFAULTS = {

    # Overall
    "laterality": "NR",

    # L Description
    "L_description": "NR",

    # L Morphology
    "L_morphology": "NR",

    # L Acuity
    "L_acuity": "NR",

    # L Density
    "L_mixed_density": "NR",
    "L_hyper_dense": "NR",
    "L_hypo_dense": "NR",
    "L_iso_dense": "NR",
    "L_density_dominance": "NR",

    # L Thickness
    "L_thickness_mm": "NR",
    "L_thickness_cm": "NR",
    "L_thickness_max": "NR",

    # L Locations
    "L_frontal_lobe": "NR",
    "L_parietal_lobe": "NR",
    "L_temporal_lobe": "NR",
    "L_occipital_lobe": "NR",
    "L_convexity_hematoma": "NR",
    "L_holohemispheric_hematoma": "NR",
    "L_vertex": "NR",
    "L_middle_fossa": "NR",

    # R Description
    "R_description": "NR",

    # R Morphology
    "R_morphology": "NR",

    # R Acuity
    "R_acuity": "NR",

    # R Density
    "R_mixed_density": "NR",
    "R_hyper_dense": "NR",
    "R_hypo_dense": "NR",
    "R_iso_dense": "NR",
    "R_density_dominance": "NR",

    # R Thickness
    "R_thickness_mm": "NR",
    "R_thickness_cm": "NR",
    "R_thickness_max": "NR",

    # R Locations
    "R_frontal_lobe": "NR",
    "R_parietal_lobe": "NR",
    "R_temporal_lobe": "NR",
    "R_occipital_lobe": "NR",
    "R_convexity_hematoma": "NR",
    "R_holohemispheric_hematoma": "NR",
    "R_vertex": "NR",
    "R_middle_fossa": "NR",

    # Falx / tentorium compartments
    "parafalcine_hematoma": "NR",
    "parafalcine_location": "NR",
    "parafalcine_density": "NR",
    "parafalcine_description": "NR",
    "tentorium_description": "NR",
    "tentorium_density": "NR",
    "parafalcine_thickness_mm": "NR",
    "parafalcine_thickness_cm": "NR",
    "parafalcine_acuity": "NR",
    "tentorium_hematoma": "NR",
    "tentorium_thickness_mm": "NR",
    "tentorium_thickness_cm": "NR",
    "tentorium_acuity": "NR",

    # Mass Effect
    "mass_effect_present": "NR",

    # Midline Shift
    "midline_shift_present": "NR",
    "midline_shift_mm": "NR",
    "midline_shift_cm": "NR",
    "midline_shift_direction": "NR",

    # Sulcal Effacement
    "effacement_sulci": "NR",
    "effacement_sulci_laterality": "NR",

    # Ventricular Effacement
    "effacement_ventricular_system": "NR",
    "effacement_lateral_ventricle": "NR",
    "l_ventricle_laterality": "NR",
    "effacement_third_ventricle": "NR",
    "effacement_forth_ventricle": "NR",

    # Ventricular Prominence
    "ventricular_prominence": "NR",
    "l_v_prominence_laterality": "NR",
    "ex_vacuo_dilatation": "NR",

    # Herniation
    "herniation_subfalcine": "NR",
    "herniation_uncal": "NR",
    "herniation_transtentorial": "NR",
    "herniation_tonsillar": "NR",

    # Other Pathologies
    "encephalomalacia": "NR",
    "hydrocephalus": "NR",
    "microangiopathy": "NR",
    "cerebral_atrophy": "NR",
    "white_dark_differentiation": "NR",
    "mass": "NR",
    "posterior_fossa_abnormality": "NR",
    "unspecified_infarct": "NR",
    "territorial_infarct": "NR",
    "lacunar_infarct": "NR",
    "cortical_infarct": "NR",
    "territorial_ischemic": "NR",
    "hyperdense_vessel": "NR",

    # Trauma
    "soft_tissue_lesion": "NR",
    "scalp_hematoma": "NR",
    "contusion": "NR",
    "edema": "NR",
    "aneurysm": "NR",

    # Other hemorrhage
    "subarachnoid_hemorrhage": "NR",
    "intraparenchymal_hemorrhage": "NR",
    "intraventricular_hemorrhage": "NR",
    "epidural_hemorrhage": "NR",

    # Fracture
    "orbital_lesion": "NR",
    "bony_lesion": "NR",
    "fracture_location": "NR",
    "fracture_type": "NR",

    # Sinuses / mastoids
    "paranasal_sinuses": "NR",
    "maxillary_sinus": "NR",
    "sphenoid_sinus": "NR",
    "ethmoid_sinus": "NR",
    "mastoid_air_cells": "NR",

}

# variables written as "<prefix><value>" e.g. "mass_effect_present_Mild"
# (longest prefixes first so "effacement_sulci_laterality_" wins over
#  "effacement_sulci_")
VALUE_PREFIXES = [
    "effacement_sulci_laterality_",
    "effacement_sulci_",
    "effacement_ventricular_system_",
    "effacement_lateral_ventricle_",
    "effacement_third_ventricle_",
    "effacement_forth_ventricle_",
    "l_ventricle_laterality_",
    "l_v_prominence_laterality_",
    "mass_effect_present_",
    "midline_shift_present_",
    "herniation_subfalcine_",
    "herniation_uncal_",
    "herniation_transtentorial_",
    "herniation_tonsillar_",
    "cerebral_atrophy_",
    "microangiopathy_",
    "hydrocephalus_",
    "parafalcine_hematoma_",
    "parafalcine_location_",
    "tentorium_hematoma_",
    "ventricular_prominence_",
    "ex_vacuo_dilatation_",
    "subarachnoid_hemorrhage_",
]

INFARCT_KEYS = [
    "territorial_ischemic",
    "unspecified_infarct",
    "territorial_infarct",
    "lacunar_infarct",
    "cortical_infarct",
]

ACUITY_VALUES = {
    "acuity_acute": "Acute",
    "acuity_subacute": "Subacute",
    "acuity_chronic": "Chronic",
    "acuity_acute_on_chronic": "Acute on Chronic",
    "acuity_subacute_on_chronic": "Subacute on Chronic",
    "acuity_acute_subacute": "Acute on Subacute",
}

MORPHOLOGY_VALUES = {
    "dependent": "Dependent",
    "layering": "Layering",
    "septations": "Septations",
}

DESCRIPTION_VALUES = {
    "description_minimal": "Minimal",
    "description_small": "Small",
    "description_moderate": "Medium",
    "description_large": "Large",
}


def value_rank(value):
    """
    How much a value says. A later mention only overwrites an earlier
    one if it says more:  NR < No... < Yes / Yes NS < a grade / a value
    ("Cerebral atrophy." in the impression must not wipe out "Severe"
    from the findings). Equal rank -> the first one said is kept.
    """

    if value is None or value == "NR":
        return 0

    if isinstance(value, str) and value.startswith("No"):
        return 1

    if value in ("Yes", "Yes NS"):
        return 2

    return 3


FRACTURE_LOCATION_ORDER = ["Calvarial", "Skull base", "Facial"]
FRACTURE_TYPE_ORDER = ["Linear", "Nondisplaced", "Displaced", "Depressed", "Comminuted"]


def put(final_variables, key, value):

    old = final_variables.get(key, "NR")

    # fractures in different sentences add up: "Calvarial" + "Facial"
    if key == "fracture_location" and value_rank(old) == 3 and value_rank(value) == 3:
        parts = set(old.split(" + ")) | set(value.split(" + "))
        final_variables[key] = " + ".join(p for p in FRACTURE_LOCATION_ORDER if p in parts)
        return

    # the most severe fracture type is kept
    # infarcts: "No recent territorial infarct" + "chronic PICA territory
    # infarct" -> No Acute (the column is about an acute infarct)
    if key in INFARCT_KEYS:
        if value == "No Acute" and str(old).startswith("Yes Chronic"):
            final_variables[key] = value
            return
        if old == "No Acute" and str(value).startswith("Yes Chronic"):
            return

    if key == "fracture_type" and old in FRACTURE_TYPE_ORDER and value in FRACTURE_TYPE_ORDER:
        if FRACTURE_TYPE_ORDER.index(value) > FRACTURE_TYPE_ORDER.index(old):
            final_variables[key] = value
        return

    if value_rank(value) > value_rank(old):
        final_variables[key] = value


def final_array_function(
    mapped_variables,
    direct_variables,
    final_variables
):

    # -------------------------
    # Populate defaults
    # -------------------------

    for key, value in ARRAY_DEFAULTS.items():

        if key not in final_variables:

            final_variables[key] = value

    # -------------------------
    # Apply mapped variables
    # -------------------------

    for variable in mapped_variables:

        negated = variable.startswith("N_")
        target = variable[2:] if negated else variable

        # -------------------------
        # Which sides have an SDH (for "laterality")
        # -------------------------

        if variable == "sdh_side_L":
            final_variables["_sdh_left"] = True
            continue

        if variable == "sdh_side_R":
            final_variables["_sdh_right"] = True
            continue

        # -------------------------
        # White / gray differentiation
        # -------------------------

        if target in ("white_dark_differentiation", "loss_white_dark_differentiation"):

            if negated or target == "white_dark_differentiation":
                put(final_variables, "white_dark_differentiation", "Preserved")
            else:
                put(final_variables, "white_dark_differentiation", "Loss")

            continue

        # -------------------------
        # "No brain herniation" -> every subtype
        # -------------------------

        if target.startswith("herniation_generic_"):

            if negated:
                for key in ("herniation_subfalcine", "herniation_uncal",
                            "herniation_transtentorial", "herniation_tonsillar"):
                    put(final_variables, key, "No")

            continue

        # -------------------------
        # Infarcts: "No Acute", "No NS", "Yes Chronic", ...
        # -------------------------

        infarct_key = next(
            (key for key in INFARCT_KEYS if target.startswith(key + "_")),
            None
        )

        if infarct_key:

            qualifier = target[len(infarct_key) + 1:]
            put(
                final_variables,
                infarct_key,
                f"{'No' if negated else 'Yes'} {qualifier}"
            )
            continue

        # -------------------------
        # L_ / R_ acuity, morphology, description
        # -------------------------

        if target[:2] in ("L_", "R_") and not negated:

            prefix = target[:2]
            rest = target[2:]

            if rest in ACUITY_VALUES:
                put(final_variables, f"{prefix}acuity", ACUITY_VALUES[rest])
                continue

            if rest in MORPHOLOGY_VALUES:
                put(final_variables, f"{prefix}morphology", MORPHOLOGY_VALUES[rest])
                continue

            if rest in DESCRIPTION_VALUES:
                put(final_variables, f"{prefix}description", DESCRIPTION_VALUES[rest])
                continue

            if rest.startswith("density_dominance_"):
                put(final_variables, f"{prefix}density_dominance", rest[len("density_dominance_"):])
                continue

        # compartment acuity ("parafalcine_acuity_acute")
        if (
            target.startswith(("parafalcine_acuity_", "tentorium_acuity_"))
            and not negated
        ):
            key = "parafalcine_acuity" if target.startswith("parafalcine") else "tentorium_acuity"
            rest = target[len(key) + 1:]
            put(final_variables, key, ACUITY_VALUES.get(rest, ACUITY_VALUES.get("acuity_" + rest, rest)))
            continue

        # -------------------------
        # Simple Yes / No variables
        # -------------------------

        if target in ARRAY_DEFAULTS:

            put(final_variables, target, "No" if negated else "Yes")
            continue

        # -------------------------
        # "<prefix><value>" variables
        # -------------------------

        prefix = next(
            (p for p in VALUE_PREFIXES if target.startswith(p)),
            None
        )

        if prefix:

            key = prefix[:-1]
            value = target[len(prefix):]

            if negated:
                # a negated laterality says nothing
                if "laterality" not in key:
                    put(final_variables, key, "No")
            else:
                put(final_variables, key, value)

            continue

    # -------------------------
    # Apply direct variables (first value said wins)
    # -------------------------

    for key, value in direct_variables.items():

        put(final_variables, key, value)

    return final_variables
```

## BLEED TREE FUNCTION

```python
MEASUREMENT_BUCKETS = (
    "numerical",
    "measurement_linker",
    "measurement_linker_max",
    "measurement_max",
    "measurement_type"
)

CLAUSE_STOP = {
    "period", "comma", "semicolon", "colon",
    "association_linker", "stop_and",
}

# anatomy that means a density / morphology word is NOT about the SDH
NON_SDH_ANATOMY = {
    "cistern", "ventricle", "intraparenchymal", "extracranial_anatomy",
    "bone", "sulci", "tonsils", "sinus",
}

ANAPHOR_WORDS = {"it", "this", "these"}


def number_belongs_to_effect(sent_tokens, i):
    """
    Is the number / measurement at i describing an effect
    ("2 cm midline shift", "midline shift measuring 15 mm")?
    Looks at the clause the number sits in, not a fixed window.
    """

    # back to the start of the clause (max 4 tokens)
    k = i - 1
    steps = 0
    while k >= 0 and steps < 4:
        bucket = sent_tokens[k][1]
        if bucket in CLAUSE_STOP:
            break
        if bucket in EFFECT_FAMILY:
            return True
        if bucket == "bleed_pathology":
            return False
        k -= 1
        steps += 1

    # forward to the end of the clause
    k = i + 1
    while k < len(sent_tokens):
        bucket = sent_tokens[k][1]
        if bucket in CLAUSE_STOP:
            break
        if bucket in EFFECT_FAMILY:
            return True
        if bucket in ("bleed_pathology", "bleed_reference"):
            return False
        k += 1

    return False


def laterality_belongs_to_effect(sent_tokens, i):

    k = i + 1

    while k < len(sent_tokens) and sent_tokens[k][1] in (
        "magnitude", "numerical", "measurement_unit", "uncertainty_linker"
    ):
        k += 1

    return k < len(sent_tokens) and sent_tokens[k][1] in EFFECT_FAMILY


def secondary_thickness(sent_tokens):
    """
    Thickness in a sentence without a bleed word is only about the SDH
    when the sentence points back to it ("It measures 8 mm",
    "This has a maximal depth of 1 cm", "The hematoma measures ...",
    "Maximal thickness 14 mm").
    """

    words = {text.lower() for text, _ in sent_tokens}

    points_back = (
        bool(words & ANAPHOR_WORDS)
        or sentence_has(sent_tokens, {"bleed_reference", "measurement_max"})
        or bool(words & {"thickness", "depth", "width", "thick", "thickest"})
        # "The right measures up to 14 mm and the left measures up to 10 mm"
        or (
            sentence_has(sent_tokens, {"laterality"})
            and bool(words & {"measures", "measuring", "measure", "measured"})
            and not sentence_has(sent_tokens, {"ventricle", "pathology", "bone", "sinus", "intraparenchymal"})
        )
    )

    if not points_back or sentence_has(sent_tokens, {"ignore"}):
        return []

    results = []
    i = 0

    while i < len(sent_tokens):

        bucket = sent_tokens[i][1]

        if bucket in MEASUREMENT_BUCKETS and not number_belongs_to_effect(sent_tokens, i):

            result, next_i = parse_thickness_branch(sent_tokens, i)

            if result:
                said_after_now = any(
                    t.lower() == "now" for t, _ in sent_tokens[max(0, i - 4):i]
                )
                results.append({"thickness": result, "now": said_after_now})

            i = max(next_i, i + 1)
            continue

        i += 1

    # "It was 14 mm and is now 15 mm thick": keep the current size
    if any(r["now"] for r in results):
        results = [r for r in results if r["now"]]

    return [{"thickness": r["thickness"]} for r in results]


def respectively_pairs(sent_tokens):

    words = [t.lower() for t, _ in sent_tokens]

    if "respectively" not in words:
        return []

    sides = [t for t, b in sent_tokens if b == "laterality" and side_of(t) in ("L", "R")]

    sizes = []
    for k, (t, b) in enumerate(sent_tokens):
        if b == "numerical":
            for t2, b2 in sent_tokens[k + 1:k + 4]:
                if b2 == "measurement_unit":
                    sizes.append(f"{t}{t2.lower()}")
                    break

    if len(sides) < 2 or len(sides) != len(sizes):
        return []

    is_max = any(b in ("measurement_max",) for _, b in sent_tokens)

    return [
        {"thickness": {"value": size, "max": is_max, "side": side,
                       "locations": [], "compartment": None}}
        for side, size in zip(sides, sizes)
    ]


EXTENSION_WORDS = {"extension", "extends", "extending", "tracking", "layering", "outlines"}


def compartment_mentions(sent_tokens):
    """
    Laterality + falx / tentorium words in a sentence that talks about
    the SDH extending there ("minor extension along the right side of
    the falx cerebri and right tentorium cerebelli").
    """

    words = {t.lower() for t, _ in sent_tokens}

    if not words & EXTENSION_WORDS:
        return []

    found = [
        (t, b) for t, b in sent_tokens
        if b == "laterality" or (b == "cerebral_location" and t.lower() in COMPARTMENT_WORDS)
    ]

    if not any(b == "cerebral_location" for _, b in found):
        return []

    return found


def falx_subregion(sent_tokens):
    """
    "along the anterior falx", "the falx posteriorly", "posterior aspect of
    the falx" -> parafalcine_location
    """

    text = " ".join(t.lower() for t, _ in sent_tokens)

    # "falcotentorial extension": where the falx meets the tentorium, at
    # the back of the falx
    if re.search(r"falco-?\s?tentorial", text):
        return [{"falx_subregion": "Posterior"}]

    match = re.search(
        r"(anterior|posterior)\w*\s+(aspect\s+of\s+the\s+|part\s+of\s+the\s+|left\s+|right\s+)?(falx|falcine|para-?falcine|interhemispheric)"
        r"|(falx|falcine|interhemispheric)\w*(\s+\w+){0,2}\s+(anterior|posterior)",
        text,
    )

    if not match:
        return []

    where = match.group(1) or match.group(6)

    # "parafalcine area anteriorly and posteriorly"
    if re.search(r"anterior", text) and re.search(r"posterior", text):
        return [{"falx_subregion": "Anterior/posterior"}]

    return [{"falx_subregion": where.capitalize()}]


def build_bleed_tree(sent_tokens, is_bleed_sentence, existing_tree=None):

    # ============================
    # Initialize / reuse tree
    # ============================
    if existing_tree is not None:
        tree = existing_tree
    else:
        tree = {"pathology": None, "children": []}

        for text, bucket in sent_tokens:
            if bucket == "bleed_pathology":
                tree["pathology"] = text
                break

        if not tree["pathology"]:
            return None

    # -----------------------------------
    # SECONDARY SENTENCE
    #   only the association (effects, density, ...) and an
    #   anaphoric thickness. Locations / lateralities in these
    #   sentences describe the effect, not the SDH.
    # -----------------------------------
    if not is_bleed_sentence:

        assoc, _ = parse_association(sent_tokens, 0)

        # density / acuity / morphology words next to other anatomy
        # ("hyperdensity within the basal cisterns") are not the SDH
        words = {t.lower() for t, _ in sent_tokens}
        points_back = (
            sentence_has(sent_tokens, {"bleed_reference"})
            or bool(words & ANAPHOR_WORDS)
            or bool(words & {"component", "components", "blood", "compartment", "hematoma"})
            or any(re.search(r"hematocrit|fluid-fluid|fluid level|abnormality|collection|septation", w) for w in words)
        )

        # "Hypoattenuation within the left frontal lobe", "hypodense nodules in
        # both thyroid lobes", "hyperdensity within the basal cisterns"
        if (
            (sentence_has(sent_tokens, NON_SDH_ANATOMY | {"cerebral_location", "ignore"})
             or not points_back)
            and not points_back
        ):
            assoc = [
                a for a in assoc
                if not any(k in a for k in ("density", "acuity", "morphology"))
            ]

        if assoc:
            tree["children"].append({"association": assoc})

        tree["children"].extend(
            secondary_thickness(sent_tokens)
        )

        tree["children"].extend(falx_subregion(sent_tokens))

        # "Mild extension along the anterior falx" -> compartment only
        compartment_tokens = compartment_mentions(sent_tokens)
        if compartment_tokens:
            tree["children"].append(
                {"compartment_location": compartment_tokens}
            )

        return tree

    # -----------------------------------
    # PRIMARY SENTENCE
    # -----------------------------------
    tree["children"].append({"sentence_start": True})

    # "right frontoparietal and left ... convexities, measuring 1.2 cm and
    # 2.3 cm in thickness respectively" -> pair sides and sizes in order
    # (added after the sentence is read, so the sides are said first)
    paired = respectively_pairs(sent_tokens)

    i = 0
    while i < len(sent_tokens):

        text, bucket = sent_tokens[i]

        # -----------------------------------
        # which bleed is being talked about now
        # -----------------------------------
        if bucket == "bleed_pathology":

            mention = {
                "bleed_mention": bleed_type(text, sent_tokens, i),
                "text": text
            }

            # "subdural fluid collections, fairly low in density consistent
            # with chronic subdural hematomas": the same bleed named again,
            # not a second one
            earlier_bleed = any(b == "bleed_pathology" for _, b in sent_tokens[:i])
            k = i - 1
            while k >= 0 and (
                sent_tokens[k][1] in ("acuity", "density", "magnitude", "laterality", "morphology")
                or sent_tokens[k][0].lower() in ("a", "an", "the")
            ):
                k -= 1
            said_before = " ".join(t.lower() for t, _ in sent_tokens[max(0, k - 1):k + 1])
            if earlier_bleed and re.search(
                r"(consistent with|compatible with|in keeping with|suggestive of|representing|represents|suspicious for)$",
                said_before
            ):
                mention["restates"] = True

            if i + 1 < len(sent_tokens) and sent_tokens[i + 1][1] == "laterality" and not re.search(
                r"falx|falcine|tentori|interhemispheric",
                " ".join(t.lower() for t, _ in sent_tokens[i + 2:i + 5])
            ):
                mention["side_after"] = sent_tokens[i + 1][0]

            tree["children"].append(mention)

            i += 1
            continue

        # "small contusion in the temporal lobe with intra-axial/subdural
        # component": the size and place said for a contusion are the
        # contusion's, not the SDH's (read like another bleed)
        if bucket == "pathology" and "contusion" in text.lower():
            tree["children"].append({"bleed_mention": "contusion", "text": text})
            i += 1
            continue

        if bucket == "negation":

            branches, next_i = parse_negation_branch(
                sent_tokens,
                i
            )

            if branches:
                tree["children"].append(
                    {"negation": branches}
                )

            i = max(next_i, i + 1)
            continue

        # -----------------------------------
        # ASSOCIATION (primary bleed sentence)
        # -----------------------------------
        if bucket == "association_linker":

            next_bucket = sent_tokens[i+1][1] if i+1 < len(sent_tokens) else None

            # "with 5 mm rightward midline shift" / "with left-sided midline shift"
            opens_numeric_effect = (
                next_bucket in ("numerical", "laterality")
                and (
                    number_belongs_to_effect(sent_tokens, i + 1)
                    if next_bucket == "numerical"
                    else laterality_belongs_to_effect(sent_tokens, i + 1)
                )
            )

            # "with mild mass effect", "with sulcal effacement", "with left
            # ventricular effacement" open an association, but "with the
            # large intermediate density right SDH" does not: the words
            # after "with" have to lead to an effect
            k = i + 1
            while k < len(sent_tokens) and (
                sent_tokens[k][1] in BETWEEN_GRADE_AND_EFFECT
                or sent_tokens[k][1] == "subregional_anatomy"
            ):
                k += 1
            grade_leads_to_effect = (
                k > i + 1
                and k < len(sent_tokens)
                and sent_tokens[k][1] in EFFECT_FAMILY
            )

            # Open association ONLY if next bucket is magnitude or mass effect
            if (
                grade_leads_to_effect
                or next_bucket in EFFECT_FAMILY
                or opens_numeric_effect
            ):
                assoc, next_i = parse_association(sent_tokens, i)
                if assoc:
                    tree["children"].append({"association": assoc})
                i = next_i
                continue

            # Otherwise → SKIP association_linker
            i += 1
            continue

        # -----------------------------------
        # DIRECT ATTACH (after association logic)
        # -----------------------------------
        if bucket in DIRECT_ATTACH:

            # "small volume of ... blood", "small amount of", "small component":
            # the size of a part, not of the whole SDH
            if bucket == "magnitude" and (
                " ".join(t.lower() for t, _ in sent_tokens[i + 1:i + 3])
                in ("volume of", "amount of", "component", "components", "foci of", "areas of")
                or re.match(r"(acute|subacute|chronic|hyperdense|hypodense|high density|low density)\s+(superimposed\s+)?(components?|foci|focus)\b",
                            " ".join(t.lower() for t, _ in sent_tokens[i + 1:i + 4]))
            ):
                i += 1
                continue

            # "with left-sided midline shift": a side in front of an effect
            # is the effect's direction, not the side of the SDH
            if bucket == "laterality" and laterality_belongs_to_effect(sent_tokens, i):
                i += 1
                continue

            tree["children"].append((text, bucket))

            # "acute blood components" / "acute hemorrhagic components"
            # are the hyperdense part of the SDH
            following = " ".join(
                t.lower() for t, _ in sent_tokens[i + 1:i + 3]
            )
            if (
                bucket == "acuity"
                and text.lower() in ("acute", "acute on chronic")
                and following.startswith(("component", "blood", "hemorrhagic component", "superimposed component"))
            ):
                tree["children"].append(("hyperdense", "density"))
                # an acute COMPONENT = more than one density, and the
                # collection is older than the acute part: acute on chronic
                # ("acute blood products over the parietal region" is the
                # whole bleed: hyperdense only)
                if following.startswith(("component", "hemorrhagic component", "superimposed component")):
                    tree["children"].append(("mixed density", "density"))
                    tree["children"].append(("acute on chronic", "acuity"))

            # "high density component", "hyperdense components": a part of
            # different density
            if (
                bucket == "density"
                and re.search(r"hyper|high", text.lower())
                and following.startswith(("component", "area", "focus", "foci"))
            ):
                tree["children"].append(("mixed density", "density"))

            i += 1
            continue

        # -----------------------------------
        # THICKNESS
        # -----------------------------------
        if bucket in MEASUREMENT_BUCKETS:

            if paired:
                i += 1
                continue

            if not number_belongs_to_effect(sent_tokens, i):

                result, next_i = parse_thickness_branch(
                    sent_tokens,
                    i
                )

                if result:
                    tree["children"].append(
                        {"thickness": result}
                    )

                i = max(next_i, i + 1)
                continue

        # -----------------------------------
        # LOCATION BRANCH
        # -----------------------------------
        if bucket == "location_linker":
            branch, next_i = parse_location_branch(sent_tokens, i)
            if branch:
                tree["children"].append({"location": branch})
            i = next_i
            continue

        # -----------------------------------
        # Direct Mass Effect Opener
        # -----------------------------------
        if bucket in EFFECT_FAMILY:

            assoc, next_i = parse_association(
                sent_tokens,
                i
            )

            if assoc:
                tree["children"].append(
                    {"association": assoc}
                )

            i = next_i
            continue

        # -----------------------------------
        # DEFAULT
        # -----------------------------------
        i += 1

    tree["children"].extend(paired)
    tree["children"].extend(falx_subregion(sent_tokens))

    return tree
```

#### Thickness Helper

```python
def get_measurement_branch(
    tokens,
    start_i
):

    branch = []
    i = start_i

    while i < len(tokens):

        _, bucket = tokens[i]

        if bucket in (
            "association_linker",
            "stop_and",
            "period",
            "comma",
            "semicolon",
        ):
            break

        branch.append(
            tokens[i]
        )

        i += 1

    return branch
```

```python
def measurement_branch_contains_effect(
    branch
):

    return any(
        bucket in EFFECT_FAMILY
        for _, bucket in branch
    )
```

### AND function

```python
def classify_and(sent_tokens, idx):

    if idx <= 0 or idx >= len(sent_tokens) - 1:
        return "unknown_and"

    left_text, left_bucket = sent_tokens[idx - 1]
    right_text, right_bucket = sent_tokens[idx + 1]

    # -----------------------------
    # 1. Paired measurements
    # -----------------------------
    if "respectively" in right_text.lower():
        return "paired_and"

    # -----------------------------
    # 1.5 Directional laterality
    # -----------------------------
    if right_text.lower() in (
        "rightward",
        "leftward"
    ):
        return "stop_and"

    # -----------------------------
    # 2. Anatomy families → sandwich_and
    # -----------------------------
    ANATOMY_FAMILY = {
        "intraparenchymal",
        "ventricle",
        "cistern",
        "cerebral_location",
        "extracranial_anatomy",
        "subregional_anatomy",
        "laterality",
        "sulci",
        "sinus",
        "bone",
    }

    if left_bucket in ANATOMY_FAMILY and right_bucket in ANATOMY_FAMILY:
        return "sandwich_and"

    # -----------------------------
    # 3. Pathology families → stop_and
    # -----------------------------
    PATHOLOGY_FAMILY = {
        "bleed_pathology",
        "mass_effect",
        "effacement",
        "compression",
        "midline_shift",
        "pathology",
        "bone_pathology",
    }

    if left_bucket in PATHOLOGY_FAMILY and right_bucket in PATHOLOGY_FAMILY:
        return "stop_and"

    # -----------------------------
    # 4. Mixed categories → stop_and
    # -----------------------------
    if left_bucket != right_bucket:
        return "stop_and"

    # -----------------------------
    # 5. Same bucket → sandwich_and
    # -----------------------------
    return "sandwich_and"
```

### Thickness Parser

```python
COMPARTMENT_WORDS = {
    "para-falcine": "parafalcine",
    "falcotentorial": "parafalcine",
    "falco tentorial": "parafalcine",
    "interhemispheric fissure": "parafalcine",
    "falx": "parafalcine",
    "falx cerebri": "parafalcine",
    "parafalcine": "parafalcine",
    "falcine": "parafalcine",
    "interhemispheric": "parafalcine",
    "tentorium": "tentorium",
    "tentorium cerebelli": "tentorium",
    "tentorial": "tentorium",
}


def clause_laterality_before(tokens, start_i):
    """
    Laterality said earlier in the same clause
    ("On the right this measures up to 1.4 cm").
    """

    k = start_i - 1

    while k >= 0:

        text, bucket = tokens[k]

        # a comma right after an opening "On the left," does not end the clause
        if bucket == "comma" and k <= 4 and any(b == "laterality" for _, b in tokens[:k]):
            k -= 1
            continue

        if bucket in ("period", "comma", "semicolon", "colon", "stop_and"):
            break

        if bucket in ("numerical",):
            break

        if bucket == "laterality":
            return text

        k -= 1

    return None


# "measuring 1.9 cm in depth over the right parietal convexity": should the
# lobe named for the place of the measurement be a location of the SDH?
# The human sheets so far always record it (9 of 9 reports), so True.
# False keeps only lobes said as an extent ("extending ... into ...").
MEASUREMENT_SITE_IS_LOCATION = True


def parse_thickness_branch(
    tokens,
    start_i
):
    """
    Save tokens from a measurement word until the clause ends.
    Returns ({"value": "2.3cm", "max": bool, "side": "left",
              "locations": [...], "compartment": "tentorium"}, next_i)

    - stops at a SECOND number, so "2.5 cm ... left frontal region and
      4 mm ... right frontal region" gives two measurements
    - "11.0 x 2.5 cm" keeps the smallest dimension (the thickness)
    - laterality / location words met inside the clause tell which side
      and where the measurement was taken
    """

    has_thickness_context = any(
        bucket in (
            "measurement_linker_max",
            "measurement_max",
            "measurement_type",
        )
        for _, bucket in tokens
    )

    # "Right subdural, 1.0 cm"
    if not has_thickness_context and tokens[start_i][1] == "numerical":
        before = tokens[max(0, start_i - 3):start_i]
        has_thickness_context = (
            any(b in ("bleed_pathology", "bleed_reference") for _, b in before)
            and start_i + 1 < len(tokens)
            and tokens[start_i + 1][1] == "measurement_unit"
        )

    # "a 16 mm left parafalcine subdural hematoma" / "a hypodense 9 mm collection"
    if not has_thickness_context and tokens[start_i][1] == "numerical":

        window = tokens[start_i + 1: start_i + 6]

        has_thickness_context = (
            any(b == "measurement_unit" for _, b in window[:1])
            and any(b in ("bleed_pathology", "bleed_reference") for _, b in window)
        )

    if not has_thickness_context:

        return None, start_i+1

    i = start_i
    numbers = []
    unit = None
    is_max = False
    side = None
    locations = []

    # "1.9 cm in depth over the right parietal convexity": a lobe named for
    # the place of the measurement is not where the whole bleed is; lobes
    # after "extending from / along / into" are (how far the bleed reaches)
    extent = False
    unit_at = None

    STOP_BUCKETS = {
        "mass_effect",
        "herniation_subtype",
        "association_linker",
        "comma",
        "period",
        "semicolon",
        "colon",
        "stop_and",
        "negation",
    }

    SKIP_BUCKETS = {
        "noticing",
        "location_linker",
        "measurement_type",
        "measurement_linker",
        "uncertainty_linker",
        "subregional_anatomy",
        "magnitude",
        "bleed_pathology",
        "bleed_reference",
        "intraparenchymal",
        "extracranial_anatomy",
        "bone",
        "sandwich_and",
        "acuity",
        "density",
        None,
    }

    while i < len(tokens):
        text, bucket = tokens[i]

        if bucket in STOP_BUCKETS:
            break

        if bucket == "numerical":

            previous = tokens[i - 1][0].lower() if i > 0 else ""

            # "3.6 mm anteriorly and posteriorly 1 cm": two sizes of the same
            # collection in one breath -> keep the larger
            between = [b for _, b in tokens[unit_at + 1:i]] if unit_at is not None else ["x"]
            if (
                numbers and unit is not None and previous != "x"
                and all(b in ("subregional_anatomy", "sandwich_and", None) for b in between)
                and i + 1 < len(tokens) and tokens[i + 1][1] == "measurement_unit"
            ):
                def in_mm(n, u):
                    return float(n) * (10 if u.startswith("c") else 1)
                try:
                    if in_mm(text, tokens[i + 1][0].lower()) > in_mm(min(numbers, key=float), unit):
                        numbers, unit = [text], tokens[i + 1][0].lower()
                except ValueError:
                    pass
                unit_at = i + 1
                i += 2
                continue

            # second number that is not "A x B" -> next measurement
            if numbers and unit is not None and previous != "x":
                break

            if numbers and previous != "x" and unit is None:
                numbers = []

            numbers.append(text)
            i += 1
            continue

        if bucket == "measurement_unit":
            if numbers and unit is None:
                unit = text.lower()
                unit_at = i
            i += 1
            continue

        if bucket in ("measurement_max", "measurement_linker_max"):
            is_max = True
            i += 1
            continue

        if bucket == "laterality":
            if side is None:
                side = text
            i += 1
            continue

        if bucket == "location_linker":
            linker = text.lower()
            if re.search(r"extend|along|into|from|through|across|wrap|toward", linker):
                extent = True
            elif re.search(r"\b(over|overlying|in|at|within|adjacent|near)\b", linker):
                extent = False
            i += 1
            continue

        # "... into the middle cranial fossa": a fossa is a location here
        if bucket == "cerebral_location" or (bucket == "bone" and "fossa" in text.lower()):
            is_lobe = re.search(r"frontal|parietal|temporal|occipital", text.lower())
            if is_lobe and not extent and not MEASUREMENT_SITE_IS_LOCATION:
                i += 1
                continue
            locations.append((text, bucket))
            i += 1
            continue

        if bucket in SKIP_BUCKETS or text.lower() == "x":
            i += 1
            continue

        break

    if not numbers or not unit:
        return None, i

    try:
        value_number = min(float(n) for n in numbers)
    except ValueError:
        return None, i

    if side is None:
        side = clause_laterality_before(tokens, start_i)

    compartment = None
    for text, bucket in locations:
        compartment = COMPARTMENT_WORDS.get(text.lower(), compartment)

    result = {
        "value": f"{value_number:g}{unit}",
        "max": is_max,
        "side": side,
        "locations": locations,
        "compartment": compartment,
    }

    return result, i
```

### Mass Effect Parser

```python
def parse_numeric_mass_effect(tokens, start_i):

    i = start_i

    value = None
    unit = None
    modifier = None
    direction = None

    # ============================
    # Step 1: numeric value
    # ============================
    if tokens[i][1] == "numerical":
        value = tokens[i][0]
        i += 1
    else:
        return None, start_i

    # ============================
    # Step 2: measurement unit
    # ============================
    if i < len(tokens) and tokens[i][1] == "measurement_unit":
        unit = tokens[i][0]
        i += 1

    # ============================
    # Step 3: optional modifier
    # ============================
    if i < len(tokens) and tokens[i][1] == "modifier":
        modifier = tokens[i][0]
        i += 1

    # ============================
    # Step 4: scan forward until STOP
    # ============================
    while i < len(tokens):
        text, bucket = tokens[i]

        # STOP CONDITIONS
        if bucket in MASS_NUM_STOP:
            break

        # Laterality (direction)
        if bucket == "laterality":
            direction = text
            i += 1
            continue

        # Skip harmless joiners
        if bucket in ("sandwich_and",):
            i += 1
            continue

        # Unknown → STOP
        break

    # ============================
    # Return clean measurement block
    # ============================
    measurement = {
        "value": value,
        "unit": unit,
        "modifier": modifier,
        "direction": direction
    }

    return measurement, i
```

### Negation Parser

```python
def parse_negation_branch(tokens, start_i):
    """
    Read a negated LIST, the way a reader does:

        "No evidence of acute infarct, mass or hydrocephalus"
        "No acute parenchymal or subarachnoid hemorrhage"
        "no midline shift or brain herniation"

    Every head (pathology / bleed / effect / density) after the negation
    becomes its own negated branch. Tokens are saved until a head is hit:
      - magnitude / laterality / territorial words  -> the NEXT head only
      - acuity ("acute", "recent")                   -> every head in the list
      - anatomy before a head                        -> that head
      - anatomy after a head (via "of the", "in the") -> that head
    The list ends at a period / semicolon / colon or at "but".
    """

    branches = []
    pending_modifiers = []
    pending_anatomy = []
    list_acuity = []
    current = None

    i = start_i + 1

    EFFECTS = {
        "mass_effect",
        "effacement",
        "compression",
        "herniation_subtype"
    }

    STOP_BUCKETS = {
        "period",
        "semicolon",
        "colon",
        "title",
    }

    STOP_WORDS = {"but", "however", "although", "whereas"}

    ANATOMY = {
        "extracranial_anatomy", "bone", "sinus", "intraparenchymal",
        "sulci", "ventricle", "cistern", "cerebral_location", "tonsils",
    }

    PASS_THROUGH = {
        None, "comma", "logical_connector", "sandwich_and", "stop_and",
        "noticing", "association_linker", "uncertainty_linker",
        "subregional_anatomy", "negation", "normal_state",
        "ignore", "spine",
    }

    # "The temporal horns are not dilated", "Left-sided cerebral sulci are
    # not effaced": the subject said in front of "not" is the anatomy
    k = start_i - 1
    subject = []
    while k >= 0:
        t_text, t_bucket = tokens[k]
        if t_bucket in ANATOMY or t_bucket == "laterality":
            subject.append((t_text, t_bucket))
        elif t_bucket not in (None, "noticing", "subregional_anatomy", "sandwich_and"):
            break
        k -= 1
    for t_text, t_bucket in reversed(subject):
        if t_bucket == "laterality":
            pending_modifiers.append((t_text, t_bucket))
        else:
            pending_anatomy.append((t_text, t_bucket))

    def open_head(head):
        head["modifiers"] = pending_modifiers + list_acuity
        head["anatomy"] = list(pending_anatomy)
        pending_modifiers.clear()
        pending_anatomy.clear()
        branches.append(head)
        return head

    while i < len(tokens):

        text, bucket = tokens[i]

        if bucket in STOP_BUCKETS or text.lower() in STOP_WORDS:
            break

        # -----------------------------
        # anatomy AFTER a head ("fractures of the calvarium")
        # -----------------------------
        if bucket == "location_linker":

            k = i + 1
            found = []

            while k < len(tokens):

                t_text, t_bucket = tokens[k]

                if t_bucket in ANATOMY or t_bucket == "laterality":
                    found.append((t_text, t_bucket))
                    k += 1
                    continue

                if t_bucket in (None, "subregional_anatomy", "location_linker", "sandwich_and"):
                    k += 1
                    continue

                break

            if current is not None:
                if "effect_name" in current:
                    current.setdefault("location", []).extend(found)
                else:
                    current["anatomy"].extend(found)
            else:
                pending_anatomy.extend(found)

            i = k
            continue

        # -----------------------------
        # saved modifiers
        # -----------------------------
        if bucket == "acuity":
            list_acuity.append((text, bucket))
            i += 1
            continue

        if bucket in ("magnitude", "laterality", "territorial", "cortical"):
            pending_modifiers.append((text, bucket))
            i += 1
            continue

        if bucket in ANATOMY:
            pending_anatomy.append((text, bucket))
            i += 1
            continue

        # -----------------------------
        # heads
        # -----------------------------
        if bucket in ("pathology", "bone_pathology"):
            current = open_head({"pathology": [(text, bucket)]})
            i += 1
            continue

        # "no extradural or subarachnoid hemorrhage": a bare "extradural"
        # is only a bleed inside a negated list (elsewhere it can be
        # "ventral extradural thickening" of the spine)
        if bucket is None and text.lower() in ("extradural", "extra-dural", "epidural"):
            current = open_head({"bleed": (text, "epidural")})
            i += 1
            continue

        if bucket == "bleed_pathology":

            btype = bleed_type(text, tokens, i)

            # "no intraventricular extension of hemorrhage": the generic
            # word repeats the head before it
            if btype == "generic" and current is not None and "bleed" in current:
                i += 1
                continue

            # "No new or expanding intracranial hemorrhage" is about the SDH
            # not growing, not about other bleeds ("No other sites of acute
            # intracranial hemorrhage" does negate the other bleeds)
            said_before = " ".join(t.lower() for t, _ in tokens[start_i:i])
            if btype == "generic" and (
                re.search(r"\b(expanding|enlarging|increas\w*)\b", said_before)
                or re.search(r"\b(component|portion|part)s?\s+of\s+(this|the)\s*$", said_before)
            ):
                i += 1
                continue

            # "no new or expanding intracranial hematoma / collection" is
            # about the SDH, not about an intraparenchymal bleed
            if btype == "generic" and text.lower().split()[-1] in ("hematoma", "collection", "hematomas", "collections"):
                btype = "subdural"

            current = open_head({
                "bleed": (text, btype)
            })
            i += 1
            continue

        if (
            bucket in EFFECTS
            and text.lower() in ("herniation", "brain herniation")
            and current is not None
            and current.get("effect_name", ("", ""))[1] == "herniation_subtype"
        ):
            i += 1
            continue

        if bucket in EFFECTS:

            magnitude = [
                (t, b) for t, b in pending_modifiers
                if b == "magnitude"
            ]
            location = [
                (t, b) for t, b in pending_modifiers + pending_anatomy
                if b != "magnitude"
            ]

            current = open_head({"effect_name": (text, bucket)})

            if magnitude:
                current["magnitude"] = magnitude
            if location:
                current["location"] = location

            i += 1
            continue

        if bucket == "density":
            current = open_head({"density": [(text, bucket)]})
            i += 1
            continue

        # "no ventricular dilatation", "temporal horns are not dilated"
        if bucket == "prominence":
            current = open_head({"prominence": (text, bucket)})
            i += 1
            continue

        if bucket in PASS_THROUGH:
            i += 1
            continue

        # anything else (numbers, measurements, sinus findings...) ends the list
        break

    return branches, i
```

### Location Parser

```python
def parse_location_branch(sent_tokens, start_idx):

    branch = []
    i = start_idx + 1

    while i < len(sent_tokens):
        text, bucket = sent_tokens[i]

        # --- CONDITIONAL STOP: association_linker ---
        if bucket == "association_linker":
            # Look ahead safely
            if i + 1 < len(sent_tokens):
                next_bucket = sent_tokens[i + 1][1]

                # If next token is a location linker → continue scanning
                if next_bucket == "location_linker":
                    i += 1
                    continue

            # Otherwise → STOP
            break

        # --- NORMAL STOP TOKENS ---
        if bucket in LOCATION_STOP:
            break

        # --- ACCEPTABLE LOCATION TOKENS ---
        if bucket in LOCATION_ACCEPTABLE:
            branch.append((text, bucket))
            i += 1
            continue

        # --- OTHERWISE IGNORE AND CONTINUE ---
        i += 1

    return branch, i
```

### Association Parser

```python
# what may sit between a grade and its effect: "significant 1.4 cm midline
# shift", "mild leftward midline shift", "partial effacement"
BETWEEN_GRADE_AND_EFFECT = {
    "magnitude", "numerical", "measurement_unit", "laterality",
    "uncertainty_linker", "sulci", "ventricle", "uncus",
}


def adjacent_magnitude(tokens):
    """
    Walk back from the end of `tokens` (the words just before an effect)
    and keep the grades until something else is hit, so "marked
    progression in the midline shift" does not grade the shift.
    """

    found = []

    for text, bucket in reversed(tokens):

        if bucket not in BETWEEN_GRADE_AND_EFFECT:
            break

        if bucket == "magnitude":
            found.append((text, bucket))

    return list(reversed(found))


def parse_association(tokens, start_i):

    # 1. Slice tokens into segments using STOP buckets
    segments = []
    current_segment = []

    negation_groups = []

    i = start_i
    while i < len(tokens):
        text, bucket = tokens[i]

        # NEGATION → read the whole "no A or B" list on the full sentence,
        # before "or" / commas cut it into segments
        if bucket == "negation":
            if current_segment:
                segments.append(current_segment)
                current_segment = []

            neg_branches, next_i = parse_negation_branch(tokens, i)

            if neg_branches:
                negation_groups.append({"negation": neg_branches})

            i = max(next_i, i + 1)
            continue

        # "secondary to left-sided midline shift, measuring 1.6 cm":
        # a comma followed by a measurement keeps the clause going
        if (
            bucket == "comma"
            and i + 1 < len(tokens)
            and tokens[i + 1][1] in ("measurement_type", "measurement_linker_max")
        ):
            i += 1
            continue

        # HARD STOP → end entire association function
        if bucket in ASSOCIATION_HARD_STOP:
            if current_segment:
                segments.append(current_segment)
                current_segment = []
            break

        # "severe associated mass effect": the magnitude is still waiting
        # for its effect, so the linker does not cut the segment
        if (
            bucket == "association_linker"
            and current_segment
            and current_segment[-1][1] == "magnitude"
        ):
            i += 1
            continue

        # BRANCH STOP → end current segment, start new one
        if bucket in ASSOCIATION_BRANCH_STOP:
            if current_segment:
                segments.append(current_segment)
                current_segment = []
            i += 1
            continue

        current_segment.append((text, bucket))
        i += 1

    # Add last segment if any
    if current_segment:
        segments.append(current_segment)

    # 2. Process each segment
    associations = list(negation_groups)
    last_effect = None
    carried = []                # "Adjacent sulcal, right lateral ventricle ... effacement"

    for seg in segments:

        seg = carried + seg
        carried = []

        # "effacement of the right lateral ventricle as well as the third
        # ventricle": a segment with only more ventricles / sulci and no
        # effect of its own continues the previous effect's location
        content = [(t, b) for t, b in seg if b not in (None, "sandwich_and", "subregional_anatomy")]
        only_anatomy = (
            content
            and all(b in ("ventricle", "sulci", "cistern", "laterality") for _, b in content)
            and any(b != "laterality" for _, b in content)
        )

        if only_anatomy and last_effect is not None:
            last_effect.setdefault("location", []).extend(content)
            continue

        # no effect yet: the anatomy waits for the effect said after it
        if only_anatomy:
            carried = list(seg)
            continue
               
        # -------------------------
        # Numeric Mass Effect
        # -------------------------

        if (
            any(b == "numerical" for _, b in seg)
            and
            any(b in EFFECT_FAMILY for _, b in seg)
        ):

            associations.append(
                num_mass(seg)
            )

        current = None
        pre_effect_tokens = []
        j = 0

        while j < len(seg):

            text, bucket = seg[j]

            # ⭐ NEGATION HANDLING
            if bucket == "negation":

                neg_branch, next_j = (
                    parse_negation_branch(
                        seg,
                        j
                    )
                )

                associations.append(
                    {"negation": [neg_branch]}
                )

                j = next_j
                continue

            # EFFECT OPENER
            if bucket in EFFECT_FAMILY:
               
                if current:
                    associations.append(current)

                current = {
                    "effect_name": (text, bucket)
                }

                for t_text, t_bucket in pre_effect_tokens:

                    if t_bucket in (
                        "laterality",
                        "cerebral_location",
                        "sulci",
                        "ventricle",
                        "uncus",
                    ):

                        current.setdefault(
                            "location",
                            []
                        ).append(
                            (t_text, t_bucket)
                        )

                for t_text, t_bucket in adjacent_magnitude(pre_effect_tokens):

                    current.setdefault(
                        "magnitude",
                        []
                    ).append(
                        (t_text, t_bucket)
                    )

                # "adjacent sulcal effacement" -> the side of the SDH
                if any(t.lower() == "adjacent" for t, _ in pre_effect_tokens):
                    current["adjacent"] = True

                pre_effect_tokens = []
                j += 1
                continue

            # DENSITY OPENER
            if bucket == "density":
            
                if current:
                    associations.append(current)
            
                current = {
                    "density": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # ACUITY OPENER
            if bucket == "acuity":
            
                if current:
                    associations.append(current)
            
                current = {
                    "acuity": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # MORPHOLOGY OPENER
            if bucket == "morphology":
            
                if current:
                    associations.append(current)
            
                current = {
                    "morphology": [
                        (text, bucket)
                    ]
                }
            
                j += 1
                continue
            
            
            # PROMINENCE OPENER
            if bucket == "prominence":
            
                if current:
                    associations.append(current)
            
                current = {
                    "prominence": (text, bucket)
                }

                grades = adjacent_magnitude(pre_effect_tokens)
                if grades:
                    current["magnitude"] = grades
            
                for t_text, t_bucket in pre_effect_tokens:
            
                    if t_bucket in (
                        "laterality",
                        "ventricle",
                    ):
            
                        current.setdefault(
                            "location",
                            []
                        ).append(
                            (t_text, t_bucket)
                        )
                        
                pre_effect_tokens = []
                j += 1
                continue

            # LOCATION LINKER
            if bucket == "location_linker" and current:

                # "effacement in both cerebral hemispheres"
                if "both" in text.lower():
                    current["both"] = True

                loc_list = []
                k = j + 1

                while k < len(seg):

                    t_text2, t_bucket2 = seg[k]

                    if t_bucket2 in ASSO_LOCATION_STOP:
                        break

                    if t_bucket2 == "sandwich_and":
                        k += 1
                        continue

                    if t_bucket2 in (
                        None,
                        "location_linker",
                        "subregional_anatomy"
                    ):
                        k += 1
                        continue

                    loc_list.append(
                        (t_text2, t_bucket2)
                    )

                    k += 1

                if loc_list:

                    current.setdefault(
                        "location",
                        []
                    ).extend(
                        loc_list
                    )

                j = k 
                continue

            # Store pre-effect tokens
            pre_effect_tokens.append(
                (text, bucket)
            )

            j += 1

        # location words AFTER the effect with no linker
        # ("Mild effacement left lateral ventricle")
        if current and pre_effect_tokens:

            # "sulcal effacement in both cerebral hemispheres"
            if any(t.lower().startswith("both") for t, _ in pre_effect_tokens):
                current["both"] = True

            trailing = [
                (t_text, t_bucket)
                for t_text, t_bucket in pre_effect_tokens
                if t_bucket in (
                    "laterality", "sulci", "ventricle", "cistern",
                    "uncus", "intraparenchymal", "cerebral_location",
                )
            ]

            if trailing and "location" not in current:
                current["location"] = trailing

            elif trailing and not any(
                b == "laterality" for _, b in current.get("location", [])
            ):
                current["location"] = current["location"] + [
                    (t, b) for t, b in trailing if b == "laterality"
                ]

        if current:
            associations.append(current)

        last_effect = next(
            (a for a in reversed(associations) if "effect_name" in a), None
        )

    return associations, i
```

### Mass Effect Num Collector

```python
def num_mass(collected):

    value = None
    unit = None
    direction = None
    effect_name = None
    magnitude = []
    numbers = []                # numbers said before the unit ("8-9 mm")

    # Extract components from collected tokens
    for item in collected:

        if (
            isinstance(item, tuple)
            and isinstance(item[1], str)
            and item[1] in (
                "mass_effect",
                "herniation_subtype"
            )
        ):

            text = item[0].lower()

            # Always prefer midline shift
            if "midline" in text and text != "off the midline":

                effect_name = item

            # Only use off the midline
            # if nothing better found
            elif (
                text == "off the midline"
                and effect_name is None
            ):

                effect_name = (
                    "midline shift",
                    "mass_effect"
                )

            elif effect_name is None:

                effect_name = item

        # numeric value: a range "8-9 mm" keeps the lower number,
        # the same as hematoma thickness
        if (
            isinstance(item, tuple)
            and item[1] == "numerical"
            and unit is None
        ):
            numbers.append(item[0])
            value = item[0]

        # measurement unit
        if (
            isinstance(item, tuple)
            and item[1] == "measurement_unit"
            and unit is None
        ):
            unit = item[0]
            try:
                value = min(numbers, key=float) if numbers else value
            except ValueError:
                pass


        # laterality
        if (
            isinstance(item, tuple)
            and item[1] == "laterality"
        ):
            direction = item[0]

    # Combine number + unit into one string
    combined_value = None

    if value:

        if unit:
            combined_value = f"{value} {unit}"
        else:
            combined_value = value

    branch = {
        "effect_name": effect_name,
        "numeric_mass_effect": {
            "value": combined_value,
            "direction": direction
        }
    }

    # magnitude said directly in front of the effect
    # ("significant 1.4 cm midline shift")
    effect_index = next(
        (k for k, item in enumerate(collected)
         if isinstance(item, tuple) and item[0].lower() == str(effect_name[0]).lower()),
        None
    ) if effect_name else None

    if effect_index is not None:
        magnitude = adjacent_magnitude(collected[:effect_index])

    if magnitude:
        branch["magnitude"] = magnitude

    return branch
```

```python
def parse_association_location(tokens, start_i):
   
    i = start_i + 1
    locations = []

    while i < len(tokens):
        text, bucket = tokens[i]

        # STOP CONDITIONS
        if bucket in ASSO_LOCATION_STOP:
            break

        # ACCEPTABLE LOCATION TOKENS
        if bucket in LOCATION_ACCEPTABLE:
            locations.append(
                (text, bucket)
            )
            i += 1
            continue


        # JOINERS (sandwich_and only)
        if bucket == "sandwich_and":
            i += 1
            continue

        # UNKNOWN → STOP
        break

    return locations, i
```

## Short Sentences Function

### Other Pathology

```python
def parse_other_pathology(sent_tokens):

    negation_branches = []
    normal_state_token = None

    anatomy_tokens = []
    items = []              # one item per pathology head
    pending_modifiers = []  # saved until the next pathology head

    i = 0
    while i < len(sent_tokens):

        text, bucket = sent_tokens[i]

        # ============================
        # 1. NEGATION (reads the whole "No A, B or C" list)
        # ============================
        if bucket == "negation":
            branches, next_i = parse_pathology_negation_branch(
                sent_tokens,
                i
            )
            negation_branches.extend(branches)
            i = next_i
            continue

        # ============================
        # 2. NORMAL STATE / OUTCOME
        # ============================
        if (
            bucket in {"normal_state", "outcome"}
            or text.lower() in ("clear", "otherwise clear")
        ):
            normal_state_token = (text, bucket)
            i += 1
            continue

        # ============================
        # 3. ANATOMY
        # ============================
        if bucket in {
            "extracranial_anatomy",
            "sinus",
            "intraparenchymal",
            "sulci",
            "cistern",
            "ventricle",
            "cerebral_location",
            "bone",
            "tonsils",
        }:
            anatomy_tokens.append((text, bucket))
            i += 1
            continue

        # ============================
        # 4. PATHOLOGY (modifiers saved so far belong to this head)
        # ============================
        if bucket in {
            "pathology",
            "bone_pathology",
            "general_pathology"
        }:
            items.append({
                "pathology": [(text, bucket)],
                "modifiers": pending_modifiers,
            })
            pending_modifiers = []
            i += 1
            continue

        # "significant mass effect with ... developing hydrocephalus":
        # the grade before an effect belongs to the effect
        if bucket in {"mass_effect", "herniation_subtype"}:
            pending_modifiers = [m for m in pending_modifiers if m[1] != "magnitude"]
            i += 1
            continue

        # ============================
        # 5. MODIFIERS
        # ============================
        if bucket in {
            "magnitude",
            "acuity",
            "territorial",
            "cortical",
        }:
            pending_modifiers.append((text, bucket))
            i += 1
            continue

        # ============================
        # 6. LOCATION LINKER
        # ============================
        if bucket == "location_linker":
            loc_branch, next_i = parse_association_location(
                sent_tokens,
                i
            )

            anatomy_tokens.extend(loc_branch)
            i = max(next_i, i + 1)
            continue

        i += 1

    # modifiers after the last head ("... which may represent mild chronic
    # microangiopathic disease" is handled above; "infarct, old" here)
    if items and pending_modifiers:
        items[-1]["modifiers"] = items[-1]["modifiers"] + pending_modifiers

    # ============================
    # BUILD FINAL OBJECT
    # ============================
    result = {}

    # NEGATION OVERRIDES EVERYTHING, but a normal anatomy said before it
    # ("... maintained with no evidence of acute infarct") is kept too
    if negation_branches:
        result["negation"] = negation_branches
        if normal_state_token and anatomy_tokens:
            result["negation"] = negation_branches + [{
                "outcome": {
                    "normal_state": normal_state_token,
                    "anatomy": anatomy_tokens,
                }
            }]
        return result

    # NORMAL OUTCOME
    if normal_state_token and not items:
        result["outcome"] = {
            "normal_state": normal_state_token,
            "anatomy": anatomy_tokens if anatomy_tokens else None
        }
        return result

    # POSITIVE PATHOLOGY
    if items:
        result["outcome"] = {
            "items": items,
            "anatomy": anatomy_tokens if anatomy_tokens else None,
        }

    return result
```

### Other Path Negation

```python
def parse_pathology_negation_branch(tokens, start_i):
    """
    Negated pathology sentences use the same list reader as the bleed tree,
    so "No intracranial herniation or hydrocephalus" negates BOTH the
    herniation (an effect) and the hydrocephalus (a pathology).
    Returns (list_of_branches, next_index).
    """

    return parse_negation_branch(tokens, start_i)
```

### Sinus / Mastoid

```python
def sinus_column(text):

    text = text.lower()

    if "mastoid" in text:
        return "mastoid_air_cells"
    if "maxillary" in text:
        return "maxillary_sinus"
    if "sphenoid" in text:
        return "sphenoid_sinus"
    if "ethmoid" in text:
        return "ethmoid_sinus"
    if "frontal sinus" in text:
        return None
    return "paranasal_sinuses"


def sinus_finding_value(text, magnitude_words):

    text = text.lower()
    mild = bool(magnitude_words & {"mild", "trace", "minimal", "small", "partial", "slight", "minor", "some"})
    moderate = "moderate" in magnitude_words
    severe = bool(magnitude_words & {"severe", "complete", "near complete", "extensive", "marked"})

    if text in ("good aeration", "aeration",
                "clear", "otherwise clear", "aerated", "well aerated", "well-aerated",
                "normally aerated", "well pneumatized", "well-pneumatized"):
        return "Well aerated"

    if text in ("unremarkable", "normal", "normal limits"):
        return "Unremarkable"

    if "mucosal thickening" in text:
        # the list has no "severe" mucosal thickening: marked counts as moderate
        if moderate or magnitude_words & {"marked", "severe", "extensive"}:
            return "Moderate mucosal thickening"
        return "Mild mucosal thickening" if mild else "Mucosal thickening NS"

    if "retention cyst" in text:
        return "Retention cyst"

    # "minimal inflammatory changes within the left maxillary sinus":
    # graded like mucosal thickening
    if "inflammat" in text:
        if moderate or magnitude_words & {"marked", "severe", "extensive"}:
            return "Moderate mucosal thickening"
        return "Mild mucosal thickening" if mild else "Mucosal thickening NS"

    if "opacif" in text:
        if severe:
            return "Severe opacification"
        return "Mild opacification" if (mild or "partial" in magnitude_words) else "Opacification NS"

    if text in ("fluid", "air-fluid levels"):
        return "Fluid"

    return None


def parse_sinus_sentence(sent_tokens):
    """
    "Visualized paranasal sinuses and mastoid air cells are clear."
    "Retention cyst left maxillary sinus."
    "Mild mucosal thickening in bilateral maxillary sinuses.
     Paranasal sinuses ... are otherwise clear."

    Save every sinus named in the sentence, then give it the finding
    (the first abnormal finding wins over "clear").
    Returns {column: value}.
    """

    # one clause at a time: "Mucosal thickening in ethmoid region and fluid
    # leveling in sphenoid sinus" gives each sinus its own finding
    clauses, current = [], []
    for k, token in enumerate(sent_tokens):
        # "..., the other / remaining paranasal sinuses are clear" is its own clause
        after = " ".join(t.lower() for t, _ in sent_tokens[k + 1:k + 3])
        if token[1] in ("stop_and", "semicolon") or (
            token[1] == "comma" and re.match(r"(the )?(other|remaining|rest)\b", after)
        ):
            clauses.append(current)
            current = []
        else:
            current.append(token)
    clauses.append(current)

    parsed = []
    for clause in clauses:
        columns, findings = sinus_clause(clause)
        if columns or findings:
            parsed.append([columns, findings])

    # "X and Y are clear": a sinus with no finding takes the next finding;
    # a finding with no sinus goes to the sinus said before it
    for i, (columns, findings) in enumerate(parsed):
        if columns and not findings:
            later = next((f for c, f in parsed[i + 1:] if f), [])
            earlier = next((f for c, f in reversed(parsed[:i]) if f), [])
            parsed[i][1] = list(later or earlier)
        if findings and not columns and i > 0:
            parsed[i][0] = list(parsed[i - 1][0])

    result = {}
    for columns, findings in parsed:
        if not columns or not findings:
            continue
        abnormal = [f for f in findings if f not in ("Well aerated", "Unremarkable")]
        value = abnormal[0] if abnormal else findings[0]
        for column in columns:
            # an abnormal finding is not overwritten by "otherwise clear"
            if column in result and result[column] not in ("Well aerated", "Unremarkable"):
                continue
            result[column] = value

    return result


SINUS_BLOOD_WORDS = {"blood", "hemorrhage", "hemorrhagic", "hemosinus", "bleeding", "haemorrhage", "clot"}


def sinus_clause(clause):

    columns = []
    findings = []
    # grades anywhere in the clause ("opacification of some ... air cells")
    magnitude_words = {t.lower() for t, b in clause if b == "magnitude"}

    for text, bucket in clause:

        if bucket == "sinus":
            column = sinus_column(text)
            if column and column not in columns:
                columns.append(column)

        elif bucket in ("sinus_finding", "normal_state"):
            value = sinus_finding_value(text, magnitude_words)
            if value:
                findings.append(value)

        elif bucket == "magnitude":
            magnitude_words.add(text.lower())

        elif bucket == "negation" and not findings:
            return [], []

    # "fluid in the right sphenoid sinus ... inflammatory sinus disease":
    # the generic word repeats the specific sinus
    if len(columns) > 1 and "paranasal_sinuses" in columns:
        specific = [c for c in columns if c != "paranasal_sinuses"]
        if not any("paranasal" in t.lower() for t, b in clause if b == "sinus"):
            columns = specific

    words = " ".join(t.lower().strip("?") for t, _ in clause)

    # "blood within the sphenoid sinus", "hemorrhagic material", "hemosinus"
    if set(words.split()) & SINUS_BLOOD_WORDS:
        findings.insert(0, "Hemorrhage")

    # "fluid leveling", "air-fluid level"
    elif "fluid" in words and "Fluid" not in findings:
        findings.append("Fluid")

    return columns, findings
```

### Fractures

```python
# ============================
# FRACTURES  (three variables)
#   bony_lesion        any fracture = Yes; "No ... fracture" = No
#   fracture_location  Calvarial / Skull base / Facial (joined with " + ");
#                      "No skull fracture", "calvarium intact" = No;
#                      "No fracture" with no bone named = NR (cleaned downstream)
#   fracture_type      most severe said: Comminuted > Depressed > Displaced > Nondisplaced
# ============================

LOCATION_NAMES = {"calvarial": "Calvarial", "skullbase": "Skull base", "facial": "Facial"}

FRACTURE_BONES = {
    "facial": [
        "nasal", "orbit", "orbital", "maxilla", "maxillary", "zygom", "mandib",
        "lamina papyracea", "facial", "pterygoid", "infraorbital",
        "palat", "alveol", "vomer", "ethmoid",
    ],
    "skullbase": [
        "skull base", "skullbase", "petrous", "clivus", "condyle", "foramen magnum",
        "cranial fossa", "sphenoid", "basilar skull",
        "middle fossa", "temporal fossa", "mastoid",
        "sella", "dorsum", "clinoid", "cribriform",
    ],
    "calvarial": [
        "calvari", "skull", "parietal", "frontal bone", "occipital bone", "squamous", "temporal bone",
        "vertex", "cranial vault", "frontal", "occipital",
    ],
}

INTACT_WORDS = ("intact", "unremarkable", "normal")

# "intact / unremarkable" only says "no fracture" when it is about BONE:
# "orbital rims are intact" yes, "the orbits are unremarkable" no
INTACT_BONE_WORDS = (
    "bone", "bony", "calvari", "skull", "rim", "wall", "arch", "mandible",
    "maxilla", "condyle", "clivus", "petrous", "lamina papyracea",
)


def fracture_bone_groups(text):
    """
    Which bone groups a sentence names. 'skull base' is not the calvarium,
    and 'temporal bone' / 'petrous' is skull base.
    """

    # wording that names one bone with another bone's word in it
    rest = re.sub(r"zygomatic process of the (left |right )?frontal bone", "frontal bone", text)
    rest = re.sub(r"temporo-?\s?mandibular|temporal mandibular", " ", rest)
    # "mandibular condyle" is the jaw (facial), not the skull base condyle
    rest = re.sub(r"mandibular condyles?|condyl\w* of the (left |right )?mandible", "mandible", rest)
    rest = re.sub(r"occipital condyles?", "condyle", rest)
    # "petrous / mastoid part of the temporal bone" is skull base only
    rest = re.sub(r"\b(petrous|mastoid)\s+((portion|part|segment)\s+of\s+the\s+)?((left|right)\s+)?temporal\s+bone", r"\1", rest)
    rest = re.sub(r"\btemporal\s+bone\s*,?\s*(petrous|mastoid)", r"\1", rest)
    rest = re.sub(r"base of the (\w+ )?pterygoids?", "skull base pterygoid", rest)
    # the orbital roof is the floor of the anterior cranial fossa (skull base)
    rest = re.sub(r"orbital roofs?|roof of the (left |right )?orbits?", "skull base", rest)
    # frontal sinus: the anterior table is the face, the posterior table
    # is the skull base
    rest = re.sub(r"anterior (and posterior )?(wall|table)s? of the (left |right )?frontal sinus(es)?|frontal sinus(es)? anterior (wall|table)s?",
                  lambda m: "facial skull base" if m.group(1) else "facial", rest)
    rest = re.sub(r"posterior (wall|table)s? of the (left |right )?frontal sinus(es)?|frontal sinus(es)? posterior (wall|table)s?", "skull base", rest)

    groups = []

    for group in ("facial", "skullbase", "calvarial"):
        for word in FRACTURE_BONES[group]:
            if word in rest:
                groups.append(group)
                rest = rest.replace(word, " ")
                break

    return groups


def parse_fracture_sentence(sent_tokens):
    """
    Returns {variable: "Yes" / "No"} for one sentence.
    """

    words = [t.lower() for t, _ in sent_tokens]
    text = " ".join(words)

    has_fracture = "fractur" in text
    is_intact = any(w in words for w in INTACT_WORDS)

    # vertebral fractures are not head fractures
    if has_fracture and re.search(r"compression fractur|vertebra|\bspin(e|al)\b|\blamina\b|spinous|\bc\d\b", text) \
            and not fracture_bone_groups(text):
        return {}

    if not has_fracture and not is_intact:
        return {}

    groups = fracture_bone_groups(text)
    result = {}

    if has_fracture:

        fracture_at = next(i for i, w in enumerate(words) if "fractur" in w)

        # "No skull fracture but scalp ..." is negated; "No X, but a ...
        # fracture" is not; "Scalp hematoma but no skull fracture" is
        after_but = re.split(r"\b(?:but|however)\b", " ".join(words[:fracture_at]))[-1]
        negated = any(
            b == "negation" for _, b in sent_tokens[:fracture_at]
        ) and (
            not any(w in ("but", "however") for w in words[:fracture_at])
            or re.search(r"\b(no|not|without|negative)\b", after_but)
        )

        if negated:
            result["bony_lesion"] = "No"
            if groups:
                result["fracture_location"] = "No"
            return result

        # "This may be from old fracture", "healed fracture": the fracture
        # variables are for trauma now, so an old fracture is not one
        if re.search(r"\b(old|healed|remote|prior|previous|chronic)\s+(\w+\s+)?fractur", text):
            result["bony_lesion"] = "No"
            return result

        result["bony_lesion"] = "Yes"

        if groups:
            result["fracture_location"] = " + ".join(
                LOCATION_NAMES[g] for g in ("calvarial", "skullbase", "facial") if g in groups
            )

        # the worst type of any fracture (fracture_type_ordinal)
        if re.search(r"(?<!non-)(?<!non)comminut", text):
            result["fracture_type"] = "Comminuted"
        elif re.search(r"(?<!non-)(?<!non)depressed|with (mild |slight |minimal )?depression", text):
            result["fracture_type"] = "Depressed"
        elif re.search(r"non-?displaced|undisplaced", text):
            result["fracture_type"] = "Nondisplaced"
        elif "displaced" in text:
            result["fracture_type"] = "Displaced"
        elif re.search(r"\blinear\b", text):
            result["fracture_type"] = "Linear"

        return result

    # "The bony calvarium is intact", "Orbital rims are intact"
    if not any(w in text for w in INTACT_BONE_WORDS):
        return {}

    if groups and not any(b == "negation" for _, b in sent_tokens):
        result["bony_lesion"] = "No"
        result["fracture_location"] = "No"

    return result
```

## Detect Pathologies Function

```python
def detect_pathology_root(sent_tokens):
    for text, bucket in sent_tokens:
        if bucket in ("pathology", "bone_pathology"):
            return bucket
    for text, bucket in sent_tokens:
        if bucket == "bleed_pathology":
            return "bleed_pathology"
    return None
```

```python
def detect_other_pathology_root(sent_tokens):

    has_target_anatomy = any(bucket in {
        "extracranial_anatomy",
        "sinus",
        "intraparenchymal",
        "sulci",
        "cistern",
        "ventricle",
        "cerebral_location",
        "bone",
        "tonsils",
    } for _, bucket in sent_tokens)

    has_state = any(
        bucket in {"normal_state", "outcome"}
        or text.lower() in ("clear", "otherwise clear")
        for text, bucket in sent_tokens
    )

    if has_target_anatomy and has_state:
        return True

    return False
```

```python
def detect_relevant_sentence(sent_tokens):

    RELEVANT_BUCKETS = {

        "bleed_pathology",
        "pathology",
        "bone_pathology",

        "mass_effect",
        "herniation_subtype",

        "density",
        "acuity",

        "cerebral_location",
        "intraparenchymal",
        "ventricle",
        "cistern",
        "sulci",
        "extracranial_anatomy",
        "bone",

        "normal_state",
        "negation",

        "prominence",
        "morphology",
        "bleed_reference",
        "measurement_max",

    }

    return any(
        bucket in RELEVANT_BUCKETS
        for _, bucket in sent_tokens
    )
```

### Bleed routing helpers

```python
def bleed_type(text, sent_tokens=None, idx=None):
    """
    Which compartment a bleed_pathology token refers to.
    subdural / subarachnoid / intraparenchymal / intraventricular / epidural / generic

    A generic word ("hematoma", "collection", "bleed") takes its type from
    an anatomy qualifier directly in front of it ("parenchymal hemorrhage").
    """

    text = text.lower()

    if "subdural" in text or "sdh" in text or "extra-axial" in text or "extra- axial" in text:
        return "subdural"

    if "parafalcine" in text:
        return "subdural"

    if "subarachnoid" in text or text == "sah":
        return "subarachnoid"

    if "intraventricular" in text or text == "ivh":
        return "intraventricular"

    if text in ("iph", "ich") or "shear" in text:
        return "intraparenchymal"

    if text == "edh":
        return "epidural"

    if "parenchymal" in text or "intra-axial" in text or "intra- axial" in text:
        return "intraparenchymal"

    if "epidural" in text or "extradural" in text or "extra-dural" in text:
        return "epidural"

    if sent_tokens is not None and idx is not None:

        # look back a couple of tokens for a qualifier
        for back in range(1, 3):

            if idx - back < 0:
                break

            prev_text, prev_bucket = sent_tokens[idx - back]
            prev_text = prev_text.lower()

            if prev_text in ("subdural", "extra-axial"):
                return "subdural"

            if prev_text == "subarachnoid":
                return "subarachnoid"

            if prev_text in ("intraparenchymal", "parenchymal", "intra-axial"):
                return "intraparenchymal"

            if prev_text == "intraventricular":
                return "intraventricular"

            if prev_bucket not in ("magnitude", "acuity", "laterality",
                                   "cerebral_location", "density", None):
                break

        # "moderate hemorrhage in both lateral ventricles" -> intraventricular,
        # "hemorrhage in the basal cisterns / sulci" -> subarachnoid
        if text.split()[-1] in ("hemorrhage", "haemorrhage", "blood", "bleed", "bleeding", "clot"):
            inside = False
            for ahead in range(idx + 1, min(idx + 7, len(sent_tokens))):
                next_text, next_bucket = sent_tokens[ahead]
                # "focus of hemorrhage in the left temporal lobe": inside the
                # brain, not over it
                if next_bucket == "location_linker":
                    inside = next_text.lower().strip() in ("in", "in the", "within", "within the")
                if next_bucket == "cerebral_location" and inside and "lobe" in next_text.lower():
                    return "intraparenchymal"
                if next_bucket == "ventricle":
                    return "intraventricular"
                if next_bucket in ("cistern", "sulci"):
                    return "subarachnoid"
                if next_bucket == "intraparenchymal":
                    return "intraparenchymal"
                if next_bucket not in ("location_linker", "laterality", "subregional_anatomy",
                                       "noticing", "uncertainty_linker", None) \
                        and next_text.lower() != "also":
                    break

    return "generic"


def bleed_is_negated(sent_tokens, idx):
    """
    A bleed token is negated when a negation word appears before it
    in the same sentence with no 'but' in between
    ("No acute intracranial hemorrhage", "There is no evidence of
    parenchymal hemorrhage").
    """

    for back in range(idx - 1, -1, -1):

        text, bucket = sent_tokens[back]

        if text.lower() in ("but", "however", "although"):
            return False

        if bucket == "negation":
            return True

    return False


def bleed_mentions(sent_tokens):
    """
    [(idx, text, type, negated), ...] for every bleed token in a sentence
    """

    mentions = []

    for idx, (text, bucket) in enumerate(sent_tokens):

        if bucket != "bleed_pathology":
            continue

        mentions.append(
            (
                idx,
                text,
                bleed_type(text, sent_tokens, idx),
                bleed_is_negated(sent_tokens, idx)
            )
        )

    return mentions


ANAPHORA_WORDS = {"lesion", "collection", "hematoma", "abnormality"}


def resolve_anaphora(sent_tokens):
    """
    "The lesion ..." / "This collection ..." after an SDH has been
    described refers back to the SDH, not to a new pathology.
    """

    resolved = list(sent_tokens)

    for idx in range(1, len(resolved)):

        text, bucket = resolved[idx]
        prev_text = resolved[idx - 1][0].lower()

        if (
            text.lower() in ANAPHORA_WORDS
            and bucket in ("pathology", "bleed_pathology")
            and prev_text in ("the", "this")
        ):
            resolved[idx] = (text, "bleed_reference")

    return resolved


def sentence_has(sent_tokens, buckets):
    return any(
        bucket in buckets
        for _, bucket in sent_tokens
    )
```

## Orchestrator Function

```python
def bleed_side(sent_tokens, idx):
    """
    Side of a bleed word: the nearest laterality before it in the sentence
    ("right extradural hemorrhage"), else the first one after it
    ("epidural hematoma over the left convexity"). L / R / B or None.
    """

    for k in range(idx - 1, -1, -1):
        text, bucket = sent_tokens[k]
        if bucket == "laterality" and side_of(text):
            return side_of(text)

    for k in range(idx + 1, len(sent_tokens)):
        text, bucket = sent_tokens[k]
        if bucket == "bleed_pathology":
            break
        if bucket == "laterality" and side_of(text):
            return side_of(text)

    return None


PRONOUN_OPENERS = (
    "this is", "this has", "this contains", "this demonstrates", "this appears",
    "it is", "it has", "it contains", "it demonstrates", "it appears",
    "these are", "these have", "these contain", "they are", "they have",
    "these measure", "they measure",
)


def describes_last_bleed(sent_tokens):
    """
    "This is mostly hyperdense with areas of lower density.",
    "This has mixed high and low density.": a pronoun sentence that only
    describes density / age / inner look is about the bleed just named.
    """

    if not sent_tokens:
        return False

    opener = sent_tokens[0][0].lower()
    if len(sent_tokens) > 1 and opener in ("this", "it", "these", "they"):
        opener = f"{opener} {sent_tokens[1][0].lower()}"

    # "These measure approximately 0.7 and 0.8 cm ... right and left ...
    # respectively": the sizes of the bleeds just named
    measures = opener in ("these measure", "they measure") and sentence_has(sent_tokens, {"numerical"})

    return (
        opener in PRONOUN_OPENERS
        and (measures or sentence_has(sent_tokens, {"density", "acuity", "morphology"}))
        and not sentence_has(sent_tokens, {"pathology", "bone_pathology", "bleed_pathology"})
    )


NORMAL_SIZE_WORDS = (
    "age appropriate", "age-appropriate", "normal", "normal in size",
    "within normal limits", "unremarkable", "normal size", "appropriate for age",
)


def normal_ventricles_sulci(sent_tokens):
    """
    "The ventricles and sulci are age appropriate", "The lateral ventricles
    are normal in size": nothing is effaced. Returns negated variables.
    """

    words = " ".join(t.lower() for t, _ in sent_tokens)

    if not any(w in words for w in NORMAL_SIZE_WORDS):
        return []

    if sentence_has(sent_tokens, {"negation", "mass_effect", "herniation_subtype",
                                  "prominence", "pathology", "bleed_pathology", "spine"}):
        return []

    # "normal vertebral body height": "body" is not the ventricle here
    if re.search(r"vertebra|\bdisc\b|spin", words):
        return []

    variables = []
    ventricles = [t.lower() for t, b in sent_tokens if b == "ventricle"]

    if any("sulc" in t.lower() for t, b in sent_tokens if b == "sulci"):
        variables.append("N_effacement_sulci")

    if any("lateral" in v for v in ventricles):
        variables.append("N_effacement_lateral_ventricle")

    elif ventricles:
        variables += ["N_effacement_lateral_ventricle", "N_effacement_third_ventricle",
                      "N_effacement_ventricular_system"]

    return variables


def orchestrator(sentence_lists):
    """
    Route every sentence:

      sinus / mastoid sentence           -> parse_sinus_sentence
      positive subdural (or generic) bleed -> THE one SDH tree (all sides,
                                            all wordings: collection, SDH,
                                            hematoma, extra-axial ...)
      positive SAH / IPH / IVH           -> {"hemorrhage": ...}
      negated bleed / other pathology    -> parse_other_pathology
      anything else relevant             -> secondary sentence of the SDH tree
                                            (effects seen before the SDH is
                                            described are saved in `pending`)
    """

    bleed_trees = []
    sdh_tree = None
    pending = []
    last_bleed_type = None
    last_fracture = None

    def add_secondary(sent_tokens):

        if sdh_tree is not None:
            build_bleed_tree(
                sent_tokens,
                is_bleed_sentence=False,
                existing_tree=sdh_tree
            )
            return

        # no SDH yet: keep the effects until it shows up
        holder = {"pathology": None, "children": []}

        build_bleed_tree(
            sent_tokens,
            is_bleed_sentence=False,
            existing_tree=holder
        )

        pending.extend(
            child for child in holder["children"]
            if isinstance(child, dict) and "association" in child
        )

    for sent_tokens in sentence_lists:

        sent_tokens = resolve_anaphora(sent_tokens)

        # "The patient has a history of a left orbital floor fracture",
        # "Previously the patient had a right frontal SDH": the past, not this scan
        if sentence_has(sent_tokens, {"history"}):
            continue
        opening = " ".join(t.lower() for t, _ in sent_tokens[:5])
        if re.match(r"(previously|in the past|prior to this),?\s+(the\s+)?patient\s+(had|has)\b", opening):
            continue

        # cervical spine / prevertebral findings are not head findings
        if (
            sentence_has(sent_tokens, {"spine"})
            and not sentence_has(sent_tokens, {"bleed_pathology"})
        ):
            continue

        # ============================
        # FRACTURES (own variables, any sentence)
        # ============================
        fracture = parse_fracture_sentence(sent_tokens)

        # "There is a fracture of the squamous temporal bone. This extends
        # into the petrous bone." -> the second bone is the same fracture
        sentence_text = " ".join(t.lower() for t, _ in sent_tokens)
        if (
            not fracture and last_fracture is not None
            and re.match(r"(this|it|the fracture)\s+(also\s+)?(extends|extending|continues)", sentence_text)
        ):
            extra = [LOCATION_NAMES[g] for g in fracture_bone_groups(sentence_text)]
            if extra:
                have = set(filter(None, last_fracture.get("fracture_location", "").split(" + ")))
                last_fracture["fracture_location"] = " + ".join(sorted(have | set(extra)))

        if fracture:
            bleed_trees.append({"fracture": fracture})
            if fracture.get("bony_lesion") == "Yes":
                last_fracture = fracture

        # ============================
        # CASE 0: SINUS / MASTOID
        # ============================
        if sentence_has(sent_tokens, {"sinus"}):

            sinus = parse_sinus_sentence(sent_tokens)

            if sinus:
                bleed_trees.append({"sinus": sinus})

            remaining = [
                (text, bucket) for text, bucket in sent_tokens
                if bucket not in ("sinus", "sinus_finding")
            ]

            if not detect_relevant_sentence(remaining):
                continue

        # ============================
        # CASE 1: BLEEDS
        # ============================
        mentions = bleed_mentions(sent_tokens)
        positive = [m for m in mentions if not m[3]]

        # "scalp soft tissue density ... likely relates to hematoma",
        # "hemorrhage with fluid levels in the maxillary sinuses",
        # "small left intraorbital hematoma": a generic hematoma / hemorrhage
        # said about the scalp, a sinus or the orbit is not the SDH
        sentence_words = " ".join(t.lower() for t, _ in sent_tokens)
        if (
            re.search(r"\b(scalp|soft tissue|subgaleal|occipitalis|temporalis|periorbital|preseptal|"
                      r"orbit\w*|intraorbital|extraconal|intraconal|retrobulbar|globe|nasal|nasopharyn\w*|"
                      r"ethmoids?|hemosinus|mastoid\w*)\b", sentence_words)
            or sentence_has(sent_tokens, {"sinus"})
        ):
            positive = [m for m in positive if m[2] != "generic"]

        # "at the site of presumed prior hematoma", "old hematoma": not now
        positive = [
            m for m in positive
            if not (m[2] == "generic" and re.search(
                r"\b(prior|previous|old|remote|resolved)\s+(\w+\s+)?$",
                " ".join(t.lower() for t, _ in sent_tokens[max(0, m[0] - 3):m[0]]) + " "))
        ]

        sdh_positive = False
        other_types = []
        other_sides = {}
        other_index = {}

        for idx, text, btype, negated in positive:

            if btype == "generic":
                btype = last_bleed_type or "subdural"

            # uncertain findings ("possible", "suspicious for", "cannot be
            # excluded") are recorded as present, like every other bleed

            if btype == "subdural":
                sdh_positive = True
            elif btype not in other_types:
                other_types.append(btype)
                other_sides[btype] = bleed_side(sent_tokens, idx)
                other_index[btype] = idx

            last_bleed_type = btype

        # "small focus of hemorrhage ... concerning for contusion": the bleed
        # is the contusion, recorded as parenchymal_contusion_present
        if "intraparenchymal" in other_types and re.search(r"contusion", sentence_words):
            other_types.remove("intraparenchymal")

        for btype in other_types:
            bleed_trees.append(
                {"hemorrhage": {
                    "type": btype,
                    "grade": None,      # subarachnoid hemorrhage is Yes / No only
                    "side": other_sides.get(btype),
                }}
            )

        if (
            sdh_positive
            and sent_tokens[0][0].lower() in ("these", "they", "both")
            and not sentence_has(sent_tokens, {"laterality"})
        ):
            sent_tokens = [("bilateral", "laterality")] + list(sent_tokens)

        if sdh_positive:

            if sdh_tree is None:

                sdh_tree = {
                    "pathology": "subdural",
                    "children": list(pending)
                }
                pending = []
                bleed_trees.append(sdh_tree)

            build_bleed_tree(
                sent_tokens,
                is_bleed_sentence=True,
                existing_tree=sdh_tree
            )

            # "left SDH with bilateral SAH and right frontal contusion": other
            # findings named in an SDH sentence are read too
            if sentence_has(sent_tokens, {"pathology"}):
                other_result = parse_other_pathology(sent_tokens)
                if other_result:
                    bleed_trees.append(other_result)
            continue

        if other_types:

            # "There is a large midline shift" style effects in a sentence
            # about another bleed still describe the brain
            if sentence_has(sent_tokens, EFFECT_FAMILY):
                add_secondary(sent_tokens)

            # "intraparenchymal bleed ... with surrounding edema"
            if sentence_has(sent_tokens, {"pathology"}):
                other_result = parse_other_pathology(sent_tokens)
                if other_result:
                    bleed_trees.append(other_result)
            continue

        refers_back = (
            sentence_has(sent_tokens, {"bleed_reference"})
            or describes_last_bleed(sent_tokens)
        )

        # "the hematoma ..." -> still about the SDH
        if refers_back and last_bleed_type not in (None, "subdural"):
            continue

        # "These are of mixed attenuation" after two collections: both sides
        if (
            describes_last_bleed(sent_tokens)
            and sent_tokens[0][0].lower() in ("these", "they")
            and not sentence_has(sent_tokens, {"laterality"})
        ):
            sent_tokens = [("bilateral", "laterality")] + list(sent_tokens)

        # "The hematoma underlies the right frontal parietal convexity
        # measuring up to 9 mm": where and how big, read like a bleed
        # sentence; the effects in it are read over the whole sentence
        if (
            refers_back
            and sdh_tree is not None
            and not sentence_has(sent_tokens, {"pathology", "bone_pathology"})
        ):
            holder = {"pathology": None, "children": []}

            build_bleed_tree(sent_tokens, is_bleed_sentence=True, existing_tree=holder)

            # mark it: "This hematoma ..." continues the bleed of the last sentence
            for child in holder["children"]:
                if isinstance(child, dict) and "sentence_start" in child:
                    child["anaphor"] = True

            sdh_tree["children"].extend(
                child for child in holder["children"]
                if not (isinstance(child, dict) and "association" in child)
            )

            assoc, _ = parse_association(sent_tokens, 0)
            assoc = [a for a in assoc if not any(k in a for k in ("density", "acuity", "morphology"))]

            if assoc:
                sdh_tree["children"].append({"association": assoc})

            continue

        normal_size = normal_ventricles_sulci(sent_tokens)
        if normal_size:
            bleed_trees.append({"mapped": normal_size})
            continue

        # ============================
        # CASE 2: OTHER PATHOLOGY / NEGATED BLEED / NORMAL ANATOMY
        # ============================
        root_bucket = detect_pathology_root(sent_tokens)

        if (
            root_bucket in ("pathology", "bone_pathology")
            or mentions
            or detect_other_pathology_root(sent_tokens)
        ):

            other_result = parse_other_pathology(sent_tokens)

            if other_result:
                bleed_trees.append(other_result)

            # mixed sentence: "atrophy with ex vacuo dilatation of the
            # ventricles", "fullness ... concerning for early trapping"
            if sentence_has(sent_tokens, EFFECT_FAMILY | {"prominence"}):
                add_secondary(sent_tokens)

            continue

        # ============================
        # NOT RELEVANT
        # ============================
        # "This measures 11.0 x 2.5 cm" points back to the SDH
        if not detect_relevant_sentence(sent_tokens) and not secondary_thickness(sent_tokens):
            continue

        # ============================
        # CASE 3: NO PATHOLOGY -> secondary sentence of the SDH
        # ============================
        add_secondary(sent_tokens)

    if sdh_tree is None and pending:
        bleed_trees.append(
            {"pathology": None, "children": pending}
        )

    return bleed_trees
```

```python
# ============================
# FINAL OUTPUT VARIABLES
#   The column names, order and value wording of the study dataset.
#   Left = column in SDH_results.csv, right = the variable the pipeline
#   fills. Edit names here; the extraction code does not need to change.
# ============================

FINAL_COLUMNS = [
    # names, spelling and order exactly as on the extraction sheet

    # Overall
    ("laterality", "laterality"),

    # Left hematoma
    ("L_description", "L_description"),
    ("L_acuity", "L_acuity"),
    ("L_mixed_density", "L_mixed_density"),
    ("L_internal_appearence", "L_morphology"),
    ("L_hyperdense", "L_hyper_dense"),
    ("L_hypodense", "L_hypo_dense"),
    ("L_isodense", "L_iso_dense"),
    ("L_density_dominance", "L_density_dominance"),
    ("L_thickness_mm", "L_thickness_mm"),
    ("L_thickness_cm", "L_thickness_cm"),
    ("L_convexity_hematoma", "L_convexity_hematoma"),
    ("L_holohemispheric _ hematoma", "L_holohemispheric_hematoma"),
    ("L_vertex", "L_vertex"),
    ("L_frontal_lobe", "L_frontal_lobe"),
    ("L_parietal_lobe", "L_parietal_lobe"),
    ("L_temporal_lobe", "L_temporal_lobe"),
    ("L_occipital_lobe", "L_occipital_lobe"),
    ("L_middle_fossa", "L_middle_fossa"),

    # Right hematoma
    ("R_description", "R_description"),
    ("R_acuity", "R_acuity"),
    ("R_mixed_density", "R_mixed_density"),
    ("R_internal_appearence", "R_morphology"),
    ("R_hyperdense", "R_hyper_dense"),
    ("R_hypodense", "R_hypo_dense"),
    ("R_isodense", "R_iso_dense"),
    ("R_density_dominance", "R_density_dominance"),
    ("R_thickness_mm", "R_thickness_mm"),
    ("R_thickness_cm", "R_thickness_cm"),
    ("R_convexity_hematoma", "R_convexity_hematoma"),
    ("R_holohemispheric _ hematoma", "R_holohemispheric_hematoma"),
    ("R_vertex", "R_vertex"),
    ("R_frontal_lobe", "R_frontal_lobe"),
    ("R_parietal_lobe", "R_parietal_lobe"),
    ("R_temporal_lobe", "R_temporal_lobe"),
    ("R_occipital_lobe", "R_occipital_lobe"),
    ("R_middle_fossa", "R_middle_fossa"),

    # Falx / tentorium
    ("parafalcine_description", "parafalcine_description"),
    ("parafalcine_hematoma", "parafalcine_hematoma"),
    ("parafalcine_location", "parafalcine_location"),
    ("parafalcine_thickness_mm", "parafalcine_thickness_mm"),
    ("parafalcine_thickness_cm", "parafalcine_thickness_cm"),
    ("parafalcine_density", "parafalcine_density"),
    ("parafalcine_acuity", "parafalcine_acuity"),
    ("tentorium_description", "tentorium_description"),
    ("tentorium_hematoma", "tentorium_hematoma"),
    ("tentorium_thickness_mm", "tentorium_thickness_mm"),
    ("tentorium_thickness_cm", "tentorium_thickness_cm"),
    ("tentorium_density", "tentorium_density"),
    ("tentorium_acuity", "tentorium_acuity"),

    # Mass effect
    ("mass_effect_present", "mass_effect_present"),
    ("midline_shift_present", "midline_shift_present"),
    ("midline_shift_mm", "midline_shift_mm"),
    ("midline_shift_cm", "midline_shift_cm"),
    ("midline_shift_direction", "midline_shift_direction"),

    # Effacement
    ("effacement_sulci", "effacement_sulci"),
    ("effacement_sulci_laterality", "effacement_sulci_laterality"),
    ("effacement_lateral_ventricle", "effacement_lateral_ventricle"),
    ("l_ventricle_laterality", "l_ventricle_laterality"),
    ("effacement_third_ventricle", "effacement_third_ventricle"),
    ("effacement_ventricular_system", "effacement_ventricular_system"),
    ("l_ventricular_prominence", "ventricular_prominence"),
    ("l_v_prominence_laterality", "l_v_prominence_laterality"),
    ("ex_vacuo_ventriculomegaly", "ex_vacuo_dilatation"),

    # Herniation
    ("herniation_subfalcine", "herniation_subfalcine"),
    ("herniation_uncal", "herniation_uncal"),
    ("herniation_transtentorial", "herniation_transtentorial"),
    ("herniation_tonsillar", "herniation_tonsillar"),

    # Other findings
    ("encephalomalacia", "encephalomalacia"),
    ("hydrocephalus", "hydrocephalus"),
    ("microangiopathy", "microangiopathy"),
    ("cerebral_atrophy", "cerebral_atrophy"),
    ("white_dark_differentiation", "white_dark_differentiation"),
    ("mass", "mass"),
    ("territorial_ infarct", "territorial_infarct"),
    ("lacune_infarct", "lacunar_infarct"),
    ("unspecified__infarct", "unspecified_infarct"),

    # Fractures / extracranial
    ("soft_tissue_lesion", "soft_tissue_lesion"),
    ("scalp_hematoma", "scalp_hematoma"),
    ("orbital_lesion", "orbital_lesion"),
    ("bony_lesion", "bony_lesion"),
    ("fracture_location", "fracture_location"),
    ("fracture_type_ordinal", "fracture_type"),

    # Other bleeds
    ("parenchymal_contusion_present", "contusion"),
    ("parenchymal_edema", "edema"),
    ("parenchymal_hemorrhage", "intraparenchymal_hemorrhage"),
    ("subarachnoid_hemorrhage", "subarachnoid_hemorrhage"),
    ("intraventricular_hemorrhage", "intraventricular_hemorrhage"),

    # Sinuses / mastoids
    ("paranal_sinuses", "paranasal_sinuses"),
    ("mastoid_air_cells", "mastoid_air_cells"),
    ("sphenoid_sinus", "sphenoid_sinus"),
    ("ethmoid_sinus", "ethmoid_sinus"),
    ("maxillary_sinus", "maxillary_sinus"),

    # Not on the sheet (delete if not wanted)
    ("epidural_hemorrhage", "epidural_hemorrhage"),
    ("aneurysm", "aneurysm"),
]

# value wording of the final dataset (applies to every column)
FINAL_VALUE_WORDING = {
    "Acute on Chronic": "Acute on chronic",
    "Subacute on Chronic": "Subacute on chronic",
    "Age Appropriate": "Age appropriate",
    "Dependent": "Dependant",
    "Minimal": "Small",
    "Diffuse": "Yes NS",
    "Moderate/ Severe": "Severe",      # "moderate to severe" -> the higher grade
    "Mild/ Moderate": "Moderate",
    "Nondisplaced": "Non-displaced",
    "Preserved": "Yes",                # gray-white differentiation present
    "Loss": "No",
    "Mild Opacification": "Mild opacification",
    "Well aerated": "Clear",            # sinuses / mastoids: one wording
    "Unremarkable": "Clear",
}

# a plain "Yes" is "Yes NS" in these graded columns
YES_NS_COLUMNS = {
    "mass_effect_present", "midline_shift_present", "effacement_sulci",
    "effacement_lateral_ventricle", "effacement_ventricular_system",
    "l_ventricular_prominence", "ex_vacuo_ventriculomegaly", "effacement_third_ventricle",
    "herniation_subfalcine", "herniation_uncal", "herniation_transtentorial",
    "herniation_tonsillar", "hydrocephalus",
    "microangiopathy", "cerebral_atrophy", "parafalcine_hematoma", "tentorium_hematoma",
}


def final_fracture_location(value):
    # "Calvarial + Skull base" -> "Calvarial & Skull base" in the order of the list
    # the final list has no "No": no fracture leaves the location NR
    if value in ("NR", "No", None):
        return "NR"
    parts = set(value.split(" + "))
    order = ["Calvarial", "Facial", "Skull base"]
    return " & ".join(p for p in order if p in parts)


SIDE_WORDS = {"L": "left", "R": "right", "B": "both sides"}


def review_flag(variables):
    """
    Reports a person should read: the code keeps to what the report
    says, so these are flagged instead of guessed.
    """

    reasons = []

    sdh_sides = {
        "Left": {"L"}, "Right": {"R"}, "Bilateral": {"L", "R"}
    }.get(variables.get("laterality"), set())

    # "right extradural hemorrhage" ... "large right subdural hemorrhage":
    # two bleeds on one side, or one bleed called two names
    for side in variables.get("_epidural_sides", []):

        epidural_sides = {"L", "R"} if side == "B" else ({side} if side else set())

        if not sdh_sides:
            continue

        if not epidural_sides:
            reasons.append("epidural and subdural both reported (epidural side not given)")
        elif epidural_sides & sdh_sides:
            reasons.append(f"epidural and subdural both on the {SIDE_WORDS[side]}")

    return "; ".join(dict.fromkeys(reasons))


def final_output(variables):
    """
    Pipeline variables -> one row of the final dataset (ordered dict).
    """

    row = {}

    # infarct types without their own column fold into the nearest one
    variables = dict(variables)
    if variables.get("unspecified_infarct", "NR") == "NR":
        variables["unspecified_infarct"] = variables.get("cortical_infarct", "NR")
    if variables.get("territorial_infarct", "NR") == "NR":
        variables["territorial_infarct"] = variables.get("territorial_ischemic", "NR")

    for column, source in FINAL_COLUMNS:

        value = variables.get(source, "NR")

        if column == "fracture_location":
            value = final_fracture_location(value)

        if isinstance(value, str):
            value = FINAL_VALUE_WORDING.get(value, value)

            if value == "Yes" and column in YES_NS_COLUMNS:
                value = "Yes NS"

        row[column] = value

    row["review_flag"] = review_flag(variables)

    return row
```

```python
# ============================
# SETTINGS - the only cell you need to edit
#   Put the CSV in the same folder as this notebook and write its name here.
# ============================

INPUT_CSV = "FINAL_BLANK_SHEET - Copy.csv"   # your sheet of reports
OUTPUT_CSV = "SDH_results.csv"                # what the NLP extracted
VALIDATION_PREFIX = "SDH_validation"          # files with the comparison

ID_COLUMNS = (0, 1)        # columns A and B: report numbers (0 = first column)
REPORT_COLUMN = 2          # column C: the report text

# compare the NLP with the human values in the same CSV (columns D onwards)?
RUN_VALIDATION = True
```

```python
# ============================
# RUN ALL REPORTS
#   Reads INPUT_CSV (settings cell): report numbers in ID_COLUMNS, the report
#   in REPORT_COLUMN. Rows with no report text are skipped.
#   Output OUTPUT_CSV: the two report numbers, the report, the tree (before
#   variable mapping), every variable in FINAL_COLUMNS, review_flag, error.
# ============================

import contextlib
import io
import pprint
import pandas as pd


def tree_text(bleed_trees):
    """
    The tree before variable mapping, as readable text for one cell.
    """
    return pprint.pformat(bleed_trees, width=110, sort_dicts=False)


def read_sheet(path):
    """
    A CSV from Excel: try UTF-8 first, then the Windows encoding.
    Returns (table, has_header).
    """

    for encoding in ("utf-8-sig", "cp1252", "latin-1"):
        try:
            table = pd.read_csv(path, header=None, dtype=str, keep_default_na=False, encoding=encoding)
            break
        except UnicodeDecodeError:
            continue

    # a header row: its report-number cells are not numbers ("#", "report_num")
    first = [str(table.iloc[0, c]).strip() for c in ID_COLUMNS] if len(table) else []
    has_header = bool(first) and not any(v.replace(".", "").isdigit() for v in first)

    if has_header:
        table.columns = [str(c).strip() for c in table.iloc[0]]
        table = table.iloc[1:].reset_index(drop=True)

    return table, has_header


def id_names(table, has_header):
    if has_header:
        return [table.columns[c] or f"id_{c + 1}" for c in ID_COLUMNS]
    return ["report_num", "report_index"]


def run_csv(input_path=None, output_path=None):

    input_path = input_path or INPUT_CSV
    output_path = output_path or OUTPUT_CSV

    table, has_header = read_sheet(input_path)
    names = id_names(table, has_header)

    rows = []
    skipped = 0

    for _, report in table.iterrows():

        text = str(report.iloc[REPORT_COLUMN])

        if not text.strip():
            skipped += 1
            continue

        row = {name: str(report.iloc[c]).strip() for name, c in zip(names, ID_COLUMNS)}
        row["report"] = text

        try:
            # the parsers print debugging output - hide it for batch runs
            with contextlib.redirect_stdout(io.StringIO()):
                sentence_lists = tokenize_report(text)
                bleed_trees = [orchestrator(sentence_lists)]
                variables = master_variable_mapping_function(bleed_trees)

            row["tree"] = tree_text(bleed_trees)
            row.update(final_output(variables))
            row["error"] = ""

        except Exception as error:
            # keep going: this report gets NR everywhere and the error is recorded
            row["tree"] = ""
            row.update(final_output({}))
            row["error"] = f"{type(error).__name__}: {error}"

        rows.append(row)

    results = pd.DataFrame(rows)

    # utf-8-sig so Excel opens it with the right characters
    results.to_csv(output_path, index=False, encoding="utf-8-sig")

    failed = (results["error"] != "").sum() if len(results) else 0
    print(f"{len(results)} reports -> {output_path}  ({failed} with errors, {skipped} rows without report text skipped)")

    return results
```

## Validation metrics

```python
# ============================
# VALIDATION: NLP vs the human values in the same CSV
#
# For every variable and every report the two values are compared after
# tidying the wording (case, spaces, "Yes" = "Yes NS", blank = NR, ...).
# Numbers (mm / cm) agree when they are within NUMBER_TOLERANCE.
#
#   a "finding"        = a value that is not NR
#   TP  true positive  = the human has a finding and the NLP has the same one
#   FN  false negative = the human has a finding and the NLP has NR or another value
#   FP  false positive = the NLP has a finding and the human has NR or another value
#   TN  true negative  = both NR
#   (a wrong value is counted as both an FN and an FP)
#
#   accuracy    = reports where the two agree / all reports
#   sensitivity = TP / (TP + FN)        (recall)
#   PPV         = TP / (TP + FP)        (precision)
#   F1          = 2 * TP / (2 * TP + FN + FP)   (= harmonic mean of PPV and sensitivity)
#   specificity = TN / (TN + NLP findings where the human has NR)
#   kappa       = Cohen's kappa over all values (agreement beyond chance)
#   95% CI      = Wilson interval
#
# Writes
#   <prefix>_by_variable.csv   one row per variable
#   <prefix>_summary.csv       all variables together
#   <prefix>_mismatches.csv    every report / variable where the two differ
#   <prefix>_confusion.csv     human value x NLP value counts per variable
# ============================

import math
import re

NUMBER_TOLERANCE = {"mm": 0.5, "cm": 0.05}

# sheet column (after tidying the name) -> NLP column
# (the NLP columns already use the sheet's names; these catch other spellings)
SHEET_COLUMN_ALIASES = {
    "l_internal_appearance": "L_internal_appearence",
    "r_internal_appearance": "R_internal_appearence",
    "effacement_lateral_ventricle_laterality": "l_ventricle_laterality",
    "ventricular_prominence": "l_ventricular_prominence",
    "calvarial_fracture_type": "fracture_type_ordinal",
    "fracture_type": "fracture_type_ordinal",
    "paranasal_sinus": "paranal_sinuses",
    "paranasal_sinuses": "paranal_sinuses",
    "ex_vacio_ventriculomegaly": "ex_vacuo_ventriculomegaly",
}

SAME_VALUE = {
    "yes": "yes ns",
    "well aerated": "clear",
    "unremarkable": "clear",
    "muscal thickening ns": "mucosal thickening ns",
    "dependent": "dependant",
    "acute/subacute": "acute on subacute",
}

BLANK_VALUES = {"", "nr", "nan", "none", "n/a", "na"}


def tidy_name(name):
    return re.sub(r"[\s_]+", "_", str(name).strip().lower())


def tidy_value(column, value):

    value = re.sub(r"\s+", " ", str(value).strip().lower())

    if value in BLANK_VALUES:
        return "nr"

    if column.endswith(("_mm", "_cm")):
        try:
            return float(value)
        except ValueError:
            return value

    # "Calvarial & Skull base" in any order
    if column == "fracture_location" and "&" in value:
        return " & ".join(sorted(p.strip() for p in value.split("&")))

    return SAME_VALUE.get(value, value)


def same(column, h, c):
    if isinstance(h, float) and isinstance(c, float):
        return abs(h - c) <= NUMBER_TOLERANCE["mm" if column.endswith("_mm") else "cm"] + 1e-9
    return h == c


def wilson(k, n, z=1.96):
    if n == 0:
        return (None, None)
    p = k / n
    centre = (p + z * z / (2 * n)) / (1 + z * z / n)
    half = z * math.sqrt(p * (1 - p) / n + z * z / (4 * n * n)) / (1 + z * z / n)
    return (round(centre - half, 3), round(centre + half, 3))


def ratio(a, b):
    return round(a / b, 3) if b else None


def cohen_kappa(pairs):
    n = len(pairs)
    if n == 0:
        return None
    observed = sum(1 for h, c in pairs if h == c) / n
    labels = {h for h, _ in pairs} | {c for _, c in pairs}
    expected = sum(
        (sum(1 for h, _ in pairs if h == lab) / n) * (sum(1 for _, c in pairs if c == lab) / n)
        for lab in labels
    )
    if expected >= 1:
        return None          # only one value seen: kappa is undefined
    return round((observed - expected) / (1 - expected), 3)


def validate(input_path=None, results=None, prefix=None):

    input_path = input_path or INPUT_CSV
    prefix = prefix or VALIDATION_PREFIX

    table, has_header = read_sheet(input_path)

    if not has_header:
        print("No header row in the CSV, so there are no human columns to compare with.")
        return None

    names = id_names(table, has_header)
    table = table[table.iloc[:, REPORT_COLUMN].astype(str).str.strip() != ""].reset_index(drop=True)

    if results is None:
        results = pd.read_csv(OUTPUT_CSV, dtype=str, keep_default_na=False)

    # match the human columns to the NLP columns by name
    nlp_columns = {tidy_name(c): c for c, _ in FINAL_COLUMNS}
    used = set(ID_COLUMNS) | {REPORT_COLUMN}
    pairs = {}
    not_in_nlp = []

    for position, column in enumerate(table.columns):
        if position in used or not str(column).strip() or str(column).startswith("Unnamed"):
            continue
        key = tidy_name(column)
        nlp = SHEET_COLUMN_ALIASES.get(key) or nlp_columns.get(key)
        if nlp:
            pairs[nlp] = column
        else:
            not_in_nlp.append(str(column).strip())

    not_in_sheet = [c for c, _ in FINAL_COLUMNS if c not in pairs]

    # rows: same report numbers in the same order
    key_h = table[names[0]].astype(str).str.strip() + "|" + table[names[1]].astype(str).str.strip()
    key_c = results[names[0]].astype(str).str.strip() + "|" + results[names[1]].astype(str).str.strip()
    nlp_rows = dict(zip(key_c, results.to_dict("records")))

    by_variable, mismatches, confusion = [], [], {}
    total = dict(n=0, agree=0, TP=0, FN=0, FP=0, TN=0, FP_on_NR=0)

    for nlp_column, sheet_column in pairs.items():

        values = []
        for k, (_, row) in zip(key_h, table.iterrows()):
            if k not in nlp_rows:
                continue
            h = tidy_value(nlp_column, row[sheet_column])
            c = tidy_value(nlp_column, nlp_rows[k].get(nlp_column, "NR"))
            values.append((k, h, c, row[sheet_column], nlp_rows[k].get(nlp_column, "NR")))

        s = dict(n=len(values), agree=0, TP=0, FN=0, FP=0, TN=0, FP_on_NR=0)
        abs_errors = []

        for k, h, c, h_raw, c_raw in values:

            agree = same(nlp_column, h, c)
            s["agree"] += agree

            if h != "nr" and agree:
                s["TP"] += 1
            if h != "nr" and not agree:
                s["FN"] += 1
            if c != "nr" and not agree:
                s["FP"] += 1
            if h == "nr" and c == "nr":
                s["TN"] += 1
            if h == "nr" and c != "nr":
                s["FP_on_NR"] += 1

            if isinstance(h, float) and isinstance(c, float):
                abs_errors.append(abs(h - c))

            if not agree:
                mismatches.append({names[0]: k.split("|")[0], names[1]: k.split("|")[1],
                                   "variable": nlp_column, "human": h_raw, "nlp": c_raw})

            label_h = "NR" if h == "nr" else str(h_raw).strip()
            label_c = "NR" if c == "nr" else str(c_raw).strip()
            confusion[(nlp_column, label_h, label_c)] = confusion.get((nlp_column, label_h, label_c), 0) + 1

        for key in total:
            total[key] += s[key]

        sens, ppv = ratio(s["TP"], s["TP"] + s["FN"]), ratio(s["TP"], s["TP"] + s["FP"])
        numeric = nlp_column.endswith(("_mm", "_cm"))

        by_variable.append({
            "variable": nlp_column,
            "reports": s["n"],
            "human_findings": s["TP"] + s["FN"],
            "nlp_findings": s["TP"] + s["FP"],
            "TP": s["TP"], "FN": s["FN"], "FP": s["FP"], "TN": s["TN"],
            "accuracy": ratio(s["agree"], s["n"]),
            "accuracy_95CI": wilson(s["agree"], s["n"]),
            "sensitivity": sens,
            "sensitivity_95CI": wilson(s["TP"], s["TP"] + s["FN"]),
            "PPV": ppv,
            "PPV_95CI": wilson(s["TP"], s["TP"] + s["FP"]),
            "F1": ratio(2 * s["TP"], 2 * s["TP"] + s["FN"] + s["FP"]),
            "specificity": ratio(s["TN"], s["TN"] + s["FP_on_NR"]),
            "kappa": None if numeric else cohen_kappa([(h, c) for _, h, c, _, _ in values]),
            "mean_abs_error": round(sum(abs_errors) / len(abs_errors), 2) if abs_errors else None,
        })

    by_variable = pd.DataFrame(by_variable)

    sens = ratio(total["TP"], total["TP"] + total["FN"])
    ppv = ratio(total["TP"], total["TP"] + total["FP"])
    with_findings = by_variable[by_variable["human_findings"] > 0]

    summary = pd.DataFrame([
        ("reports compared", len(set(key_h) & set(nlp_rows))),
        ("variables compared", len(pairs)),
        ("report x variable values", total["n"]),
        ("accuracy (all values)", ratio(total["agree"], total["n"])),
        ("accuracy 95% CI", wilson(total["agree"], total["n"])),
        ("human findings (not NR)", total["TP"] + total["FN"]),
        ("NLP findings (not NR)", total["TP"] + total["FP"]),
        ("sensitivity / recall (micro)", sens),
        ("sensitivity 95% CI", wilson(total["TP"], total["TP"] + total["FN"])),
        ("PPV / precision (micro)", ppv),
        ("PPV 95% CI", wilson(total["TP"], total["TP"] + total["FP"])),
        ("F1 (micro)", ratio(2 * total["TP"], 2 * total["TP"] + total["FN"] + total["FP"])),
        ("specificity (micro)", ratio(total["TN"], total["TN"] + total["FP_on_NR"])),
        ("F1 (macro, variables with a human finding)", round(with_findings["F1"].dropna().mean(), 3) if len(with_findings) else None),
        ("kappa (mean over variables)", round(by_variable["kappa"].dropna().mean(), 3) if by_variable["kappa"].notna().any() else None),
        ("variables in the sheet but not made by the NLP", ", ".join(not_in_nlp) or "-"),
        ("variables made by the NLP but not in the sheet", ", ".join(not_in_sheet) or "-"),
    ], columns=["measure", "value"])

    confusion = pd.DataFrame(
        [(v, h, c, n) for (v, h, c), n in sorted(confusion.items())],
        columns=["variable", "human", "nlp", "count"],
    )
    mismatches = pd.DataFrame(mismatches)

    by_variable.to_csv(f"{prefix}_by_variable.csv", index=False, encoding="utf-8-sig")
    summary.to_csv(f"{prefix}_summary.csv", index=False, encoding="utf-8-sig")
    mismatches.to_csv(f"{prefix}_mismatches.csv", index=False, encoding="utf-8-sig")
    confusion.to_csv(f"{prefix}_confusion.csv", index=False, encoding="utf-8-sig")

    print(f"Validation files: {prefix}_by_variable.csv, {prefix}_summary.csv, "
          f"{prefix}_mismatches.csv, {prefix}_confusion.csv\n")
    print(summary.to_string(index=False))

    return {"summary": summary, "by_variable": by_variable,
            "mismatches": mismatches, "confusion": confusion}
```
